```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')
nltk.download('punkt_tab')

raw_text = "BinX Tech is an amazing internship program in Palestine! I learned so much about AI and Machine Learning."

tokens = word_tokenize(raw_text)

print(f"Original text : {raw_text}")
print(f"Number of tokens: {len(tokens)}")
print(f"Tokens: {tokens}")
```

```python
import re

text_lower = raw_text.lower()
text_clean = re.sub(r'[^a-z\s]', '', text_lower)

print(f"Original  : {raw_text}")
print(f"Lowercase : {text_lower}")
print(f"Cleaned   : {text_clean}")
```

```python
nltk.download('stopwords')
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
tokens_clean = text_clean.split()
tokens_no_stop = [word for word in tokens_clean if word not in stop_words]

print(f"Before: {tokens_clean}")
print(f"After : {tokens_no_stop}")
```

```python
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
tokens_lemmatized = [lemmatizer.lemmatize(word) for word in tokens_no_stop]

print(f"Before: {tokens_no_stop}")
print(f"After : {tokens_lemmatized}")
```

```python
print("=" * 60)
print("BEFORE vs AFTER — Full Cleaning Pipeline")
print("=" * 60)
print(f"Original text : {raw_text}")
print(f"Final tokens  : {tokens_lemmatized}")
print("=" * 60)
print(f"Tokens before cleaning : {len(word_tokenize(raw_text))}")
print(f"Tokens after cleaning  : {len(tokens_lemmatized)}")
print(f"Tokens removed         : {len(word_tokenize(raw_text)) - len(tokens_lemmatized)}")
```
