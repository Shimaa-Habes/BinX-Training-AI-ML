```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize

sentence = "The movie was great! I really enjoyed watching it."
tokens = word_tokenize(sentence)
print(tokens)
```

```python
print(f"Original sentence: {sentence}")
print(f"Number of tokens: {len(tokens)}")
print(f"Tokens: {tokens}")
```

<strong>Removing punctuation</strong> → strips non-word noise → <code>"great!"</code> → <code>"great"</code>

<br>

<strong>Stop-word removal</strong> → drops common low-signal words → <code>"the", "is", "a"</code> → removed

<br>

<strong>Lemmatization</strong> → reduces words to base form → <code>"running", "ran"</code> → <code>"run"</code>

<br><br>

Lemmatization is preferred over stemming because it produces real words and preserves meaning.

```python
import re

text = "The movie was Great! I really Enjoyed watching it. 10/10 would recommend!"

# lowercase
text_lower = text.lower()

# remove punctuation and numbers
text_clean = re.sub(r'[^a-z\s]', '', text_lower)

print(f"Original : {text}")
print(f"Lowercase: {text_lower}")
print(f"Cleaned  : {text_clean}")
```

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
tokens = text_clean.split()

tokens_no_stop = [word for word in tokens if word not in stop_words]

print(f"Before stop-word removal: {tokens}")
print(f"After stop-word removal : {tokens_no_stop}")
```

```python
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
tokens_lemmatized = [lemmatizer.lemmatize(word) for word in tokens_no_stop]

print(f"Before lemmatization: {tokens_no_stop}")
print(f"After lemmatization : {tokens_lemmatized}")
```

```python
print("=" * 55)
print("BEFORE vs AFTER CLEANING")
print("=" * 55)
print(f"Original  : {text}")
print(f"Final     : {' '.join(tokens_lemmatized)}")
print("=" * 55)
print(f"Tokens before: {len(text.split())}")
print(f"Tokens after : {len(tokens_lemmatized)}")
print(f"Reduction    : {len(text.split()) - len(tokens_lemmatized)} tokens")
```
