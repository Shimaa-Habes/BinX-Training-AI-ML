</div>

<div style="

TF-IDF represents text using numerical values based on <b>word importance and frequency</b>.

<div style="

Word embeddings represent words as <b>dense numerical vectors</b> that capture semantic relationships.

<br><br>

Words with similar meanings tend to have similar positions in the vector space.

<br><br>

<b>Best idea:</b> word meaning & relationships

<div style="

">
What is Bag-of-Words?
</div>

<div style="
font-size:16px;
line-height:1.8;
color:#34443B;
">
<b>Bag-of-Words (BoW)</b> is one of the simplest ways to represent text numerically.
It represents a document based on <b>which words appear</b> and <b>how often each word appears</b>.
</div>

<div style="
margin:18px 0;
padding:16px 20px;
background:#F7F4ED;
border-radius:12px;
border:1px solid #E2D8C7;
font-size:16px;
line-height:1.8;
color:#5A4B38;
">

<b>Example:</b><br>

"The movie was good and the movie was exciting."

<br><br>

Bag-of-Words focuses on the word counts:

<br>

<b>movie → 2</b>&nbsp;&nbsp;&nbsp;
<b>was → 2</b>&nbsp;&nbsp;&nbsp;
<b>good → 1</b>&nbsp;&nbsp;&nbsp;
<b>exciting → 1</b>

</div>

<div style="
font-size:15px;
line-height:1.7;
color:#5B685F;
">
The important point is that <b>word order is ignored</b>.
The model knows which words are present and how frequently they occur, but it does not understand the sequence in which those words appear.
</div>

</div>

<div style="
width:100%;
box-sizing:border-box;
background:#FFFFFF;

">

<div style="

</div>

<div style="
font-size:16px;

color:#34443B;

<b>TF-IDF</b> gives each word a score based on two main ideas:

01 · Frequent in the document

<div style="

">
Although TF-IDF is useful for measuring word importance, it treats each word
<b>independently</b>. It does not understand the meaning of a word or the

</div>

<div style="

TF-IDF knows that a word appears, but it does not understand its
<b>semantic meaning</b> or relationship to other words.

</div>

<b style="color:#203B2F;">Key takeaway:</b>

<b>meaning or context</b> — which is where word embeddings become useful.
</div>

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
cleaned_texts = [
    "movie great really enjoyed watching",
    "movie boring waste time",
    "excellent movie amazing acting",
    "terrible movie bad acting",
    "great story excellent performance"
]

print("Number of documents:", len(cleaned_texts))
print("First document:", cleaned_texts[0])
```

```python
vectorizer = TfidfVectorizer(max_features=5000)

X_tfidf = vectorizer.fit_transform(cleaned_texts)

print("TF-IDF Matrix Shape:", X_tfidf.shape)
```

```python
feature_names = vectorizer.get_feature_names_out()

tfidf_scores = X_tfidf.mean(axis=0).A1

top_indices = tfidf_scores.argsort()[-20:][::-1]

print("Top 20 TF-IDF Features:\n")

for index in top_indices:
    print(f"{feature_names[index]:<20} {tfidf_scores[index]:.4f}")
```

```python
from gensim.downloader import load

word_vectors = load("glove-wiki-gigaword-50")

print("Embedding model loaded successfully.")
print("Vector size:", word_vectors.vector_size)
```

```python
words = ["king", "queen", "computer"]

for word in words:
    print(f"\nNearest neighbors for '{word}':")
    
    for neighbor, similarity in word_vectors.most_similar(word, topn=5):
        print(f"{neighbor:<15} {similarity:.4f}")
```
