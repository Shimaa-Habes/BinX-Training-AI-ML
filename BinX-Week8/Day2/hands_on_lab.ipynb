2.1 Load Pre-trained Word Embeddings

<div style="

```python
from gensim.downloader import load

word_vectors = load("glove-wiki-gigaword-50")

print("Embedding model loaded successfully.")
print("Vector size:", word_vectors.vector_size)
```

```python
words = ["king", "queen", "computer"]

for word in words:
    vector = word_vectors[word]
    
    print(f"\nWord: {word}")
    print("Vector shape:", vector.shape)
    print("First 10 values:", vector[:10])
```

2.3 Find Nearest Neighbors
</div>

<div style="

">
We use <strong style="color:#355C4A;">cosine similarity</strong> to find words whose vectors are closest to a target word.
</div>

<div style="

line-height:1.75;
color:#59675E;

">
If two words have similar vectors, they are likely to have similar meanings or appear in similar contexts.
</div>

<div style="

border-radius:10px;
border:1px solid #DDD4C3;
font-size:14px;
color:#5E665F;
">
<strong style="color:#355C4A;">Key idea:</strong>
A smaller distance between vectors means greater similarity in the embedding space.
</div>

<div style="
width:52px;
height:4px;
background:#B88746;
border-radius:10px;
margin-top:15px;
"></div>

</div>

```python
words = ["king", "queen", "computer"]

for word in words:
    print(f"\nNearest neighbors for '{word}':")
    
    for neighbor, similarity in word_vectors.most_similar(word, topn=5):
        print(f"{neighbor:<15} {similarity:.4f}")
```
