```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
df = pd.read_csv("Student_Performance.csv")

plt.figure(figsize=(10, 5))

sns.scatterplot(
    data=df,
    x="Hours Studied",
    y="Performance Index",
    color="#42A5F5",
    s=80,
    alpha=0.6
)

plt.title("Relationship Between Hours Studied and Performance", fontsize=14)
plt.xlabel("Hours Studied")
plt.ylabel("Performance Index")

plt.grid(alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Extracurricular Activities",
    y="Performance Index",
    color="#F4C95D"
)

plt.title(
    "Performance Distribution by Extracurricular Activities",
    fontsize=14
)

plt.xlabel("Extracurricular Activities")
plt.ylabel("Performance Index")

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.show()
```

<h3 style="color:#243B53; margin:0 0 12px 0; font-size:24px; position:relative;">
Step 2: Correlation Analysis
</h3>

<p style="color:#3D4A57; font-size:15px; line-height:1.8; margin:0;">
Correlation analysis helps us understand how <b>numerical variables move together</b>.
In this step, we calculate the correlation matrix and visualize the relationships using a heatmap.
</p>

<div style="margin-top:16px; padding:12px 15px; background:#E7EEF2; border-left:5px solid #F4C95D; border-radius:6px; color:#243B53; font-size:13px; line-height:1.7;">
📈 <b>Positive</b> relationships &nbsp; • &nbsp;
📉 <b>Negative</b> relationships &nbsp; • &nbsp;
🔍 <b>Useful feature relationships</b>
</div>

<p style="color:#3D4A57; font-size:14px; line-height:1.7; margin:16px 0 0 0;">
The correlation results can help guide future Machine Learning decisions by showing which numerical features are strongly related.
</p>

<div style="margin-top:14px; padding:9px 13px; background:#FFF3CD; border-left:5px solid #D89B2B; border-radius:6px; color:#5D4E37; font-size:13px; font-weight:bold;">
⚠️ CORRELATION ≠ CAUSATION
</div>

</div>

```python
corr = df.corr(numeric_only=True)

display(corr.round(2))
```

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(9, 6))

sns.heatmap(
    corr,
    annot=True,
    cmap="YlGnBu",
    fmt=".2f",
    linewidths=0.8,
    linecolor="#F8F5E9",
    square=True,
    annot_kws={"size": 10, "weight": "bold"}
)

plt.title(
    "Correlation Between Student Performance Variables",
    fontsize=16,
    fontweight="bold",
    color="#243B53",
    pad=15
)

plt.xticks(
    rotation=35,
    ha="right",
    fontsize=10,
    color="#243B53"
)

plt.yticks(
    rotation=0,
    fontsize=10,
    color="#243B53"
)

plt.tight_layout()
plt.show()
```
