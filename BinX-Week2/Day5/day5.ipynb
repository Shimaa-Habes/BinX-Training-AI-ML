## <span style="color:#F78BA0">5.1 Bivariate Analysis</span>

<blockquote style="border-left: 3px solid #FD1D1D; padding-left: 12px; margin-left: 0;">

Bivariate Analysis focuses on studying the relationship between two variables.

Unlike Univariate Analysis, which examines one variable at a time, Bivariate Analysis helps us discover patterns, trends, and relationships between features.

In Machine Learning, understanding relationships between variables is important because models learn from these patterns.

Common visualization techniques include:

- 📈 Scatter Plot
- 📦 Grouped Box Plot

</blockquote>

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("data.csv")

df.head()
```

```python
plt.figure(figsize=(10,5))

sns.scatterplot(
    data=df,
    x="year",
    y="value",
    color="#42A5F5",
    s=80,
    alpha=0.6
)

plt.title("Relationship Between Year and Value", fontsize=14)

plt.xlabel("Year")
plt.ylabel("Value")

plt.grid(alpha=0.3)

plt.tight_layout()

plt.show()
```

## <span style="color:#66BB6A">✅ Scatter Plot Observation</span>

<blockquote style="border-left: 3px solid #4CAF50; padding-left: 12px; margin-left: 0;">

The scatter plot shows the relationship between **Year** and **Value**.

Each point represents one observation from the dataset.

This visualization helps us identify patterns, trends, and unusual values over time.

By observing the distribution of points, we can get an initial understanding of how values change across different years.

</blockquote>

## <span style="color:#F78BA0">📦 Grouped Box Plot</span>

<blockquote style="border-left: 3px solid #FD1D1D; padding-left: 12px; margin-left: 0;">

A grouped box plot compares the distribution of a numeric variable across different categories.

It helps us understand differences between groups by showing:

- 📍 Median values
- 📍 Data spread
- 📍 Possible outliers

This visualization is useful for discovering patterns and differences between categories before building Machine Learning models.

</blockquote>

```python
top_countries = df["country_name"].value_counts().head(10).index

plt.figure(figsize=(12,6))

sns.boxplot(
    data=df[df["country_name"].isin(top_countries)],
    x="country_name",
    y="value",
    color="#81C784"
)

plt.title("Value Distribution by Top 10 Countries", fontsize=14)

plt.xlabel("Country")
plt.ylabel("Value")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
```

## <span style="color:#66BB6A">✅ Box Plot Observation</span>

<blockquote style="border-left: 3px solid #4CAF50; padding-left: 12px; margin-left: 0;">

The box plot compares the distribution of **Value** across different countries.

It helps us understand:

- 📍 The median value for each group
- 📍 The spread and variability of the data
- 📍 Potential outliers that may require further investigation

Different countries show different value distributions, which can reveal important patterns before applying Machine Learning models.

</blockquote>

## <span style="color:#F78BA0">5.2 Correlation</span>

<blockquote style="border-left: 3px solid #FD1D1D; padding-left: 12px; margin-left: 0;">

Correlation measures the strength and direction of the relationship between two numeric variables.

The correlation value ranges from **-1 to +1**:

- 🔵 **+1** → Perfect positive relationship
- ⚪ **0** → No linear relationship
- 🔴 **-1** → Perfect negative relationship

A correlation matrix shows the correlation between all numerical variables in the dataset.

</blockquote>

```python
corr = df.corr(numeric_only=True)

corr
```

```python
plt.figure(figsize=(8,5))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap", fontsize=14)

plt.tight_layout()

plt.show()
```

## <span style="color:#66BB6A">✅ Correlation Heatmap Observation</span>

<blockquote style="border-left: 3px solid #4CAF50; padding-left: 12px; margin-left: 0;">

The correlation heatmap provides a visual summary of relationships between numerical variables.

It helps identify:

- 📈 Strong positive relationships
- 📉 Negative relationships
- 🔍 Variables that may be useful for Machine Learning models

However, correlation does not mean causation. Two variables can move together without one causing the other.

</blockquote>

## <span style="color:#F78BA0">5.3 The Pairplot</span>

<blockquote style="border-left: 3px solid #FD1D1D; padding-left: 12px; margin-left: 0;">

A pairplot creates multiple plots between numerical variables in one figure.

It helps us quickly explore:

- Relationships between features
- Distribution of each variable
- Possible patterns and clusters

Pairplots are useful during early Exploratory Data Analysis to decide which relationships need deeper investigation.

</blockquote>

```python
sns.pairplot(
    df.select_dtypes(include="number")
)

plt.show()
```

## <span style="color:#66BB6A">✅ Pairplot Observation</span>

<blockquote style="border-left: 3px solid #4CAF50; padding-left: 12px; margin-left: 0;">

The pairplot provides an overview of relationships between all numerical variables.

It helps discover:

- Possible correlations
- Data patterns
- Unusual observations
- Variable distributions

This makes it easier to decide which features may be important for future Machine Learning models.

</blockquote>

## <span style="color:#42A5F5">🔗 Step 2: Correlation Analysis</span>

<blockquote style="border-left: 3px solid #42A5F5; padding-left: 12px; margin-left: 0;">

Correlation analysis measures how numerical variables move together.

In this step, we calculate the correlation matrix and visualize it using a heatmap.

The heatmap helps identify:

- 📈 Strong positive relationships
- 📉 Negative relationships
- 🔍 Variables that may be important for future Machine Learning models

Remember: correlation does not mean causation.

</blockquote>

```python
corr = df.corr(numeric_only=True)

corr
```

```python
plt.figure(figsize=(8,5))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap", fontsize=14)

plt.tight_layout()

plt.show()
```
