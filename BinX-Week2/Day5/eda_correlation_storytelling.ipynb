```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="Extracurricular Activities",
    y="Performance Index",
    color="#F4C95D"
)

plt.title(
    "Performance Distribution by Extracurricular Activities",
    fontsize=14
)

plt.xlabel("Extracurricular Activities")
plt.ylabel("Performance Index")

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
corr = df.corr(numeric_only=True)

display(corr.round(2))
```

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(9, 6))

sns.heatmap(
    corr,
    annot=True,
    cmap="YlGnBu",
    fmt=".2f",
    linewidths=0.8,
    linecolor="#F8F5E9",
    square=True,
    annot_kws={"size": 10, "weight": "bold"}
)

plt.title(
    "Correlation Between Student Performance Variables",
    fontsize=16,
    fontweight="bold",
    color="#243B53",
    pad=15
)

plt.xticks(
    rotation=35,
    ha="right",
    fontsize=10,
    color="#243B53"
)

plt.yticks(
    rotation=0,
    fontsize=10,
    color="#243B53"
)

plt.tight_layout()
plt.show()
```
