<h3 style="color:#22D3EE; margin:0 0 12px 0;">
🧮 3.3 Matrices
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 10px 0;">
A <b style="color:#67E8F9;">matrix</b> is a two-dimensional grid of numbers.
</p>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 8px 0;">
In Machine Learning, a matrix usually represents a
<b style="color:#F59E0B;">complete dataset</b>, where:
</p>

<ul style="color:#CBD5E1; line-height:1.9; margin-top:0;">
<li>📌 <b style="color:#22D3EE;">Rows</b> represent data samples.</li>
<li>📊 <b style="color:#22D3EE;">Columns</b> represent features.</li>

<div style="background:#111827; border-left:4px solid #F59E0B; padding:12px 16px; border-radius:8px; margin-top:14px;">

<p style="color:#E2E8F0; margin:0; line-height:1.6;">
The shape of a matrix is written as:

<b style="color:#FBBF24; font-size:16px;">

</p>

</div>

</div>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 8px 0;">

<h3 style="color:#22D3EE; margin:0 0 12px 0;">
🔢 3.4 The Dot Product
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 10px 0;">
The <b style="color:#67E8F9;">dot product</b> is an operation that multiplies
two vectors element-by-element and then adds the results together.
</p>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 12px 0;">
In Machine Learning, the dot product is used to
<b style="color:#F59E0B;">calculate predictions</b>.
</p>

<div style="background:#111827; border-left:4px solid #10B981; padding:14px 18px; border-radius:9px;">

<p style="color:#94A3B8; margin:0 0 6px 0; font-size:14px;">
A linear model prediction:
</p>

<p style="color:#F8FAFC; font-size:18px; margin:0; font-weight:bold;">
Prediction = <span style="color:#22D3EE;">Features · Weights</span>
+ <span style="color:#FBBF24;">Bias</span>
</p>

</div>

</div>

<div style="background:#0F172A; border:1px solid #06B6D4; border-radius:14px; padding:18px 22px; box-shadow:0 0 12px rgba(6,182,212,0.12);">

<h3 style="color:#FBBF24; margin:0 0 8px 0;">
🎯 Goal
</h3>

<p style="color:#CBD5E1; margin:0; line-height:1.7;">
In this section, we will learn how the
<b style="color:#22D3EE;">dot product</b> combines
<b style="color:#67E8F9;">features and weights</b> to produce a
<b style="color:#F59E0B;">prediction</b>.
</p>

</div>

<div style="background:linear-gradient(135deg,#0F172A,#0C4A6E); padding:20px 24px; border-radius:14px; border-left:5px solid #06B6D4; box-shadow:0 0 14px rgba(6,182,212,0.12);">

<h3 style="color:#22D3EE; margin:0 0 12px 0;">
💻 Example
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 8px 0;">
Let's calculate a <b style="color:#F59E0B;">prediction</b> using a feature vector
and a weight vector.
</p>

<p style="color:#CBD5E1; line-height:1.7; margin:0;">
<b style="color:#67E8F9;">Features</b> represent the input data, while
<b style="color:#FBBF24;">weights</b> represent the importance of each feature.
</p>

</div>

```python
import numpy as np

features = np.array([25, 50000, 3])

weights = np.array([0.1, 0.0002, 1.5])

prediction = np.dot(features, weights)

print("\033[94mFeatures:\033[0m", features)
print("\033[92mWeights:\033[0m", weights)

print("\n\033[93mDot Product Prediction:\033[0m", prediction)
```

<div style="background:linear-gradient(135deg,#0F172A,#0C4A6E); padding:20px 24px; border-radius:14px; border-left:5px solid #10B981; box-shadow:0 0 14px rgba(16,185,129,0.12);">

<h3 style="color:#34D399; margin:0 0 12px 0;">
✅ Output Explanation
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 10px 0;">
The dot product multiplies each feature by its corresponding weight and then
<b style="color:#22D3EE;">sums the results</b>:
</p>

<div style="background:#111827; border-left:4px solid #F59E0B; padding:14px 18px; border-radius:9px; margin:12px 0;">

<p style="color:#E2E8F0; margin:0; line-height:1.8; font-size:16px;">
(25 × 0.1) + (50000 × 0.0002) + (3 × 1.5)
<br>
= 2.5 + 10 + 4.5
<br>
= <b style="color:#FBBF24; font-size:18px;">17.0</b>
</p>

</div>

<p style="color:#CBD5E1; line-height:1.7; margin:0;">
The final value represents the
<b style="color:#34D399;">model prediction</b> before adding the bias.
</p>

</div>

<div style="background:#0F172A; border:1px solid #06B6D4; border-radius:14px; padding:18px 22px; box-shadow:0 0 12px rgba(6,182,212,0.12);">

<h3 style="color:#FBBF24; margin:0 0 10px 0;">
💡 Note
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 10px 0;">
The <b style="color:#22D3EE;">dot product</b> is one of the most important
operations in Machine Learning.
</p>

<p style="color:#CBD5E1; line-height:1.7; margin:0;">
Models such as <b style="color:#67E8F9;">Linear Regression</b> and

to generate <b style="color:#F59E0B;">predictions</b>.

</div>

<div style="background:linear-gradient(135deg,#0F172A,#0C4A6E); padding:20px 24px; border-radius:14px; border-left:5px solid #06B6D4; box-shadow:0 0 14px rgba(6,182,212,0.12);">

<h3 style="color:#22D3EE; margin:0 0 12px 0;">
🧮 3.5 Matrix Multiplication
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 10px 0;">
Matrix multiplication applies the <b style="color:#67E8F9;">dot product</b>
across multiple samples at once.
</p>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 12px 0;">
In Machine Learning, this allows a model to make
<b style="color:#F59E0B;">predictions for an entire dataset</b>
using a single operation.
</p>

<div style="background:#111827; border-left:4px solid #10B981; padding:14px 18px; border-radius:9px; margin-top:12px;">

<p style="color:#94A3B8; margin:0 0 6px 0; font-size:14px;">
📐 Shape Rule
</p>

<p style="color:#F8FAFC; font-size:18px; margin:0; font-weight:bold;">
(m × n) @ (n × p) = <span style="color:#34D399;">(m × p)</span>
</p>

<p style="color:#CBD5E1; margin:10px 0 0 0; line-height:1.6;">
The <b style="color:#FBBF24;">inner dimensions must match</b>
for matrix multiplication to be valid.
</p>

</div>

</div>

<div style="background:#0F172A; border:1px solid #06B6D4; border-radius:14px; padding:18px 22px; box-shadow:0 0 12px rgba(6,182,212,0.12);">

<h3 style="color:#FBBF24; margin:0 0 8px 0;">
🎯 Goal
</h3>

<p style="color:#CBD5E1; margin:0; line-height:1.7;">
In this section, we will learn how
<b style="color:#22D3EE;">matrix multiplication</b> produces
<b style="color:#67E8F9;">predictions for multiple samples</b>
and understand the importance of
<b style="color:#F59E0B;">matrix shapes</b>.
</p>

</div>

<div style="background:linear-gradient(135deg,#0F172A,#0C4A6E); padding:20px 24px; border-radius:14px; border-left:5px solid #06B6D4; box-shadow:0 0 14px rgba(6,182,212,0.12);">

<h3 style="color:#22D3EE; margin:0 0 12px 0;">
💻 Example
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 8px 0;">
Let's use <b style="color:#67E8F9;">matrix multiplication</b> to calculate
<b style="color:#F59E0B;">predictions for multiple customers at once</b>.
</p>

<p style="color:#CBD5E1; line-height:1.7; margin:0;">
Each <b style="color:#22D3EE;">row</b> represents one customer, while the
<b style="color:#FBBF24;">weight vector</b> represents the model parameters.
</p>

</div>

```python
import numpy as np

X = np.array([
    [25, 50000, 3],
    [40, 80000, 10]
])

weights = np.array([0.1, 0.0002, 1.5])

predictions = X @ weights

print("\033[94mInput Matrix X:\033[0m")
print(X)

print("\n\033[92mWeights Vector:\033[0m")
print(weights)

print("\n\033[93mPredictions:\033[0m")
print(predictions)

print("\n\033[95mPrediction Shape:\033[0m", predictions.shape)
```

<div style="background:linear-gradient(135deg,#0F172A,#0C4A6E); padding:20px 24px; border-radius:14px; border-left:5px solid #10B981; box-shadow:0 0 14px rgba(16,185,129,0.12);">

<h3 style="color:#34D399; margin:0 0 12px 0;">
✅ Output Explanation
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 10px 0;">
The matrix multiplication applies the <b style="color:#22D3EE;">dot product</b>
between each row in <b style="color:#67E8F9;">X</b> and the
<b style="color:#FBBF24;">weights vector</b>.
</p>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 8px 0;">
For each customer, the model calculates one prediction:
</p>

<ul style="color:#CBD5E1; line-height:1.9; margin-top:0;">
<li>👤 <b style="color:#22D3EE;">Customer 1</b> → Prediction</li>
<li>👤 <b style="color:#22D3EE;">Customer 2</b> → Prediction</li>
</ul>

<div style="background:#111827; border-left:4px solid #F59E0B; padding:12px 16px; border-radius:8px; margin-top:12px;">

<p style="color:#94A3B8; margin:0 0 6px 0;">
📐 Output Shape
</p>

<p style="color:#F8FAFC; margin:0; line-height:1.7;">
<b style="color:#FBBF24; font-size:18px;">(2,)</b>
<br>
This means there are <b style="color:#34D399;">two predictions</b>,
one for each sample.
</p>

</div>

</div>

<div style="background:#0F172A; border:1px solid #06B6D4; border-radius:14px; padding:18px 22px; box-shadow:0 0 12px rgba(6,182,212,0.12);">

<h3 style="color:#FBBF24; margin:0 0 12px 0;">
💡 Shape Rule
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 10px 0;">
For matrix multiplication:
</p>

<div style="background:#111827; border-left:4px solid #22D3EE; padding:12px 16px; border-radius:8px; margin:10px 0;">

<p style="color:#F8FAFC; font-size:18px; margin:0; font-weight:bold;">
(m × n) @ (n × p) = <span style="color:#34D399;">(m × p)</span>
</p>

</div>

<p style="color:#CBD5E1; line-height:1.7; margin:10px 0;">
The <b style="color:#F59E0B;">two middle dimensions must be equal</b>.
</p>

<p style="color:#67E8F9; font-weight:bold; margin:14px 0 6px 0;">
Example:
</p>

<div style="background:#111827; border-left:4px solid #10B981; padding:12px 16px; border-radius:8px;">

<p style="color:#F8FAFC; font-size:17px; margin:0; font-weight:bold;">
(2 × 3) @ (3 × 1) = <span style="color:#34D399;">(2 × 1)</span>
</p>

</div>

<p style="color:#CBD5E1; line-height:1.7; margin:12px 0 0 0;">
The result contains <b style="color:#FBBF24;">one prediction for each sample</b>.

<div style="background:linear-gradient(135deg,#0F172A,#0C4A6E); padding:20px 24px; border-radius:14px; border-left:5px solid #F59E0B; box-shadow:0 0 14px rgba(245,158,11,0.12);">

<h3 style="color:#22D3EE; margin:0 0 10px 0;">
🔹 Step 1: Create a Dataset Matrix
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 8px 0;">
Create a matrix representing <b style="color:#FBBF24;">three samples</b>.
</p>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 8px 0;">
Each <b style="color:#67E8F9;">row</b> represents one sample,
and each <b style="color:#67E8F9;">column</b> represents one feature.
</p>

<div style="background:#111827; border-radius:9px; padding:12px 16px; margin-top:12px;">

<p style="color:#94A3B8; margin:0 0 6px 0; font-size:14px;">
📊 Features
</p>

<ul style="color:#CBD5E1; line-height:1.8; margin:0;">
<li>🎂 <b style="color:#22D3EE;">Age</b></li>
<li>💰 <b style="color:#FBBF24;">Income</b></li>
<li>🗓️ <b style="color:#34D399;">Tenure</b></li>
</ul>

</div>

</div>

```python
import numpy as np

X = np.array([
    [25, 50000, 3],
    [40, 80000, 10],
    [33, 62000, 5]
])

print("\033[94mDataset Matrix:\033[0m")
print(X)

print("\n\033[92mMatrix Shape:\033[0m", X.shape)
```

<div style="background:linear-gradient(135deg,#0F172A,#0C4A6E); padding:18px 22px; border-radius:14px; border-left:5px solid #10B981; box-shadow:0 0 14px rgba(16,185,129,0.12);">

<h3 style="color:#34D399; margin:0 0 12px 0;">
✅ Explanation
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 10px 0;">
The matrix shape is <b style="color:#FBBF24; font-size:17px;">(3, 3)</b>:
</p>

<ul style="color:#CBD5E1; line-height:1.9; margin:0;">
<li><b style="color:#22D3EE;">3 rows</b> → 3 samples</li>
<li><b style="color:#67E8F9;">3 columns</b> → 3 features</li>
</ul>

<div style="background:#111827; border-left:4px solid #F59E0B; padding:12px 16px; border-radius:8px; margin-top:12px;">
<p style="color:#E2E8F0; margin:0; line-height:1.7;">
📊 This matrix represents the <b style="color:#FBBF24;">complete dataset</b>.
</p>
</div>

</div>

<div style="background:#0F172A; border-left:5px solid #06B6D4; padding:18px 22px; border-radius:12px; box-shadow:0 0 12px rgba(6,182,212,0.12);">

<h3 style="color:#22D3EE; margin:0 0 10px 0;">
🔹 Step 2: Dot Product Calculation
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 8px 0;">
Calculate the <b style="color:#67E8F9;">dot product</b> between one sample
vector and a weight vector.
</p>

<p style="color:#CBD5E1; line-height:1.7; margin:0;">
Then compare the <b style="color:#FBBF24;">manual calculation</b> with the
<b style="color:#34D399;">NumPy result</b>.
</p>

</div>

```python
sample = X[0]

weights = np.array([0.1, 0.0002, 1.5])


manual_result = (
    sample[0] * weights[0] +
    sample[1] * weights[1] +
    sample[2] * weights[2]
)

numpy_result = np.dot(sample, weights)


print("\033[93mManual Dot Product:\033[0m", manual_result)
print("\033[92mNumPy Dot Product:\033[0m", numpy_result)
```

<div style="background:#0F172A; border-left:5px solid #10B981; padding:18px 22px; border-radius:12px; box-shadow:0 0 12px rgba(16,185,129,0.12);">

<h3 style="color:#34D399; margin:0 0 10px 0;">
✅ Explanation
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 10px 0;">
Both calculations should return the <b style="color:#FBBF24;">same value</b>.
</p>

<p style="color:#CBD5E1; line-height:1.7; margin:0;">
This confirms that NumPy's <b style="color:#22D3EE;">dot()</b> performs:
</p>

<div style="background:#111827; border-left:4px solid #06B6D4; padding:12px 16px; border-radius:8px; margin-top:10px;">
<p style="color:#F8FAFC; margin:0; font-weight:bold;">
Feature values × Weights → Sum → <span style="color:#34D399;">Prediction</span>
</p>
</div>

</div>

<div style="background:linear-gradient(135deg,#0F172A,#0C4A6E); padding:20px 24px; border-radius:14px; border-left:5px solid #06B6D4; box-shadow:0 0 14px rgba(6,182,212,0.12); margin-top:18px;">

<h3 style="color:#22D3EE; margin:0 0 10px 0;">
🔹 Step 3: Matrix Multiplication
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0;">
Use <b style="color:#67E8F9;">matrix multiplication</b> to calculate
<b style="color:#F59E0B;">predictions for all samples at once</b>.
</p>

<p style="color:#CBD5E1; line-height:1.7; margin:8px 0 0 0;">
Instead of calculating one sample at a time, the model can process the
<b style="color:#34D399;">whole dataset using one operation</b>.
</p>

</div>

```python
weights = np.array([0.1, 0.0002, 1.5])

predictions = X @ weights


print("\033[94mDataset Matrix:\033[0m")
print(X)

print("\n\033[93mWeights:\033[0m")
print(weights)

print("\n\033[92mPredictions:\033[0m")
print(predictions)

print("\n\033[95mPrediction Shape:\033[0m", predictions.shape)
```

<div style="background:linear-gradient(135deg,#0F172A,#0C4A6E); padding:20px 24px; border-radius:14px; border-left:5px solid #10B981; box-shadow:0 0 14px rgba(16,185,129,0.12);">

<h3 style="color:#34D399; margin:0 0 12px 0;">
✅ Explanation
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 10px 0;">
The <b style="color:#22D3EE;">@ operator</b> performs matrix multiplication.
</p>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 8px 0;">
Each row in the dataset is multiplied with the
<b style="color:#FBBF24;">weights vector</b> to produce one prediction.
</p>

<ul style="color:#CBD5E1; line-height:1.9; margin-top:8px;">
<li>👤 <b style="color:#22D3EE;">Sample 1</b> → Prediction 1</li>
<li>👤 <b style="color:#22D3EE;">Sample 2</b> → Prediction 2</li>
<li>👤 <b style="color:#22D3EE;">Sample 3</b> → Prediction 3</li>
</ul>

<div style="background:#111827; border-left:4px solid #F59E0B; padding:12px 16px; border-radius:8px; margin-top:12px;">

<p style="color:#E2E8F0; margin:0; line-height:1.7;">
The output shape <b style="color:#FBBF24; font-size:17px;">(3,)</b>
means we have <b style="color:#34D399;">three predictions</b>.
</p>

</div>

</div>

<div style="background:#0F172A; border-left:5px solid #EF4444; padding:18px 22px; border-radius:12px; box-shadow:0 0 12px rgba(239,68,68,0.12); margin-top:18px;">

<h3 style="color:#F87171; margin:0 0 10px 0;">
🔹 Step 4: Shape Mismatch Error
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0;">
Machine Learning errors often happen because
<b style="color:#FBBF24;">matrix shapes do not match</b>.
</p>

<p style="color:#CBD5E1; line-height:1.7; margin:8px 0 0 0;">
Let's create an <b style="color:#F87171;">incorrect weight vector</b>
and observe the resulting error.
</p>

</div>

```python
wrong_weights = np.array([0.1, 0.0002])

wrong_prediction = X @ wrong_weights
```

<div style="background:#0F172A; border-left:5px solid #EF4444; padding:20px 22px; border-radius:12px; box-shadow:0 0 12px rgba(239,68,68,0.12);">

<h3 style="color:#F87171; margin:0 0 12px 0;">
⚠️ Shape Mismatch Error
</h3>

<p style="color:#CBD5E1; line-height:1.7; margin:0 0 12px 0;">
The error happened because the <b style="color:#F87171;">shapes do not match</b>.
The weight vector contains only <b style="color:#F87171;">2 values</b>,
while the input matrix <b style="color:#22D3EE;">X</b> contains
<b style="color:#34D399;">3 features per sample</b>.
</p>

<div style="background:#111827; border-radius:9px; padding:12px 16px; margin:10px 0;">

<p style="color:#CBD5E1; margin:0; line-height:1.8;">
<b style="color:#67E8F9;">X shape:</b> (3, 3)
<br>
<span style="color:#94A3B8;">→ 3 samples</span>
<br>
<span style="color:#94A3B8;">→ 3 features</span>
<br><br>
<b style="color:#FBBF24;">Weights shape:</b> (2,)
<br>
<span style="color:#94A3B8;">→ Only 2 weights</span>
</p>

</div>

<p style="color:#CBD5E1; line-height:1.7; margin:12px 0 10px 0;">
For matrix multiplication, the number of
<b style="color:#67E8F9;">features must equal the number of weights</b>.
Since <b style="color:#F87171;">3 ≠ 2</b>, NumPy raises a
<b style="color:#F87171;">ValueError</b>.
</p>

<div style="background:#111827; border-left:4px solid #10B981; padding:12px 16px; border-radius:8px;">

<p style="color:#94A3B8; margin:0 0 6px 0;">
✔️ Correct:
</p>

<p style="color:#F8FAFC; font-size:18px; font-weight:bold; margin:0;">
(3, 3) @ (3,) → <span style="color:#34D399;">(3,)</span>
</p>

</div>
