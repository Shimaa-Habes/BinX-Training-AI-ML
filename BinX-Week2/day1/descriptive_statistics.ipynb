<th style="border:1px solid #6D5AA8; padding:10px; color:#C4B5FD;">🎯 When to Use It</th>
</tr>

<tr>
<td style="border:1px solid #6D5AA8; padding:10px; color:#E5E7EB;"><b>Mean</b></td>
<td style="border:1px solid #6D5AA8; padding:10px; color:#E5E7EB;">The arithmetic average</td>
<td style="border:1px solid #6D5AA8; padding:10px; color:#E5E7EB;">Symmetric data with no extreme outliers</td>
</tr>

<tr style="background-color:#1D1730;">
<td style="border:1px solid #6D5AA8; padding:10px; color:#E5E7EB;"><b>Median</b></td>
<td style="border:1px solid #6D5AA8; padding:10px; color:#E5E7EB;">The middle value when sorted</td>
<td style="border:1px solid #6D5AA8; padding:10px; color:#E5E7EB;">Skewed data or data with outliers (more robust)</td>
</tr>

<tr>
<td style="border:1px solid #6D5AA8; padding:10px; color:#E5E7EB;"><b>Mode</b></td>
<td style="border:1px solid #6D5AA8; padding:10px; color:#E5E7EB;">The most frequent value</td>
<td style="border:1px solid #6D5AA8; padding:10px; color:#E5E7EB;">Categorical data, or finding the most common value</td>
</tr>

</table>

<div style="background-color:#211B3A; border-left:5px solid #A78BFA; padding:14px 18px; border-radius:10px; margin-top:14px;">

<p style="color:#E5E7EB; margin:0; line-height:1.7;">

💡 <b style="color:#C4B5FD;">Key Insight:</b>
The <b>mean</b> is pulled toward outliers, while the <b>median</b> is not.

</p>

</div>

<div style="background-color:#111827; border-left:6px solid #38BDF8; padding:18px; border-radius:12px;">

<h3 style="color:#38BDF8; margin:0 0 10px 0;">
📊 Mean vs Median
</h3>

<p style="color:#E5E7EB; margin:0; line-height:1.7;">
Here, we use <b style="color:#86EFAC;">NumPy</b> to calculate two measures of central tendency:
<b style="color:#38BDF8;">Mean</b> and <b style="color:#A78BFA;">Median</b>.
</p>

<p style="color:#E5E7EB; line-height:1.7;">
The dataset contains a very large value, <b style="color:#F87171;">100</b>, compared with the other values.
This value is an <b style="color:#FBBF24;">outlier</b> and has a strong effect on the Mean.
</p>

<ul style="color:#E5E7EB; line-height:1.8;">
<li><b style="color:#86EFAC;">Mean:</b> It is affected by extreme values and is pulled toward the outlier.</li>
<li><b style="color:#A78BFA;">Median:</b> It is more resistant to outliers because it represents the middle value after sorting the data.</li>
</ul>

<p style="color:#CBD5E1; margin-bottom:0;">
💡 When a dataset contains outliers, the Median can provide a more representative measure of the typical value.
</p>

</div>

```python
import numpy as np

data = np.array([10, 12, 12, 13, 100])

print("\033[92mMean:\033[0m", np.mean(data))
print("\033[94mMedian:\033[0m", np.median(data))
```

<td style="border:1px solid #6D5AA8; padding:10px; color:#E5E7EB;">The square root of variance — spread in the same units as the data</td>

</tr>

</table>

<div style="background-color:#111827; border-left:6px solid #A78BFA; padding:18px; border-radius:12px;">

<h3 style="color:#A78BFA; margin:0 0 10px 0;">
📏 Measuring Data Spread
</h3>

<p style="color:#E5E7EB; margin:0; line-height:1.7;">
Here, we use <b style="color:#86EFAC;">NumPy</b> to measure how spread out the values are using
<b style="color:#38BDF8;">Standard Deviation</b>, <b style="color:#FBBF24;">Variance</b>, and
<b style="color:#F472B6;">IQR</b>.
</p>

<ul style="color:#E5E7EB; line-height:1.8;">
<li><b style="color:#38BDF8;">Standard Deviation:</b> Measures the typical distance of values from the mean.</li>
<li><b style="color:#FBBF24;">Variance:</b> Measures the average squared distance from the mean.</li>
<li><b style="color:#F472B6;">IQR:</b> Measures the spread of the middle 50% of the data using Q3 − Q1.</li>
</ul>

<p style="color:#E5E7EB; line-height:1.7;">
Because the dataset contains the outlier <b style="color:#F87171;">100</b>,
the Standard Deviation and Variance become much larger.
The <b style="color:#F472B6;">IQR</b> is more resistant to extreme values, making it useful
when the dataset contains outliers.
</p>

<p style="color:#CBD5E1; margin-bottom:0;">
💡 Comparing these measures helps us understand whether the data is tightly grouped or widely spread.
</p>

</div>

```python
data = np.array([10, 12, 12, 13, 100])

print("\033[92mStandard deviation:\033[0m", np.std(data))
print("\033[94mVariance:\033[0m", np.var(data))

q1, q3 = np.percentile(data, [25, 75])
print("\033[93mIQR:\033[0m", q3 - q1)
```

<div style="background-color:#111827; border-left:6px solid #F59E0B; padding:18px; border-radius:12px;">

<h3 style="color:#FBBF24; margin:0 0 10px 0;">
📊 Percentiles & Quartiles
</h3>

<p style="color:#E5E7EB; margin:0; line-height:1.7;">
A <b style="color:#38BDF8;">percentile</b> indicates the value below which a certain percentage of the data falls.
It helps us understand the relative position of a value within a dataset.
</p>

<ul style="color:#E5E7EB; line-height:1.8;">
<li><b style="color:#86EFAC;">25th Percentile (Q1):</b> 25% of the data falls below this value.</li>
<li><b style="color:#A78BFA;">50th Percentile (Q2):</b> 50% of the data falls below this value. This is the Median.</li>
<li><b style="color:#FBBF24;">75th Percentile (Q3):</b> 75% of the data falls below this value.</li>
</ul>

<p style="color:#E5E7EB; line-height:1.7;">
These three values divide the dataset into four parts, or <b style="color:#F472B6;">quarters</b>.
The difference between Q3 and Q1 is called the <b style="color:#38BDF8;">Interquartile Range (IQR)</b>,
which represents the spread of the middle 50% of the data.
</p>

<p style="color:#CBD5E1; margin-bottom:0;">
💡 Quartiles are especially useful for understanding data distribution and identifying potential outliers.
</p>

</div>

<div style="background-color:#111827; border-left:6px solid #F59E0B; padding:18px; border-radius:12px;">

<h3 style="color:#FBBF24; margin:0 0 10px 0;">
🔢 Calculating Quartiles with NumPy
</h3>

<p style="color:#E5E7EB; margin:0; line-height:1.7;">
Here, we use <b style="color:#86EFAC;">np.percentile()</b> to calculate the three main quartiles of the dataset.
</p>

<ul style="color:#E5E7EB; line-height:1.8;">
<li><b style="color:#86EFAC;">Q1:</b> The 25th percentile, representing the point below which 25% of the data falls.</li>
<li><b style="color:#38BDF8;">Q2:</b> The 50th percentile, which is the Median.</li>
<li><b style="color:#FBBF24;">Q3:</b> The 75th percentile, representing the point below which 75% of the data falls.</li>
</ul>

<p style="color:#E5E7EB; line-height:1.7;">
For this dataset, the value <b style="color:#F87171;">100</b> is much larger than the other observations.
The quartiles help us understand where most of the data is concentrated without being strongly affected by this extreme value.
</p>

<p style="color:#CBD5E1; margin-bottom:0;">
💡 Q1, Q2, and Q3 are also the key values used to construct a <b style="color:#F472B6;">box plot</b>.
</p>

</div>

```python
import numpy as np

data = np.array([10, 12, 12, 13, 100])

q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)

print("\033[92mQ1 (25th percentile):\033[0m", q1)
print("\033[94mQ2 / Median (50th percentile):\033[0m", q2)
print("\033[93mQ3 (75th percentile):\033[0m", q3)
```
