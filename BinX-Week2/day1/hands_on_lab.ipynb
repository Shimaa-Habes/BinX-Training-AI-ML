<h3 style="color:#38BDF8; margin:0 0 8px 0;">
🚗 Step 1: Load a Numeric Column from the Dataset
</h3>

<p style="color:#E5E7EB; margin:0; line-height:1.7;">
We begin by selecting a numerical column from the car dataset.
The <b style="color:#FBBF24;">selling_price</b> column is a useful choice because
it contains the selling price of each car and allows us to apply descriptive
statistics to a real-world numerical variable.
</p>

</div>

```python
import pandas as pd


lab_df = pd.read_csv("cardekho.csv")

car_prices = lab_df["selling_price"]


print("\033[96mNumber of car prices:\033[0m", car_prices.shape[0])

car_prices.head()
```

<div style="background-color:#111827; border-left:5px solid #38BDF8; padding:16px; border-radius:10px;">

<h3 style="color:#38BDF8; margin:0 0 8px 0;">
📊 Step 2: Compute Descriptive Statistics
</h3>

<p style="color:#E5E7EB; margin:0; line-height:1.7;">
Next, we calculate several descriptive statistics for the
<b style="color:#FBBF24;">selling_price</b> column.
These measures help us understand the center and spread of car prices,
including the <b style="color:#34D399;">mean</b>, 
<b style="color:#34D399;">median</b>, 
<b style="color:#34D399;">mode</b>, 
<b style="color:#34D399;">standard deviation</b>, and 
<b style="color:#34D399;">IQR</b>.
</p>

</div>

```python
import numpy as np

data = np.array([10, 12, 12, 13, 100])

q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)

print("\033[92mQ1 (25th percentile):\033[0m", q1)
print("\033[94mQ2 / Median (50th percentile):\033[0m", q2)
print("\033[93mQ3 (75th percentile):\033[0m", q3)
```

```python

mean_val = car_prices.mean()
median_val = car_prices.median()
mode_val = car_prices.mode()[0]
std_val = car_prices.std()


q1 = np.percentile(car_prices, 25)
q3 = np.percentile(car_prices, 75)
iqr_val = q3 - q1


print("\033[92mMean:\033[0m", mean_val)
print("\033[94mMedian:\033[0m", median_val)
print("\033[93mMode:\033[0m", mode_val)
print("\033[95mStandard Deviation:\033[0m", std_val)
print("\033[96mIQR:\033[0m", iqr_val)
```

<div style="background-color:#111827; border-left:5px solid #F7DC6F; padding:16px; border-radius:10px;">

<h3 style="color:#F7DC6F; margin:0 0 8px 0;">
🎯 Step 3: Which Measure Better Represents a "Typical" Value?
</h3>

<p style="color:#E5E7EB; margin:0; line-height:1.7;">
<b style="color:#FBBF24;">Justification:</b>
For car selling prices, the <b style="color:#38BDF8;">median</b> is generally a
better representation of a "typical" car price than the mean.
The dataset contains a small number of cars with very high selling prices,
which can act as outliers and pull the mean upward. The median is more
robust to these extreme values and therefore gives a better idea of the
price of a typical car in the dataset.
</p>

</div>

<div style="background-color:#111827; border-left:5px solid #F7DC6F; padding:16px; border-radius:10px;">

<h3 style="color:#F7DC6F; margin:0 0 8px 0;">
📊 Step 4: Summary of Center and Spread
</h3>

<p style="color:#E5E7EB; margin:0; line-height:1.7;">
<b style="color:#FBBF24;">Summary:</b>
The selling prices in this dataset show a wide range of values, with most
cars concentrated at lower and middle price levels and a smaller number of
cars having much higher prices. This difference between typical prices and
extreme prices can cause the mean to be higher than the median.
The standard deviation helps quantify the overall spread of car prices,
while the IQR focuses on the middle 50% of the data and is less affected by
extreme values. Comparing these measures gives us a clearer understanding
of both the typical selling price and the variability of prices across the
dataset.
</p>
