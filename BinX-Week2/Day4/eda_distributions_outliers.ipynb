<p style="color:#D8E2DC; line-height:1.8; margin:0 0 10px 0;">

<h2 style="color:#A8DADC; border-left:5px solid #F4A261; padding:10px 15px; background:#17211B; border-radius:8px; margin-top:30px;">
⚠️ 4.4 Outlier Detection with the IQR Method
</h2>

<div style="background:#26352A; border-left:4px solid #84A98C; padding:16px 18px; margin:15px 0; border-radius:8px;">

<p style="color:#D8E2DC; line-height:1.8; margin:0;">
An <b style="color:#F4A261;">outlier</b> is a value that is significantly
different from the rest of the data.
</p>

<p style="color:#D8E2DC; line-height:1.8; margin:10px 0 8px 0;">
Outliers can represent:
</p>

<ul style="color:#D8E2DC; line-height:2; margin:0;">
<li>✅ A real rare event</li>
<li>❌ A data entry mistake</li>
<li>⚠️ An unusual but important observation</li>
</ul>

<p style="color:#D8E2DC; line-height:1.8; margin:12px 0 0 0;">
Finding outliers helps us understand our data before building
<b style="color:#A8DADC;">Machine Learning models</b>.
</p>

</div>

<h3 style="color:#F4A261; border-bottom:2px solid #84A98C; padding-bottom:7px; margin-top:25px;">
📐 Understanding the IQR Method
</h3>

<div style="background:#17211B; border:1px solid #84A98C; padding:16px 18px; border-radius:10px; margin:15px 0;">

<p style="color:#D8E2DC; line-height:1.8; margin:0 0 10px 0;">
The <b style="color:#A8DADC;">Interquartile Range (IQR)</b> measures the spread
of the middle 50% of the data.
</p>

<ul style="color:#D8E2DC; line-height:2; margin:0 0 12px 0;">
<li><b style="color:#F4A261;">Q1</b> → 25th percentile</li>
<li><b style="color:#F4A261;">Q2</b> → Median (50th percentile)</li>
<li><b style="color:#F4A261;">Q3</b> → 75th percentile</li>
</ul>

<div style="background:#26352A; padding:12px 16px; border-radius:8px; text-align:center; margin:10px 0;">
<p style="color:#A8DADC; font-size:18px; font-weight:bold; margin:0;">
IQR = Q3 - Q1
</p>
</div>

<p style="color:#D8E2DC; line-height:1.8; margin:12px 0 8px 0;">
A value is considered a potential outlier if it falls:
</p>

<ul style="color:#D8E2DC; line-height:2; margin:0;">
<li>⬇️ Below: <b style="color:#F4A261;">Q1 - 1.5 × IQR</b></li>
<li>⬆️ Above: <b style="color:#F4A261;">Q3 + 1.5 × IQR</b></li>
</ul>

</div>

```python
Q1 = df["Number_of_Reviews"].quantile(0.25)
Q3 = df["Number_of_Reviews"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["Number_of_Reviews"] < lower_bound) |
    (df["Number_of_Reviews"] > upper_bound)
]

outliers
```

<div style="background:#26352A; border-left:5px solid #84A98C; padding:18px 20px; border-radius:10px; margin:18px 0;">

<h3 style="color:#A8DADC; margin:0 0 10px 0;">
✅ IQR Calculation Explanation
</h3>

<p style="color:#D8E2DC; line-height:1.8; margin:0;">
The <b style="color:#F4A261;">IQR method</b> helps identify values that fall
far outside the typical range of the data.
</p>

<p style="color:#D8E2DC; line-height:1.8; margin:10px 0 0 0;">
First, we calculate <b style="color:#A8DADC;">Q1</b> and
<b style="color:#A8DADC;">Q3</b> to understand the spread of the middle
50% of the data.
</p>

<p style="color:#D8E2DC; line-height:1.8; margin:10px 0 0 0;">
Then, we calculate the <b style="color:#A8DADC;">lower and upper bounds</b>.
Any value outside these boundaries is marked as a
<b style="color:#F4A261;">potential outlier</b>.
</p>

</div>

```python
print("\033[91mNumber of outliers:\033[0m", len(outliers))
```
