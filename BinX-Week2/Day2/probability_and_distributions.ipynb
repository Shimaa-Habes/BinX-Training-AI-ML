<h3 style="color:#A78BFA; margin:0 0 10px 0;">
🎲 2.1 Probability Basics
</h3>

<p style="color:#E2E8F0; margin:0; line-height:1.8;">
<b style="color:#38BDF8;">Probability</b> quantifies uncertainty — and ML models
fundamentally output probabilities, such as <i>"85% likely to churn"</i>.
A probability is a number from <b style="color:#34D399;">0 (impossible)</b>
to <b style="color:#34D399;">1 (certain)</b>.
When outcomes are equally likely, the probability of an event is calculated
as the number of favorable outcomes divided by the total number of possible
outcomes.
</p>

</div>

<div style="background:#0B1120; border:1px solid #334155; border-top:4px solid #38BDF8; padding:20px; border-radius:10px;">

<h3 style="color:#38BDF8; margin:0 0 14px 0;">
📐 2.2 Core Rules
</h3>

<tr>
<th style="border:1px solid #475569; padding:10px; background:#172033; color:#A78BFA;">

<td style="border:1px solid #475569; padding:10px; color:#67E8F9;">
P(not A) = 1 - P(A)
</td>
<td style="border:1px solid #475569; padding:10px;">
The chance something does NOT happen
</td>
</tr>

<tr>
<td style="border:1px solid #475569; padding:10px;">

</td>
<td style="border:1px solid #475569; padding:10px; color:#67E8F9;">
P(A or B) = P(A) + P(B) - P(A and B)

<tr>

```python

p_A = 1 / 2
p_B = 1 / 2



p_not_A = 1 - p_A

print("\033[92mP(not A):\033[0m", p_not_A)



p_A_or_B = p_A + p_B

print("\033[94mP(A or B):\033[0m", p_A_or_B)



p_A_and_B = p_A * p_B

print("\033[93mP(A and B):\033[0m", p_A_and_B)
```

<h2 style="color:#67E8F9; margin:0 0 12px 0;">
🎯 2.3 Conditional Probability
</h2>

<p style="color:#E2E8F0; line-height:1.7; margin:0;">
Conditional probability, written <b style="color:#FDE68A;">P(A | B)</b>, is the probability of event A
given that event B has already happened. It is an important concept in
predictive modeling because many ML predictions can be understood as:
<b style="color:#67E8F9;">"Given these features, what is the probability of this outcome?"</b>
</p>

</div>

```python

p_A_and_B = 12 / 27
p_B = 20 / 2004



p_A_given_B = p_A_and_B / p_B



print("\033[92mP(King | Face Card):\033[0m", p_A_given_B)
```

<div style="background:linear-gradient(135deg,#111827,#1E293B); padding:20px; border-radius:14px; border-left:5px solid #A78BFA; box-shadow:0 0 12px rgba(167,139,250,0.15);">

<h2 style="color:#C4B5FD; margin:0 0 12px 0;">
🧠 2.4 Bayes' Theorem
</h2>

<p style="color:#E2E8F0; line-height:1.7; margin:0;">
Bayes' theorem allows us to reverse a conditional probability by computing
<b style="color:#FDE68A;">P(A | B)</b> from <b style="color:#67E8F9;">P(B | A)</b>.
It is an important concept in machine learning and forms the mathematical
foundation of the <b style="color:#C4B5FD;">Naive Bayes classifier</b>.
</p>

<p style="color:#CBD5E1; line-height:1.7; margin:12px 0 0 0;">
It also describes how our belief about an event changes when new evidence
becomes available.
</p>

</div>

<div style="background:#0F172A; border:1px solid #A78BFA; padding:18px; border-radius:10px; text-align:center;">

<h3 style="color:#C4B5FD; margin:0 0 12px 0;">
📐 Bayes' Theorem Formula
</h3>

<p style="color:#F8FAFC; font-size:20px; margin:0;">
P(A | B) = (P(B | A) × P(A)) / P(B)
</p>

</div>

<div style="background:#111827; padding:16px; border-radius:10px; border-left:4px solid #22C1C3;">

<p style="color:#E2E8F0; line-height:1.8; margin:0;">
🔹 <b style="color:#FDE68A;">Prior — P(A):</b> What we believed before seeing the evidence.<br>
🔹 <b style="color:#67E8F9;">Likelihood — P(B | A):</b> How strongly the evidence supports the event.<br>
🔹 <b style="color:#C4B5FD;">Posterior — P(A | B):</b> Our updated belief after considering the evidence.
</p>

</div>

```python

p_A = 0.01
p_B_given_A = 0.95
p_B_given_not_A = 0.05



p_not_A = 1 - p_A



p_B = (
    p_B_given_A * p_A
    + p_B_given_not_A * p_not_A
)



p_A_given_B = (
    p_B_given_A * p_A
) / p_B



print("\033[92mPrior P(Disease):\033[0m", p_A)
print("\033[94mLikelihood P(Positive | Disease):\033[0m", p_B_given_A)
print("\033[93mP(Positive Test) Overall:\033[0m", round(p_B, 4))
print("\033[95mPosterior P(Disease | Positive):\033[0m", round(p_A_given_B, 4))
```

<p style="color:#E2E8F0; line-height:1.7; margin:0;">
💡 <b style="color:#F0DA1A;">Important:</b>
The normal distribution is especially important because many statistical
methods and machine learning assumptions rely on data being approximately
normal. Checking the shape of a distribution is therefore an important part
of Exploratory Data Analysis (EDA).
</p>

</div>

</div>

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

normal_data = np.random.normal(loc=0, scale=1, size=1000)

axes[0].hist(
    normal_data,
    bins=30,
    color="#8B5CF6"
)
axes[0].set_title("Normal Distribution")
axes[0].set_xlabel("Value")
axes[0].set_ylabel("Frequency")


binomial_data = np.random.binomial(
    n=10,
    p=0.5,
    size=1000
)

axes[1].hist(
    binomial_data,
    bins=10,
    color="#F97316"
)
axes[1].set_title("Binomial Distribution")
axes[1].set_xlabel("Number of Successes")
axes[1].set_ylabel("Frequency")


uniform_data = np.random.uniform(
    low=0,
    high=1,
    size=1000
)

axes[2].hist(
    uniform_data,
    bins=30,
    color="#FACC15"
)
axes[2].set_title("Uniform Distribution")
axes[2].set_xlabel("Value")
axes[2].set_ylabel("Frequency")


plt.tight_layout()

plt.show()
```
