## <span style="color:#F0DA1A">2.1 Probability Basics</span>

Probability quantifies uncertainty — and ML models fundamentally output
probabilities ("85% likely to churn"). A probability is a number from 0
(impossible) to 1 (certain). The probability of an event is the number of
favorable outcomes divided by the total number of possible outcomes, when
outcomes are equally likely.

## <span style="color:#F0DA1A">2.3 Conditional Probability</span>

Conditional probability, written **P(A | B)**, is the probability of A
given that B has already happened. It is the foundation of nearly all
predictive modeling: "given these customer features, what is the
probability they churn?" is a conditional probability question.

**P(A | B) = P(A and B) / P(B)**

```python
p_A_and_B = 12/27
p_B = 20/2004

p_A_given_B = p_A_and_B / p_B
print("\033[92mP(King | Face card):\033[0m", p_A_given_B)
```

## <span style="color:#F0DA1A">2.4 Bayes' Theorem</span>

Bayes' theorem lets you reverse a conditional probability — computing
P(A | B) from P(B | A). It is the mathematical basis of the Naive Bayes
classifier and of updating beliefs as new evidence arrives:

**P(A | B) = ( P(B | A) × P(A) ) / P(B)**

**Intuition:** it combines what you believed before seeing evidence
(the **prior**, P(A)) with how strongly the evidence supports it
(the **likelihood**, P(B|A)) to produce an updated belief
(the **posterior**, P(A|B)).

```python
# Worked example: medical test scenario
# A = has the disease, B = tests positive

p_A = 0.01          # prior: 1% of population has the disease
p_B_given_A = 0.95  # likelihood: test is 95% accurate for sick people
p_B_given_not_A = 0.05  # false positive rate: 5% of healthy people test positive

# P(B) = P(B|A)*P(A) + P(B|not A)*P(not A)
p_not_A = 1 - p_A
p_B = (p_B_given_A * p_A) + (p_B_given_not_A * p_not_A)

# Bayes' theorem
p_A_given_B = (p_B_given_A * p_A) / p_B

print("\033[92mPrior P(disease):\033[0m", p_A)
print("\033[94mLikelihood P(positive | disease):\033[0m", p_B_given_A)
print("\033[93mP(positive test) overall:\033[0m", round(p_B, 4))
print("\033[95mPosterior P(disease | positive test):\033[0m", round(p_A_given_B, 4))
```

**Note:** the normal distribution matters most — many statistical methods
and ML assumptions rely on data being approximately normal, which is one
of the first things EDA checks for.

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)  # fixed seed for reproducibility

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Normal distribution
normal_data = np.random.normal(loc=0, scale=1, size=1000)
axes[0].hist(normal_data, bins=30, color="teal")
axes[0].set_title("Normal Distribution")

# Binomial distribution
binomial_data = np.random.binomial(n=10, p=0.5, size=1000)
axes[1].hist(binomial_data, bins=10, color="orange")
axes[1].set_title("Binomial Distribution")

# Uniform distribution
uniform_data = np.random.uniform(low=0, high=1, size=1000)
axes[2].hist(uniform_data, bins=30, color="gold")
axes[2].set_title("Uniform Distribution")

plt.tight_layout()
plt.show()
```

## <span style="color:#F0DA1A">🧪 Hands-On Lab: Probability & Distributions in Code</span>

### <span style="color:#F85226">Step 1: Simulate 10,000 coin flips</span>

```python
np.random.seed(42)  # fixed seed for reproducibility

n_flips = 10000
flips = np.random.randint(0, 2, size=n_flips)  # 0 = tails, 1 = heads

proportion_heads = np.mean(flips)
print("\033[92mProportion of heads:\033[0m", proportion_heads)
print("\033[94mExpected (theoretical):\033[0m 0.5")
```

### <span style="color:#F85226">Step 2: Sample from a normal distribution</span>

```python
np.random.seed(42)  # fixed seed for reproducibility

normal_sample = np.random.normal(loc=50, scale=10, size=10000)

plt.hist(normal_sample, bins=40, color="teal")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Sample from a Normal Distribution (mean=50, std=10)")
plt.show()

print("\033[92mSample mean:\033[0m", np.mean(normal_sample))
print("\033[94mSample standard deviation:\033[0m", np.std(normal_sample))
```

### <span style="color:#F85226">Step 3: Compute a conditional probability by hand and verify with simulation</span>

**Scenario:** Rolling two dice. Let A = "the sum is greater than 8",
and B = "the first die shows a 6". What is P(A | B)?

**By hand:** If the first die shows 6, the second die can be 1-6, giving
sums of 7, 8, 9, 10, 11, 12. Out of these 6 outcomes, the sum is greater
than 8 in 4 cases (9, 10, 11, 12). So P(A | B) = 4/6 ≈ 0.667.


```python
np.random.seed(42)  # fixed seed for reproducibility

n_rolls = 100000
die1 = np.random.randint(1, 7, size=n_rolls)
die2 = np.random.randint(1, 7, size=n_rolls)
total = die1 + die2

# B: first die shows a 6
B = (die1 == 6)

# A and B: sum > 8 AND first die is 6
A_and_B = (total > 8) & B

# P(A | B) = P(A and B) / P(B)
p_A_given_B_sim = np.sum(A_and_B) / np.sum(B)

print("\033[92mSimulated P(A | B):\033[0m", round(p_A_given_B_sim, 4))
print("\033[94mBy-hand calculation:\033[0m", round(4/6, 4))
```

### <span style="color:#F85226">Step 4: Documentation of results</span>

**Step 1 — Coin flips:** With 10,000 simulated flips, the proportion of
heads landed very close to 0.5, demonstrating the Law of Large Numbers —
as the number of trials grows, the observed proportion converges to the
true theoretical probability.

**Step 2 — Normal distribution:** Sampling 10,000 values from a normal
distribution with mean 50 and std 10 produced a classic bell-shaped
histogram, with the sample mean and standard deviation closely matching
the parameters used to generate the data.

**Step 3 — Conditional probability:** The simulated probability of
rolling a sum greater than 8 given the first die shows a 6 matched the
hand-calculated value of 4/6 ≈ 0.667 almost exactly, confirming that
simulation is a reliable way to verify probability calculations.
