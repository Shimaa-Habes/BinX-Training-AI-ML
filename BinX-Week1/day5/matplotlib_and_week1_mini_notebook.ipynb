📈 5.2 Matplotlib Basics
</h2>

<p style="color:#D1D1D1; font-size:15px; line-height:1.8; margin-top:15px;">
<b style="color:#E50914;">Matplotlib</b> is a Python library used to create
clear and informative data visualizations.

<p style="color:#B3B3B3; font-size:15px; line-height:1.8; margin-top:10px;">
🎬 In this example, we will create a simple <b style="color:#FFFFFF;">line plot</b>
to visualize how a value changes over time.
</p>

<p style="color:#B3B3B3; font-size:15px; line-height:1.8; margin-top:10px;">
The basic process is to provide the <b style="color:#FFFFFF;">x</b> and
<b style="color:#FFFFFF;">y</b> values, create the plot, add axis labels
and a title, and finally display the chart.
</p>

<div style="background-color:#241818; border-left:4px solid #E50914; padding:14px 17px; border-radius:10px; margin-top:16px;">

<p style="color:#D1D1D1; margin:0; line-height:1.8; font-size:14px;">

💡 <b style="color:#E50914;">Key Idea</b><br>

A line plot connects data points in order, making it useful for showing
<b style="color:#FFFFFF;">changes and trends</b> across an ordered variable,
such as time.

</p>

</div>

</div>

<div style="background-color:#181818; border:1px solid #333333; border-left:5px solid #E50914; padding:16px 20px; border-radius:11px;">

<h3 style="color:#FFFFFF; font-family:Georgia,serif; font-size:21px; margin:0;">
💻 Creating a Simple Line Plot
</h3>

<p style="color:#B3B3B3; font-size:14px; margin:8px 0 0 0; line-height:1.6;">
Create a basic line chart using Matplotlib to visualize changes in the data.
</p>

</div>

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)
plt.xlabel("Time")
plt.ylabel("Value")
plt.title("Value over Time")
plt.show()
```

<h3 style="color:#FFFFFF; font-family:Georgia,serif; font-size:21px; margin:0;">
📉 Histogram
</h3>

<p style="color:#B3B3B3; font-size:14px; line-height:1.6; margin:8px 0 0 0;">
A histogram shows the <b style="color:#FFFFFF;">distribution of a numerical variable</b>
by grouping values into intervals called <b style="color:#E50914;">bins</b>.
It helps us understand the <b style="color:#FFFFFF;">shape, spread, and frequency</b>
of the data.
</p>

</div>

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.randn(1000)

plt.hist(data, bins=30, color="#E50914")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram Example")
plt.show()
```

<div style="background-color:#1F1F1F; border:1px solid #333333; border-left:6px solid #E50914; padding:20px 22px; border-radius:12px; box-shadow:0 4px 14px rgba(0,0,0,0.22);">

<h2 style="color:#FFFFFF; font-family:Georgia,serif; font-size:24px; margin:0; border-bottom:2px solid #E50914; padding-bottom:9px;">
🧩 5.4 Subplots
</h2>

<p style="color:#D1D1D1; font-size:15px; line-height:1.8; margin-top:15px;">
<b style="color:#E50914;">Subplots</b> allow us to display multiple visualizations
within a single figure.
</p>

<p style="color:#B3B3B3; font-size:15px; line-height:1.8; margin-top:9px;">
🎬 Using <code>plt.subplots()</code>, we can create a grid of plotting areas
and control each visualization independently. This is especially useful
when we want to <b style="color:#FFFFFF;">compare different views of the data side by side</b>.
</p>

<ul style="color:#D1D1D1; font-size:14px; line-height:2; padding-left:25px;">

<li>
<b style="color:#E50914;">plt.subplots(1, 2)</b> → creates one row with two plotting areas.
</li>

<li>
<b style="color:#E50914;">axes[0]</b> → refers to the first subplot.
</li>

<li>
<b style="color:#E50914;">axes[1]</b> → refers to the second subplot.
</li>

<li>
<b style="color:#E50914;">plt.tight_layout()</b> → adjusts spacing to prevent overlapping elements.
</li>

</ul>

<div style="background-color:#241818; border-left:4px solid #E50914; padding:15px 17px; border-radius:10px; margin-top:16px;">

<p style="color:#D1D1D1; margin:0; line-height:1.8; font-size:14px;">

💡 <b style="color:#E50914;">Key Idea</b><br>

Subplots are especially useful in <b style="color:#FFFFFF;">EDA</b> because
they let us examine multiple aspects of a dataset in one figure,
making comparisons easier and more efficient.

</p>

</div>

</div>

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].hist(data, bins=30, color="#E50914")
axes[0].set_title("Distribution of Random Data")

axes[1].scatter(x, y, color="#E50914")
axes[1].set_title("Value vs Index")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# 1. NumPy: a simple computation

sample_values = np.array([10, 20, 30, 40, 50])
mean_value = np.mean(sample_values)

print("NumPy computed mean:", mean_value)


# 2. Pandas: load and clean the Netflix dataset

netflix_df = pd.read_csv("netflix_titles.csv")
netflix_df = netflix_df.dropna(subset=["release_year"])


# 3. Matplotlib: visualize the data

plt.hist(netflix_df["release_year"], bins=30, color="#E50914")
plt.xlabel("Release Year")
plt.ylabel("Frequency")
plt.title("Distribution of Netflix Content by Release Year")
plt.show()
```
