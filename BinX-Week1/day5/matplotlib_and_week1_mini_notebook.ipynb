📈 5.2 The Matplotlib Basics

</h2>

<p style="color:#E2E8F0; font-size:16px; line-height:1.8; margin-top:15px;">

<b style="color:#5EEAD4;">Matplotlib</b> is a Python library used to create data visualizations.

In this example, we will create a simple <b style="color:#5EEAD4;">line plot</b> to visualize how a value changes over time.

The basic plotting process involves providing the <b>x</b> and <b>y</b> values, creating the plot, adding axis labels and a title, and finally displaying the chart.

<p style="color:#CCFBF1; margin:0; line-height:1.8;">

<h3 style="color:#5EEAD4; margin:0;">

💻 Creating a Simple Line Plot

</h3>

<p style="color:#CBD5E1; margin:8px 0 0 0;">

Creating a basic line chart using Matplotlib.

</p>

</div>


```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 30, 25]

plt.plot(x, y)
plt.xlabel("Time")
plt.ylabel("Value")
plt.title("Value over Time")
plt.show()
```

📉 Histogram

</h3>

<p style="color:#CBD5E1; margin:8px 0 0 0;">

A histogram shows the distribution of a numerical variable by grouping values into intervals called <b>bins</b>. It helps us understand the shape, spread, and frequency of the data.

</p>

</div>


```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.randn(1000)

plt.hist(data, bins=30, color="#E50914")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram Example")
plt.show()
```

<h2 style="color:#5EEAD4; margin:0; border-bottom:2px solid #0D9488; padding-bottom:8px;">

🧩 5.4 Subplots

</h2>

<p style="color:#E2E8F0; font-size:16px; line-height:1.8; margin-top:15px;">

<b style="color:#5EEAD4;">Subplots</b> allow us to display multiple visualizations within a single figure.

Using <code>plt.subplots()</code>, we can create a grid of plotting areas and work with each plot independently. This is useful when we want to <b>compare different visualizations side by side</b>.

</p>

<ul style="color:#E2E8F0; font-size:15px; line-height:1.9;">

<li><b style="color:#5EEAD4;">plt.subplots(1, 2)</b> creates one row with two plotting areas.</li>

<li><b style="color:#5EEAD4;">axes[0]</b> refers to the first subplot.</li>

<li><b style="color:#5EEAD4;">axes[1]</b> refers to the second subplot.</li>

<li><b style="color:#5EEAD4;">plt.tight_layout()</b> automatically adjusts spacing to prevent overlapping elements.</li>

</ul>

<div style="background-color:#0A2828; border-left:5px solid #14B8A6; padding:15px; border-radius:10px; margin-top:15px;">

<p style="color:#CCFBF1; margin:0; line-height:1.7;">

💡 <b>Key Idea:</b> Subplots are especially useful in EDA because they allow us to examine multiple aspects of a dataset in one figure, making comparisons easier.

</p>

</div>

</div>


```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].hist(data, bins=30, color="#E50914")
axes[0].set_title("Distribution of Random Data")

axes[1].scatter(x, y, color="#E50914")
axes[1].set_title("Value vs Index")

plt.tight_layout()
plt.show()
```

<h3 style="color:#99F6E4; margin-top:20px;">

🔹 Step 1: Load and Inspect the Dataset with Pandas

</h3>

<p style="color:#E2E8F0; font-size:16px; line-height:1.8;">

We begin by loading the <b style="color:#5EEAD4;">Netflix Movies and TV Shows</b> dataset using Pandas.

Before cleaning the data, we will inspect its <b>shape</b> and check the number of <b>missing values</b> in each column. This gives us an initial understanding of the dataset and helps us decide how missing values should be handled.

</p>

<div style="background-color:#102F2F; border-left:5px solid #14B8A6; padding:15px; border-radius:10px;">

<p style="color:#CCFBF1; margin:0; line-height:1.7;">

💡 <b>Why inspect before cleaning?</b><br>

Different columns may contain different amounts of missing data. Instead of removing every row containing a missing value, we first examine the dataset and then apply an appropriate cleaning strategy.

</p>

</div>

</div>


```python
lab_df = pd.read_csv("netflix_titles.csv")

print("Shape before cleaning:", lab_df.shape)

print("\nMissing values per column:\n")
print(lab_df.isnull().sum())
```

<div style="background-color:#102F2F; border-left:6px solid #0D9488; padding:20px; border-radius:12px;">

<h3 style="color:#5EEAD4; margin:0; border-bottom:2px solid #0D9488; padding-bottom:8px;">

🔹 Step 2: Use NumPy to Compute a Derived Feature

</h3>

<p style="color:#E2E8F0; font-size:16px; line-height:1.8; margin-top:15px;">

In this step, we use <b style="color:#5EEAD4;">NumPy</b> to calculate summary statistics for the <b>release year</b> of the Netflix content.

We will calculate the <b>mean</b> and <b>standard deviation</b>, then create a new derived feature called <b>z_score</b>.

The z-score indicates how far each release year is from the average release year in terms of standard deviations.

</p>

<div style="background-color:#0A2828; border-left:5px solid #14B8A6; padding:15px; border-radius:10px; margin-top:15px;">

<p style="color:#CCFBF1; margin:0; line-height:1.7;">

💡 <b>Key Idea:</b> A z-score close to <b>0</b> means the value is close to the mean, while a positive or negative z-score indicates that the value is above or below the mean.

</p>

</div>

</div>


```python
# Compute the mean and standard deviation of release years using NumPy

mean_release_year = np.mean(lab_df["release_year"])
std_release_year = np.std(lab_df["release_year"])

print("Mean release year:", mean_release_year)
print("Standard deviation:", std_release_year)


# Derived feature: z-score of each release year

lab_df["z_score"] = (
    (lab_df["release_year"] - mean_release_year)
    / std_release_year
)

lab_df[["title", "type", "release_year", "z_score"]].head()
```

<div style="background-color:#102F2F; border-left:6px solid #0D9488; padding:20px; border-radius:12px;">

<h3 style="color:#5EEAD4; margin:0; border-bottom:2px solid #0D9488; padding-bottom:8px;">

🔹 Step 3: Produce Three Labeled Plots

</h3>

<p style="color:#E2E8F0; font-size:16px; line-height:1.8; margin-top:15px;">

In this step, we use <b style="color:#5EEAD4;">Matplotlib</b> to create three different visualizations that explore important aspects of the Netflix dataset.

Each plot answers a different type of question:

</p>

<ul style="color:#E2E8F0; font-size:15px; line-height:1.9;">

<li>
📉 <b style="color:#5EEAD4;">Histogram:</b> What is the distribution of Netflix content across release years?
</li>

<li>
🔵 <b style="color:#5EEAD4;">Scatter Plot:</b> How does the number of Netflix titles vary across release years?
</li>

<li>
📊 <b style="color:#5EEAD4;">Bar Plot:</b> Which countries have the largest number of titles in the dataset?
</li>

</ul>

<div style="background-color:#0A2828; border-left:5px solid #14B8A6; padding:15px; border-radius:10px; margin-top:15px;">

<p style="color:#CCFBF1; margin:0; line-height:1.7;">

💡 <b>Goal:</b> Instead of looking at raw values in a table, these visualizations help us identify patterns, concentration, and differences in the Netflix catalog more quickly.

</p>

</div>

</div>


```python
# 1. Histogram: Distribution of Release Years

plt.hist(lab_df["release_year"], bins=30, color="#E50914")

plt.xlabel("Release Year")
plt.ylabel("Frequency")
plt.title("Distribution of Netflix Content by Release Year")

plt.show()


# 2. Scatter Plot: Number of Titles by Release Year

year_counts = lab_df.groupby("release_year").size()

plt.scatter(
    year_counts.index,
    year_counts.values,
    color="#E50914",
    alpha=0.5
)

plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.title("Netflix Titles by Release Year")

plt.show()


# 3. Bar Plot: Top 10 Countries by Number of Titles

top10_countries = (
    lab_df["country"]
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)

plt.bar(
    top10_countries.index,
    top10_countries.values,
    color="#E50914"
)

plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.title("Top 10 Countries by Number of Netflix Titles")

plt.xticks(rotation=45, ha="right")

plt.show()
```

<h3 style="color:#5EEAD4; margin:0; border-bottom:2px solid #0D9488; padding-bottom:8px;">
