2.1 Core Python Data Types

```python
# The core types you will use constantly:
x_int = 42                 # int : Whole number
x_float = 3.14             # float : decimal number
x_str = "churn"            # str : text
x_bool = True              # bool : True / False
x_list = [1, 2, 3]         # list : Ordered, mutable sequence
x_tuple = (1, 2)           # tuple : Ordered, immutable sequence
x_dict = {"age": 30}       # dict : Key-value pairs
x_set = {1, 2, 3}          # set : Unordered unique values

print("\033[96mint:\033[0m", type(x_int))
print("\033[94mfloat:\033[0m", type(x_float))
print("\033[92mstr:\033[0m", type(x_str))
print("\033[93mbool:\033[0m", type(x_bool))
print("\033[95mlist:\033[0m", type(x_list))
print("\033[96mtuple:\033[0m", type(x_tuple))
print("\033[94mdict:\033[0m", type(x_dict))
print("\033[92mset:\033[0m", type(x_set))
```

2.2 Control Flow
</h2>

<p style="color:#CBD5E1; font-size:16px; line-height:1.6; margin:10px 0 0 0;">
Conditionals (<code style="color:#60A5FA;">if/elif/else</code>) branch logic; loops (<code style="color:#60A5FA;">for</code>, <code style="color:#60A5FA;">while</code>) repeat work.
</p>

```python
## if / elif / else and Loop Examples
for value in [1, 2, 3, 4]:
    if value % 2 == 0:
        print(f"\033[92m{value} is even\033[0m")
    else:
        print(f"\033[93m{value} is odd\033[0m")
```

```python
## While Loop
total = 0
i = 0
while i < 5:
    total = total + i
    print("\033[94mRunning total:\033[0m", total)
    i = i + 1
```

2.3 Functions
</h2>

<p style="color:#CBD5E1; font-size:16px; line-height:1.6; margin:10px 0 0 0;">
Functions package reusable logic. A well-named function with clear
parameters and a single responsibility is the foundation of clean,
testable data pipelines.
</p>

</div>

```python
## Function Example
def normalize(value, min_val, max_val):
    """Scale a value to the 0-1 range."""
    return (value - min_val) / (max_val - min_val)

print("\033[92mnormalize(75, 0, 100):\033[0m", normalize(75, 0, 100))
print("\033[94mnormalize(0, 0, 100):\033[0m", normalize(0, 0, 100))
print("\033[93mnormalize(5, 7, 100):\033[0m", normalize(5, 7, 100))
```

2.4 List Comprehensions
</h2>

<p style="color:#CBD5E1; font-size:16px; line-height:1.6; margin:10px 0 0 0;">
A list comprehension builds a new list in one readable line, replacing
a multi-line loop. This is idiomatic ("Pythonic") code and appears
constantly in real data-science work.
</p>

</div>

```python
## List Comprehensions
squares = [x**2 for x in range(10)]
evens = [x for x in range(20) if x % 2 == 0]

print("\033[92mSquares:\033[0m", squares)
print("\033[94mEvens:\033[0m", evens)
```

2.5 Object-Oriented Programming (Basics)
</h2>

<p style="color:#CBD5E1; font-size:16px; line-height:1.6; margin:10px 0 0 0;">
A class bundles data (attributes) and behavior (methods) together. You
will not write many classes early on, but AI/ML frameworks (Scikit-learn
estimators, PyTorch models) are all built on classes, so understanding
the syntax is essential for reading and using them later.
</p>

</div>

```python
## Object-Oriented Programming (Basics)
class DataPoint:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def distance_from_origin(self):
        return (self.x**2 + self.y**2) ** 0.5

p = DataPoint(3, 4)
print("\033[92mDistance from origin:\033[0m", p.distance_from_origin())
```
