2.3 Functions
</h2>
<p style="color:#94A3B8; font-size:16px; line-height:1.6;">
Functions package reusable logic. A well-named function with clear
parameters and a single responsibility is the foundation of clean,
testable data pipelines.
</p>
</div>

```python
## Function Example
def normalize(value, min_val, max_val):
    """Scale a value to the 0-1 range."""
    return (value - min_val) / (max_val - min_val)

print("\033[92mnormalize(75, 0, 100):\033[0m", normalize(75, 0, 100))
print("\033[94mnormalize(0, 0, 100):\033[0m", normalize(0, 0, 100))
print("\033[93mnormalize(5, 7, 100):\033[0m", normalize(5, 7, 100))
```

<h2 style="color:#FFFFFF; font-size:24px; text-shadow:0 0 5px #94A3B8; margin:0;">
2.4 List Comprehensions
</h2>
<p style="color:#94A3B8; font-size:16px; line-height:1.6;">
A list comprehension builds a new list in one readable line, replacing
a multi-line loop. This is idiomatic ("Pythonic") code and appears
constantly in real data-science work.
</p>
</div>

```python
## List Comprehensions
squares = [x**2 for x in range(10)]
evens = [x for x in range(20) if x % 2 == 0]

print("\033[92mSquares:\033[0m", squares)
print("\033[94mEvens:\033[0m", evens)
```

2.5 Object-Oriented Programming (Basics)
</h2>
<p style="color:#94A3B8; font-size:16px; line-height:1.6;">
A class bundles data (attributes) and behavior (methods) together. You
will not write many classes early on, but AI/ML frameworks (Scikit-learn
estimators, PyTorch models) are all built on classes, so understanding
the syntax is essential for reading and using them later.
</p>
</div>

```python
## Object-Oriented Programming (Basics)
class DataPoint:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def distance_from_origin(self):
        return (self.x**2 + self.y**2) ** 0.5

p = DataPoint(3, 4)
print("\033[92mDistance from origin:\033[0m", p.distance_from_origin())
```

<h3 style="color:#FFFFFF; text-shadow:0 0 5px #94A3B8;">Step 1: Write a function that returns mean, min, and max as a dictionary</h3>


```python
def stats(numbers):
    total = 0
    for num in numbers:
        total = total + num
    mean = total / len(numbers)

    minimum = min(numbers)
    maximum = max(numbers)

    return {
        "mean": mean,
        "min": minimum,
        "max": maximum
    }

list_of_numbers = [50, 55, 80, 100, 65]
print("\033[92mStats:\033[0m", stats(list_of_numbers))
```

<h3 style="color:#FFFFFF; text-shadow:0 0 5px #94A3B8;">Step 3: Define a small class with at least two attributes and one method</h3>

```python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade

    def is_pass(self):
        if self.grade >= 60:
            return f"{self.name} passed"
        else:
            return f"{self.name} failed"

student_1 = Student("Shimaa", 90)
student_2 = Student("Sara", 50)

print("\033[92m" + student_1.is_pass() + "\033[0m")
print("\033[91m" + student_2.is_pass() + "\033[0m")
```
