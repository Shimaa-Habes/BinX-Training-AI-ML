## 2.2  Control Flow
Conditionals (if/elif/else) branch logic; loops (for, while) repeat work.

```python
## if / elif / else and Loop Examples
for value in [1, 2, 3, 4]:
    if value % 2 == 0:
        print(f"{value} is even")
    else:
        print(f"{value} is odd")
```

```python
## While Loop 
sum = 0
i = 0
while i<5:
    sum = sum + i
    print(sum) 
    i = i + 1 
```

## 2.3 Functions
Functions package reusable logic. A well-named function with clear parameters and a single responsibility is the 
foundation of clean, testable data pipelines.

```python
## Function Example
def normalize(value, min_val, max_val):
    """Scale a value to the 0-1 range."""
    return (value - min_val) / (max_val - min_val)
print(normalize(75, 0, 100))  # 0.75
print(normalize(0, 0, 100))   # 0.0
print(normalize(5, 7, 100))   # -0.02 
```

## 2.4  List Comprehensions
A list comprehension builds a new list in one readable line, replacing a multi-line loop. This is idiomatic ("Pythonic") 
code and appears constantly in real data-science work.

```python
##  List Comprehensions 
squares = [x**2 for x in range(10)]
evens   = [x for x in range(20) if x % 2 == 0]
```

## 2.5  Object-Oriented Programming (Basics)
A class bundles data (attributes) and behavior (methods) together. You will not write many classes early on, but 
AI/ML frameworks (Scikit-learn estimators, PyTorch models) are all built on classes, so understanding the syntax is 
essential for reading and using them later.


```python
##  Object-Oriented Programming (Basics)
class DataPoint:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def distance_from_origin(self):
        return (self.x**2 + self.y**2) ** 0.5
p = DataPoint(3, 4)
print(p.distance_from_origin())   # 5.0
```

Step 1: Write a function that takes a list of numbers and returns the mean, min, and max as a dictionary.


```python
def stats(numbers):
    sum = 0
    for num in numbers :
        sum = sum + num
    mean = sum / len(numbers)
        
    minimum = min(numbers)
    maximum = max(numbers)
    
    return{
        "mean" : mean,
        "min"  : minimum,
        "max"  : maximum
    }
    
list_of_numbers = [50, 55, 80, 100, 65]
print(stats(list_of_numbers))
```
