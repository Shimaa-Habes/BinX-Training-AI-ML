<h3 style="color:#FACC15; font-size:21px; text-shadow:0 0 7px rgba(250,204,21,0.3); margin:0; letter-spacing:0.3px;">
Step 1: Write a function that returns mean, min, and max as a dictionary
</h3>

</div>

```python
def stats(numbers):
    total = 0
    for num in numbers:
        total = total + num
    mean = total / len(numbers)

    minimum = min(numbers)
    maximum = max(numbers)

    return {
        "mean": mean,
        "min": minimum,
        "max": maximum
    }

list_of_numbers = [50, 55, 80, 100, 65]
print("\033[92mStats:\033[0m", stats(list_of_numbers))
```

Step 2: Rewrite a for-loop that filters even numbers as a list comprehension
</h3>

</div>

```python
even_numbers = []
for num in range(40):
    if num % 2 == 0:
        even_numbers.append(num)

even_numbers2 = [num for num in range(40) if num % 2 == 0]

print("\033[94mLoop result:\033[0m", even_numbers)
print("\033[92mComprehension result:\033[0m", even_numbers2)
```

Step 3: Define a small class with at least two attributes and one method
</h3>

</div>

```python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade

    def is_pass(self):
        if self.grade >= 60:
            return f"{self.name} passed"
        else:
            return f"{self.name} failed"

student_1 = Student("Shimaa", 90)
student_2 = Student("Sara", 50)

print("\033[92m" + student_1.is_pass() + "\033[0m")
print("\033[91m" + student_2.is_pass() + "\033[0m")
```
