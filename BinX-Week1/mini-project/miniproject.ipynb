```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

```

```python
lab_df = pd.read_csv("netflix_titles.csv")

print("Shape before cleaning:", lab_df.shape)

print("\nMissing values per column:\n")
print(lab_df.isnull().sum())
```

<div style="background-color:#102F2F; border-left:6px solid #0D9488; padding:20px; border-radius:12px;">

<h3 style="color:#5EEAD4; margin:0; border-bottom:2px solid #0D9488; padding-bottom:8px;">

🔹 Step 2: Use NumPy to Compute a Derived Feature

</h3>

<p style="color:#E2E8F0; font-size:16px; line-height:1.8; margin-top:15px;">

In this step, we use <b style="color:#5EEAD4;">NumPy</b> to calculate summary statistics for the <b>release year</b> of the Netflix content.

We will calculate the <b>mean</b> and <b>standard deviation</b>, then create a new derived feature called <b>z_score</b>.

The z-score indicates how far each release year is from the average release year in terms of standard deviations.

</p>

<div style="background-color:#0A2828; border-left:5px solid #14B8A6; padding:15px; border-radius:10px; margin-top:15px;">

<p style="color:#CCFBF1; margin:0; line-height:1.7;">

💡 <b>Key Idea:</b> A z-score close to <b>0</b> means the value is close to the mean, while a positive or negative z-score indicates that the value is above or below the mean.

</p>

</div>

</div>


```python
# Compute the mean and standard deviation of release years using NumPy

mean_release_year = np.mean(lab_df["release_year"])
std_release_year = np.std(lab_df["release_year"])

print("Mean release year:", mean_release_year)
print("Standard deviation:", std_release_year)


# Derived feature: z-score of each release year

lab_df["z_score"] = (
    (lab_df["release_year"] - mean_release_year)
    / std_release_year
)

lab_df[["title", "type", "release_year", "z_score"]].head()
```

<div style="background-color:#102F2F; border-left:6px solid #0D9488; padding:20px; border-radius:12px;">

<h3 style="color:#5EEAD4; margin:0; border-bottom:2px solid #0D9488; padding-bottom:8px;">

🔹 Step 3: Produce Three Labeled Plots

</h3>

<p style="color:#E2E8F0; font-size:16px; line-height:1.8; margin-top:15px;">

In this step, we use <b style="color:#5EEAD4;">Matplotlib</b> to create three different visualizations that explore important aspects of the Netflix dataset.

Each plot answers a different type of question:

</p>

<ul style="color:#E2E8F0; font-size:15px; line-height:1.9;">

<li>
📉 <b style="color:#5EEAD4;">Histogram:</b> What is the distribution of Netflix content across release years?
</li>

<li>
🔵 <b style="color:#5EEAD4;">Scatter Plot:</b> How does the number of Netflix titles vary across release years?
</li>

<li>
📊 <b style="color:#5EEAD4;">Bar Plot:</b> Which countries have the largest number of titles in the dataset?
</li>

</ul>

<div style="background-color:#0A2828; border-left:5px solid #14B8A6; padding:15px; border-radius:10px; margin-top:15px;">

<p style="color:#CCFBF1; margin:0; line-height:1.7;">

💡 <b>Goal:</b> Instead of looking at raw values in a table, these visualizations help us identify patterns, concentration, and differences in the Netflix catalog more quickly.

</p>

</div>

</div>


```python
# Netflix Theme Color
netflix_red = "#E50914"


# 1. Histogram: Distribution of Release Years

plt.hist(
    lab_df["release_year"],
    bins=30,
    color=netflix_red
)

plt.xlabel("Release Year")
plt.ylabel("Frequency")
plt.title("Distribution of Netflix Content by Release Year")

plt.show()


# 2. Scatter Plot: Number of Titles by Release Year

year_counts = lab_df.groupby("release_year").size()

plt.scatter(
    year_counts.index,
    year_counts.values,
    color=netflix_red,
    alpha=0.5
)

plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.title("Netflix Titles by Release Year")

plt.show()


# 3. Bar Plot: Top 10 Countries by Number of Titles

top10_countries = (
    lab_df["country"]
    .dropna()
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)

plt.bar(
    top10_countries.index,
    top10_countries.values,
    color=netflix_red
)

plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.title("Top 10 Countries by Number of Netflix Titles")

plt.xticks(rotation=45, ha="right")

plt.show()
```
