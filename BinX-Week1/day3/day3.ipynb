## 3.1  Why NumPy


Import numpy and print version

```python
import numpy as np
print(np.__version__)
```

## 3.2 Creating Arrays

```python
numbers = np.array([1, 2, 3, 4, 5])
print(numbers)
zero = np.zeros((1, 2))
ones = np.ones((3))
print(zero)
print(ones)
arange = np.arange (0, 10, 2)
linspaces = np.linspace(0, 1, 5)
print(arange)
print(linspaces)
random = np.random.rand(3, 3)
print(random) 
```

## 3.3 Array Attributes, Indexing & Slicing

```python
m = np.array([[1, 2, 3], [4, 5, 6]])
print(m.shape)     
print(m.dtype)      
print(m.ndim)      
print(m[0, 2]) 
print(m[1, 1]) 
print(m[:, 1])      
print(m[1, :])  
```

## 3.4 Vectorized Operations

```python
a  = np.array([1, 2, 3, 4])
print(a * 2)         
print(a + a) 
print(a.mean()) 
print(a[a > 2]) 
```

## 3.5 Broadcasting

```python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
row    = np.array([10, 20, 30])
print(matrix + row) 
```

#  Hands-On Lab: NumPy Array Manipulation


Step 1: Create a 2D array (4,4) with values 1-16, print shape and dtype

```python
numbers = np.arange(1, 17).reshape(4, 4)

print("My 4x4 array:")
print(numbers)
print("\nShape:", numbers.shape)
print("Data type:", numbers.dtype)
```

Step 2: Use slicing to extract second column and last row


```python
second_column = numbers[:, 1]
last_row = numbers[-1, :]
print("Second column:", second_column)
print("Last row:", last_row)
```

Step 3: Use boolean mask to select values greater than the array's mean

```python
average = numbers.mean()
above_average = numbers[numbers > average]
print("Average value:", average)
print("Values greater than average:", above_average)
```

Step 4: Add a 1D row to every row of a 2D array using broadcasting

```python
extra_row = np.array([10, 20, 30, 40])
result = numbers + extra_row
print("Original array:")
print(numbers)
print("\nRow to add:", extra_row)
print("\nResult after broadcasting:")
print(result)
```
