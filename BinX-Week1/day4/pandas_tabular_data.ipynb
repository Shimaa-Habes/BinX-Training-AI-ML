4.4 Cleaning Data

<div style="height:2px; width:55px; background-color:#7FA66A; margin:10px 0;"></div>

<p style="color:#B8CBB5; font-size:15px; line-height:1.8; margin:0;">
🌱 Real-world data is rarely perfect. Before analysis, we need to check for
<b>missing values</b>, <b>duplicates</b>, and <b>incorrect data types</b>.
</p>

<p style="color:#A8BDA8; font-size:14px; line-height:1.7; margin:9px 0 0 0;">
🔍 Inspect &nbsp;•&nbsp; 🧹 Clean &nbsp;•&nbsp; 🌿 Prepare &nbsp;•&nbsp; 📊 Analyze
</p>

</div>

<h3 style="color:#B8D49F; font-family:Georgia,serif; font-size:20px; text-shadow:0 0 5px rgba(127,166,106,0.25); margin:0; letter-spacing:0.3px;">
🔍 Count Missing Values per Column
</h3>

```python
print("\033[94mMissing values per column:\033[0m")
df.isnull().sum()
```

```python
missing_count = df["value"].isnull().sum()
print("\033[93mMissing values in 'value' column:\033[0m", missing_count)

df["value"] = df["value"].fillna(df["value"].mean())
print("\033[92mFilled missing values with the mean.\033[0m")
```

```python
print("Number of duplicate rows :", df.duplicated().sum())
```

<h3 style="color:#B8D49F; font-family:Georgia,serif; font-size:20px; text-shadow:0 0 5px rgba(127,166,106,0.25); margin:0; letter-spacing:0.3px;">
♻️ Remove Duplicates if Any Exist
</h3>

```python
df.drop_duplicates(inplace=True)
print("Dataset after remove duplicates - Shape-",df.shape)
```

<h2 style="color:#E3EDD9; font-family:Georgia,serif; font-size:24px; margin:0;">
4.5 Grouping and Aggregation
</h2>

<div style="height:2px; width:55px; background-color:#7FA66A; margin:10px 0;"></div>

<p style="color:#B8CBB5; font-size:15px; line-height:1.8; margin:0;">
🌍 Group the data by <b>country</b> and calculate the <b>average CO₂ value</b> for each one.
</p>

<p style="color:#A8BDA8; font-size:14px; margin:9px 0 0 0;">
📊 Group &nbsp;→&nbsp; 🧮 Calculate Average &nbsp;→&nbsp; 🌱 Compare Countries
</p>

</div>

```python
print("\033[94mAverage CO2 value per country:\033[0m")
df.groupby("country_name")["value"].mean()
```

<h3 style="color:#B8D49F; font-family:Georgia,serif; font-size:20px; text-shadow:0 0 5px rgba(127,166,106,0.25); margin:0; letter-spacing:0.3px;">
🌿 Multiple Aggregations per Country — Mean &amp; Maximum CO₂ Value
</h3>

```python
print("\033[92mMean and max value per country:\033[0m")
df.groupby("country_name")["value"].agg(["mean", "max"])
```
