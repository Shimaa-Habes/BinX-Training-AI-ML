4.4 Cleaning Data
</h2>
<p style="color:#93C5FD; font-size:16px; line-height:1.6; margin-top:10px;">
Real data is rarely perfect — missing values, duplicates, and wrong
types are common. Cleaning it correctly is usually the biggest part of
any project, and everything downstream depends on doing it right.
</p>
</div>

count missing values per column

```python
print("\033[94mMissing values per column:\033[0m")
df.isnull().sum()
```

```python
missing_count = df["value"].isnull().sum()
print("\033[93mMissing values in 'value' column:\033[0m", missing_count)

df["value"] = df["value"].fillna(df["value"].mean())
print("\033[92mFilled missing values with the mean.\033[0m")
```

```python
print("Number of duplicate rows :", df.duplicated().sum())
```

<h3 style="color:#FFFFFF; text-shadow:0 0 5px #2563EB;">Remove Duplicates if any exist </h3>

```python
df.drop_duplicates(inplace=True)
print("Dataset after remove duplicates - Shape-",df.shape)
```

<div style="background-color:#0A1A2E; border-left:5px solid #2563EB; padding:15px 20px; border-radius:10px;">
<h2 style="color:#FFFFFF; font-size:24px; text-shadow:0 0 5px #2563EB; margin:0;">
4.5 Grouping and Aggregation
</h2>
<p style="color:#93C5FD; font-size:16px; line-height:1.6; margin-top:10px;">
Average CO2 value per country.
</p>
</div>

```python
print("\033[94mAverage CO2 value per country:\033[0m")
df.groupby("country_name")["value"].mean()
```

<h3 style="color:#FFFFFF; text-shadow:0 0 5px #2563EB;">Multiple aggregations per country (mean and max value)</h3>

```python
print("\033[92mMean and max value per country:\033[0m")
df.groupby("country_name")["value"].agg(["mean", "max"])
```

<h3 style="color:#FFFFFF; text-shadow:0 0 5px #2563EB;">Step 1: Load a CSV dataset and print shape, columns, dtypes</h3>

```python
lab_df = pd.read_csv("co2_emissions_data.csv")

print("\033[94mShape:\033[0m", lab_df.shape)
print("\033[92mColumns:\033[0m", lab_df.columns.tolist())
print("\033[93mData types:\033[0m")
print(lab_df.dtypes)
```

<h3 style="color:#FFFFFF; text-shadow:0 0 5px #2563EB;">Step 2: Count missing values and handle them (with justification)</h3>

```python
print("\033[94mMissing values per column:\033[0m")
lab_df.isnull().sum()
```

<div style="background-color:#0A1A2E; border-left:5px solid #2563EB; padding:15px 20px; border-radius:10px;">
<p style="color:#93C5FD; font-size:16px; line-height:1.6; margin:0;">
<b>Justification:</b> The dataset has zero missing values in all columns
(country_code, country_name, year, value). No filling or dropping is
needed — the data is already complete and ready for analysis.
</p>
</div>

<h3 style="color:#FFFFFF; text-shadow:0 0 5px #2563EB;">Step 3: Filter the data to a meaningful subset and describe what you find</h3>

```python
lab_threshold = lab_df["value"].mean()
high_emission = lab_df[lab_df["value"] > lab_threshold]

print("\033[92mThreshold (mean value):\033[0m", lab_threshold)
print("\033[94mRows above threshold:\033[0m", high_emission.shape[0], "out of", lab_df.shape[0])
high_emission.head()
```

```python
print("Total rows in lab_df:", lab_df.shape[0])
print("Rows above threshold:", high_emission.shape[0])
```

<div style="background-color:#0A1A2E; border-left:5px solid #2563EB; padding:15px 20px; border-radius:10px;">
<p style="color:#93C5FD; font-size:16px; line-height:1.6; margin:0;">
<b>Findings:</b> Filtering the dataset to records above the mean value
shows 1727 out of 13953 total rows represent higher-than-average CO2
emissions. This subset highlights which countries/years contribute most
to global emissions, with entities like "Arab World" appearing among the
higher-emission records.
</p>
</div>

<h3 style="color:#FFFFFF; text-shadow:0 0 5px #2563EB;">Step 4: Use groupby to compute an aggregate statistic per category and interpret it</h3>

```python
country_avg = lab_df.groupby("country_name")["value"].mean().sort_values(ascending=False)
print("\033[92mTop 10 countries by average CO2 value:\033[0m")
country_avg.head(10)
```

<p style="color:#93C5FD; font-size:16px; line-height:1.6; margin:0;">
<b>Interpretation:</b> The top 10 entities above show the highest average
CO2 values in the dataset, dominated by large aggregated regions rather
than individual countries — highlighting the need to separate country
level records from group-level ones for precise analysis.
