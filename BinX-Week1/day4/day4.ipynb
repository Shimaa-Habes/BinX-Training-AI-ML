## 4.1 Series and DataFrames


Quick example showing the difference between a Series and a DataFrame

```python
import pandas as pd 
# Series : Single labeled column 
series = pd.Series([1, 2, 3], name="Example")
print(series) 
print(type(series))

# DataFrame : full table (columns and rows)
dataframe = pd.DataFrame({"even":[2, 4, 6], "odd":[1, 3, 5]})
print(dataframe)
print(type(dataframe))

```

## 4.5 Grouping and Aggregation

```python
df.groupby("country_name").agg({"value": ["mean", "max"]})
```

# Hands-On Lab: Exploring a Real Dataset

Step 1: Load a CSV dataset and print shape, columns, dtypes

```python
lab_df = pd.read_csv("data.csv")
print("Shape : ", lab_df.shape)
print("Columns : ", lab_df.columns.tolist())
print("Data Types :",lab_df.dtypes )
```

Step 2: Count missing values per column and handle them (with a documented justification)


```python
missing_counts = lab_df.isnull().sum()
print("Missing values per column : ",missing_counts)
```

**Justification:** The dataset has zero missing values in all columns
(country_code, country_name, year, value). No filling or dropping is needed —
the data is already complete and ready for analysis.

Step 3: Filter the data to a meaningful subset and describe what you find


```python
lab_threshold = lab_df["value"].mean()
high_emission = lab_df[lab_df["value"]>lab_threshold]
print("Threshold(mean value) : ", lab_threshold)
print("Number of rows above threshold :",lab_df.shape[0])
high_emission.head()
```

```python
print("Total rows in lab_df:", lab_df.shape[0])
print("Rows above threshold:", high_emission.shape[0])
```

**Findings:** Filtering the dataset to records above the mean value shows
1727 out of 13953 total rows represent higher-than-average CO2 emissions.
This subset highlights which countries/years contribute most to global emissions,
with entities like "Arab World" appearing among the higher-emission records.

Step 4: Use groupby to compute an aggregate statistic per category and interpret it in a Markdown cell


```python
country_avg = lab_df.groupby("country_name")["value"].mean().sort_values(ascending=False)
country_avg.head(10)
```
