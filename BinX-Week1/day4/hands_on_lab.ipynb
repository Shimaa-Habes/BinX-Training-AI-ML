<div style="background-color:#203326; border-left:4px solid #9BBF87; padding:14px 18px; border-radius:10px; margin-top:10px;">

<p style="color:#D5E8D8; font-size:15px; line-height:1.7; margin:0;">

<div style="background-color:#1B2A20; border:1px solid #4F6F52; border-left:5px solid #7FA66A; padding:16px 20px; border-radius:11px;">

<h3 style="color:#B8D49F; font-family:Georgia,serif; font-size:20px; margin:0;">
🍃 Step 3 — Filter & Explore a Meaningful Subset
</h3>

<p style="color:#A8BDA8; font-size:14px; margin:7px 0 0 0;">
Filter the data using a meaningful condition and examine the resulting records.
</p>

</div>

```python
lab_threshold = lab_df["value"].mean()
high_emission = lab_df[lab_df["value"] > lab_threshold]

print("\033[92mThreshold (mean value):\033[0m", lab_threshold)
print("\033[94mRows above threshold:\033[0m", high_emission.shape[0], "out of", lab_df.shape[0])
high_emission.head()
```

```python
print("Total rows in lab_df:", lab_df.shape[0])
print("Rows above threshold:", high_emission.shape[0])
```

<div style="background-color:#203326; border-left:4px solid #9BBF87; padding:14px 18px; border-radius:10px; margin-top:10px;">

<p style="color:#D5E8D8; font-size:15px; line-height:1.7; margin:0;">
🌍 <b style="color:#B8D49F;">Findings</b><br>
Filtering for records above the mean value resulted in
<b>1,727 out of 13,953 rows</b> with higher-than-average CO₂ emissions.
The subset highlights countries and aggregated regions with larger emission
values, including entities such as <b>"Arab World"</b>.
</p>

</div>

<div style="background-color:#1B2A20; border:1px solid #4F6F52; border-left:5px solid #7FA66A; padding:16px 20px; border-radius:11px;">

<h3 style="color:#B8D49F; font-family:Georgia,serif; font-size:20px; margin:0;">
📊 Step 4 — Group & Aggregate
</h3>

<p style="color:#A8BDA8; font-size:14px; margin:7px 0 0 0;">
Use <code>groupby</code> to calculate an aggregate statistic for each category
and interpret the results.
</p>

</div>

```python
country_avg = lab_df.groupby("country_name")["value"].mean().sort_values(ascending=False)
print("\033[92mTop 10 countries by average CO2 value:\033[0m")
country_avg.head(10)
```

<div style="background-color:#203326; border-left:4px solid #9BBF87; padding:14px 18px; border-radius:10px; margin-top:10px;">

<p style="color:#D5E8D8; font-size:15px; line-height:1.7; margin:0;">
🌱 <b style="color:#B8D49F;">Interpretation</b><br>
The top 10 entities with the highest average CO₂ values are mainly
<b>aggregated regions</b> rather than individual countries. This shows why
country-level records should be separated from regional or global groups
when performing precise comparisons.
</p>
