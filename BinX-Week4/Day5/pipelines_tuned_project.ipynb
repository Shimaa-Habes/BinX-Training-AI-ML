⚠️ Data Leakage
</h3>

<p style="
font-family:'Segoe UI',Arial,sans-serif;
font-size:18px;
line-height:1.8;
color:#D9E1E2;
margin:0;
">
<strong>Data leakage</strong> happens when information from the test data accidentally influences the training process. This gives the model access to information it should not have during training and can lead to unrealistically high evaluation results.
</p>

</div>

<div style="
background:rgba(19,40,50,0.72);
border-left:4px solid #D6B870;
border-radius:14px;
padding:24px 28px;

">

<h3 style="
font-family:Georgia,'Times New Roman',serif;
color:#D6B870;
font-size:22px;
font-weight:700;
margin:0 0 12px 0;
">
🔍 Why Split Before Preprocessing?
</h3>

<p style="
font-family:'Segoe UI',Arial,sans-serif;
font-size:18px;
line-height:1.8;
color:#D9E1E2;
margin:0;
">
Preprocessing the entire dataset before splitting can cause <strong>data leakage</strong> because transformations such as scaling may use information from the test set. The model should only learn preprocessing parameters from the <strong>training data</strong>, while the test data remains completely unseen until evaluation.
</p>

</div>

<div style="
background:rgba(19,40,50,0.72);
border-left:4px solid #D6B870;
border-radius:14px;
padding:24px 28px;

">

<h3 style="

">
🔗 How Pipeline Keeps Everything Together
</h3>

<p style="
font-family:'Segoe UI',Arial,sans-serif;
font-size:18px;
line-height:1.8;
color:#D9E1E2;
margin:0;
">
A <strong>Pipeline</strong> connects preprocessing steps and the machine learning model into one workflow. This ensures that preprocessing is learned only from the training data and then applied consistently to new data, helping prevent <strong>data leakage</strong> and making the modeling process more reliable.
</p>

</div>
