<span style="font-size:12px; letter-spacing:3px; color:#9D174D; text-transform:uppercase; font-weight:600;">3.1 · Model Fit</span>

<h2 style="color:#500724; font-size:24px; font-weight:600; margin:8px 0 12px 0;">
Underfitting vs Overfitting
</h2>

<p style="color:#831843; font-size:13px; line-height:1.7; margin:0;">
Before tuning a machine learning model, we need to understand whether it is learning too little or too much from the training data.
</p>

<div style="height:1px; background:linear-gradient(90deg,#F472B6,transparent); margin:22px 0;"></div>

<div style="display:flex; gap:14px; flex-wrap:wrap; box-sizing:border-box; width:100%;">

<div style="width:calc(50% - 7px); box-sizing:border-box; background:#FFFFFF; border:1px solid #F9A8D4; border-radius:14px; padding:18px;">

<div style="display:flex; align-items:center; gap:10px; margin-bottom:10px;">
<span style="font-size:22px;">📉</span>
<span style="font-size:15px; color:#500724; font-weight:600;">Underfitting</span>
</div>

<p style="font-size:12px; color:#831843; line-height:1.7; margin:0 0 12px 0;">
The model is too simple to capture the important patterns in the data.
</p>

<div style="background:#FDF2F8; border-radius:10px; padding:12px; box-sizing:border-box;">
<div style="font-size:11px; color:#9D174D; font-weight:600; margin-bottom:6px;">Typical symptoms</div>

<div style="font-size:12px; color:#831843; line-height:1.8;">
• Low training performance<br>
• Low validation performance<br>
• Small gap between train and validation scores
</div>

</div>
</div>


<div style="width:calc(50% - 7px); box-sizing:border-box; background:#FFFFFF; border:1px solid #F9A8D4; border-radius:14px; padding:18px;">

<div style="display:flex; align-items:center; gap:10px; margin-bottom:10px;">
<span style="font-size:22px;">📈</span>
<span style="font-size:15px; color:#500724; font-weight:600;">Overfitting</span>
</div>

<p style="font-size:12px; color:#831843; line-height:1.7; margin:0 0 12px 0;">
The model learns the training data too closely, including noise and patterns that do not generalize.
</p>

<div style="background:#FDF2F8; border-radius:10px; padding:12px; box-sizing:border-box;">
<div style="font-size:11px; color:#9D174D; font-weight:600; margin-bottom:6px;">Typical symptoms</div>

<div style="font-size:12px; color:#831843; line-height:1.8;">
• Very high training performance<br>
• Lower validation performance<br>
• Large gap between train and validation scores
</div>

</div>
</div>

</div>

<div style="margin-top:16px; background:#500724; border-radius:14px; padding:16px 18px;">

<div style="display:flex; align-items:center; gap:10px;">
<span style="font-size:18px;">🎯</span>
<span style="font-size:13px; color:#FCE7F3; font-weight:600;">The goal</span>
</div>

<p style="font-size:12px; color:#FBCFE8; line-height:1.7; margin:8px 0 0 0;">
A well-fitted model should learn the underlying patterns without simply memorizing the training examples. This allows it to perform well on unseen data.
</p>

</div>

</div>
</div>

<div style="background:linear-gradient(160deg,#FDF2F8 0%,#FCE7F3 100%); border-radius:20px; overflow:hidden; border:1px solid #F9A8D4;">

<div style="padding:28px 32px;">

<div style="display:flex; align-items:center; gap:10px; margin-bottom:8px;">
<span style="font-size:20px;">⚖️</span>
<span style="font-size:12px; letter-spacing:3px; color:#9D174D; text-transform:uppercase; font-weight:600;">3.2 · Model Complexity</span>
</div>

<h2 style="color:#500724; font-size:24px; font-weight:600; margin:8px 0 12px 0;">
Bias–Variance Trade-off
</h2>

<p style="color:#831843; font-size:13px; line-height:1.7; margin:0;">
The bias–variance trade-off helps us understand why a model can fail by being either too simple or too complex.
</p>

<div style="height:1px; background:linear-gradient(90deg,#F472B6,transparent); margin:22px 0;"></div>

<div style="display:flex; gap:14px; flex-wrap:wrap;">

<div style="flex:1; min-width:220px; background:#FFFFFF; border:1px solid #F9A8D4; border-radius:14px; padding:18px;">

<div style="font-size:22px; margin-bottom:8px;">🧠</div>

<div style="font-size:15px; color:#500724; font-weight:600; margin-bottom:8px;">
High Bias
</div>

<p style="font-size:12px; color:#831843; line-height:1.7; margin:0;">
The model makes strong assumptions and is too simple to capture the real patterns in the data.
</p>

<div style="margin-top:12px; padding:10px; background:#FDF2F8; border-radius:9px; font-size:11px; color:#9D174D;">
<strong>Usually leads to:</strong> Underfitting
</div>

</div>

<div style="display:flex; align-items:center; justify-content:center; min-width:30px; color:#BE185D; font-size:20px;">
↔
</div>

<div style="flex:1; min-width:220px; background:#FFFFFF; border:1px solid #F9A8D4; border-radius:14px; padding:18px;">

<div style="font-size:22px; margin-bottom:8px;">🌀</div>

<div style="font-size:15px; color:#500724; font-weight:600; margin-bottom:8px;">
High Variance
</div>

<p style="font-size:12px; color:#831843; line-height:1.7; margin:0;">
The model is highly sensitive to the training data and may learn noise instead of general patterns.
</p>

<div style="margin-top:12px; padding:10px; background:#FDF2F8; border-radius:9px; font-size:11px; color:#9D174D;">
<strong>Usually leads to:</strong> Overfitting
</div>

</div>

</div>

<div style="margin-top:16px; background:#FFFFFF; border:1px solid #F9A8D4; border-radius:14px; padding:18px;">

<div style="display:flex; align-items:center; gap:10px; margin-bottom:10px;">
<span style="font-size:19px;">🎯</span>
<span style="font-size:13px; letter-spacing:2px; color:#9D174D; text-transform:uppercase; font-weight:600;">
Finding the balance
</span>
</div>

<p style="font-size:12px; color:#831843; line-height:1.8; margin:0;">
As model complexity increases, bias generally decreases while variance increases. 
The goal is not to minimize only one of them, but to find a level of complexity that gives good performance on both training and unseen data.
</p>

</div>

<div style="margin-top:16px; display:flex; align-items:center; gap:12px; background:#500724; border-radius:14px; padding:16px 18px;">

<span style="font-size:20px;">💡</span>

<div>
<div style="font-size:12px; color:#FCE7F3; font-weight:600; margin-bottom:4px;">
Key idea
</div>

<div style="font-size:12px; color:#FBCFE8; line-height:1.6;">
Too simple → high bias → underfitting &nbsp; | &nbsp;
Too complex → high variance → overfitting
</div>
</div>

</div>

</div>
</div>

<span style="font-size:12px; letter-spacing:3px; color:#9D174D; text-transform:uppercase; font-weight:600;">3.3 · Model Diagnosis</span>
</div>

<h2 style="color:#500724; font-size:24px; font-weight:600; margin:8px 0 12px 0;">
Diagnosing Model Fit
</h2>

<p style="color:#831843; font-size:13px; line-height:1.7; margin:0;">
One of the easiest ways to diagnose a model is to compare its performance on the training data with its performance on unseen validation data.
</p>

<div style="height:1px; background:linear-gradient(90deg,#F472B6,transparent); margin:22px 0;"></div>


<div style="display:flex; gap:14px; flex-wrap:wrap; box-sizing:border-box; width:100%;">

<div style="width:calc(50% - 7px); box-sizing:border-box; background:#FFFFFF; border:1px solid #F9A8D4; border-radius:14px; padding:18px;">

<div style="display:flex; align-items:center; gap:10px; margin-bottom:10px;">
<span style="font-size:21px;">📉</span>
<span style="font-size:14px; color:#500724; font-weight:600;">Underfitting Pattern</span>
</div>

<div style="background:#FDF2F8; border-radius:10px; padding:12px; box-sizing:border-box;">

<div style="font-size:11px; color:#9D174D; font-weight:600; margin-bottom:7px;">
What we observe
</div>

<div style="font-size:12px; color:#831843; line-height:1.8;">
Training score → Low<br>
Validation score → Low<br>
Train–Validation gap → Small
</div>

</div>

<p style="font-size:12px; color:#831843; line-height:1.7; margin:12px 0 0 0;">
The model performs poorly even on the data it was trained on, which suggests that it is not complex enough to learn the underlying patterns.
</p>

</div>


<div style="width:calc(50% - 7px); box-sizing:border-box; background:#FFFFFF; border:1px solid #F9A8D4; border-radius:14px; padding:18px;">

<div style="display:flex; align-items:center; gap:10px; margin-bottom:10px;">
<span style="font-size:21px;">📈</span>
<span style="font-size:14px; color:#500724; font-weight:600;">Overfitting Pattern</span>
</div>

<div style="background:#FDF2F8; border-radius:10px; padding:12px; box-sizing:border-box;">

<div style="font-size:11px; color:#9D174D; font-weight:600; margin-bottom:7px;">
What we observe
</div>

<div style="font-size:12px; color:#831843; line-height:1.8;">
Training score → Very high<br>
Validation score → Lower<br>
Train–Validation gap → Large
</div>

</div>

<p style="font-size:12px; color:#831843; line-height:1.7; margin:12px 0 0 0;">
The model performs extremely well on training data but loses performance on unseen data, indicating that it has learned the training examples too closely.
</p>

</div>

</div>


<div style="margin-top:16px; background:#FFFFFF; border:1px solid #F9A8D4; border-radius:14px; padding:18px; box-sizing:border-box;">

<div style="display:flex; align-items:center; gap:10px; margin-bottom:10px;">
<span style="font-size:19px;">📊</span>
<span style="font-size:13px; letter-spacing:2px; color:#9D174D; text-transform:uppercase; font-weight:600;">
Healthy Fit
</span>
</div>

<p style="font-size:12px; color:#831843; line-height:1.8; margin:0;">
A well-fitted model should achieve strong training and validation performance with a relatively small gap between them. The validation score is especially important because it gives us an idea of how well the model can generalize to new observations.
</p>

</div>


<div style="margin-top:16px; background:#500724; border-radius:14px; padding:16px 18px; box-sizing:border-box;">

<div style="display:flex; align-items:center; gap:10px;">
<span style="font-size:18px;">💡</span>
<span style="font-size:13px; color:#FCE7F3; font-weight:600;">
What should we look for?
</span>
</div>

<p style="font-size:12px; color:#FBCFE8; line-height:1.7; margin:8px 0 0 0;">
Instead of looking only at the highest training score, we should examine the relationship between training and validation performance. A large gap is often a warning sign of overfitting.
</p>

</div>

</div>
</div>

<div style="background:linear-gradient(160deg,#FDF2F8 0%,#FCE7F3 100%); border-radius:20px; overflow:hidden; border:1px solid #F9A8D4;">

<div style="padding:28px 32px; box-sizing:border-box;">

<div style="display:flex; align-items:center; gap:10px; margin-bottom:8px;">
<span style="font-size:20px;">🛡️</span>
<span style="font-size:12px; letter-spacing:3px; color:#9D174D; text-transform:uppercase; font-weight:600;">
3.4 · Model Control
</span>
</div>

<h2 style="color:#500724; font-size:24px; font-weight:600; margin:8px 0 12px 0;">
Regularization
</h2>

<p style="color:#831843; font-size:13px; line-height:1.8; margin:0;">
Regularization is a technique used to reduce overfitting by adding a penalty to the model's complexity. Instead of allowing the model to focus only on minimizing training error, regularization encourages simpler solutions that are more likely to generalize well to unseen data.
</p>

<div style="height:1px; background:linear-gradient(90deg,#F472B6,transparent); margin:22px 0;"></div>

<div style="display:flex; gap:14px; flex-wrap:wrap; box-sizing:border-box; width:100%;">

<div style="width:calc(50% - 7px); box-sizing:border-box; background:#FFFFFF; border:1px solid #F9A8D4; border-radius:14px; padding:18px;">

<div style="font-size:21px; margin-bottom:8px;">〽️</div>

<div style="font-size:15px; color:#500724; font-weight:600; margin-bottom:8px;">
L1 Regularization — Lasso
</div>

<p style="font-size:12px; color:#831843; line-height:1.7; margin:0;">
Adds a penalty based on the absolute values of the model coefficients. L1 regularization can push some coefficients all the way to zero, which can help with feature selection.
</p>

</div>

<div style="width:calc(50% - 7px); box-sizing:border-box; background:#FFFFFF; border:1px solid #F9A8D4; border-radius:14px; padding:18px;">

<div style="font-size:21px; margin-bottom:8px;">〰️</div>

<div style="font-size:15px; color:#500724; font-weight:600; margin-bottom:8px;">
L2 Regularization — Ridge
</div>

<p style="font-size:12px; color:#831843; line-height:1.7; margin:0;">
Adds a penalty based on the squared values of the model coefficients. L2 regularization reduces the influence of large coefficients without usually forcing them exactly to zero.
</p>

</div>

</div>

<div style="margin-top:16px; background:#500724; border-radius:14px; padding:16px 18px; box-sizing:border-box;">

<div style="display:flex; align-items:center; gap:10px;">
<span style="font-size:18px;">🎯</span>
<span style="font-size:13px; color:#FCE7F3; font-weight:600;">
Why use Regularization?
</span>
</div>

<p style="font-size:12px; color:#FBCFE8; line-height:1.7; margin:8px 0 0 0;">
The main goal is to control model complexity and reduce overfitting, so that the model learns useful patterns rather than memorizing the training data.
</p>

</div>

<span style="font-size:12px; letter-spacing:3px; color:#9D174D; text-transform:uppercase; font-weight:600;">
L2 · Ridge Classifier
</span>
</div>

<h2 style="color:#500724; font-size:23px; font-weight:600; margin:8px 0 12px 0;">
Ridge Classifier
</h2>

<p style="color:#831843; font-size:13px; line-height:1.8; margin:0;">

</p>

<p style="font-size:12px; color:#831843; line-height:1.8; margin:0;">

<h2 style="color:#500724; font-size:24px; font-weight:600; margin:8px 0 12px 0;">
Understanding the Loss Function
</h2>

<p style="color:#831843; font-size:13px; line-height:1.8; margin:0;">
Ridge Classifier does not only aim to make correct predictions. It also adds a regularization penalty to control the size of the model coefficients and reduce unnecessary complexity.
</p>

<div style="margin-top:20px; background:#500724; border-radius:14px; padding:18px; box-sizing:border-box; text-align:center;">

<div style="font-size:12px; color:#FCE7F3; font-weight:600; margin-bottom:10px;">
L2 Regularized Objective
</div>

<div style="font-size:17px; color:#FBCFE8; font-weight:600;">
Loss + α × Σ(w²)
</div>

</div>

<div style="margin-top:16px; display:flex; gap:12px; flex-wrap:wrap; box-sizing:border-box; width:100%;">

<div style="width:calc(50% - 6px); box-sizing:border-box; background:#FFFFFF; border:1px solid #F9A8D4; border-radius:13px; padding:16px;">

<div style="font-size:18px; margin-bottom:7px;">📉</div>

<div style="font-size:13px; color:#500724; font-weight:600;">
Original Loss
</div>

<p style="font-size:12px; color:#831843; line-height:1.7; margin:7px 0 0 0;">
Measures how well the model's predictions match the target values.
</p>

</div>

<div style="width:calc(50% - 6px); box-sizing:border-box; background:#FFFFFF; border:1px solid #F9A8D4; border-radius:13px; padding:16px;">

<div style="font-size:18px; margin-bottom:7px;">🛡️</div>

<div style="font-size:13px; color:#500724; font-weight:600;">
L2 Penalty
</div>

<p style="font-size:12px; color:#831843; line-height:1.7; margin:7px 0 0 0;">
Penalizes large coefficients by adding the squared weights to the objective.
</p>

</div>

</div>

<div style="margin-top:16px; background:#FFFFFF; border:1px solid #F9A8D4; border-radius:14px; padding:17px; box-sizing:border-box;">

🎯 The main idea
</div>

<p style="font-size:12px; color:#831843; line-height:1.8; margin:0;">
The model searches for a balance between making accurate predictions and keeping its coefficients under control. This helps prevent the model from becoming overly dependent on individual features and can improve generalization to unseen data.
</p>

</div>

</div>

<h2 style="color:#500724; font-size:24px; font-weight:600; margin:8px 0 12px 0;">
Deliberately Overfit a Decision Tree
</h2>

<p style="color:#831843; font-size:13px; line-height:1.8; margin:0;">
We will intentionally create a very deep Decision Tree with minimal restrictions. This allows the model to learn the training examples extremely closely, including patterns that may not generalize to unseen data.
</p>

<div style="margin-top:18px; background:#500724; border-radius:14px; padding:16px 18px; box-sizing:border-box;">

<div style="display:flex; align-items:center; gap:10px;">
<span style="font-size:18px;">🔎</span>
<span style="font-size:13px; color:#FCE7F3; font-weight:600;">
What are we looking for?
</span>
</div>

<p style="font-size:12px; color:#FBCFE8; line-height:1.8; margin:8px 0 0 0;">
A very high training accuracy combined with a noticeably lower validation accuracy. A large gap between the two scores is evidence of <strong>overfitting</strong>.
</p>

</div>

</div>
</div>

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import pandas as pd

overfit_tree = DecisionTreeClassifier(
    max_depth=None,
    min_samples_split=2,
    min_samples_leaf=1,
    random_state=42
)

overfit_tree.fit(X_train, y_train)

train_predictions = overfit_tree.predict(X_train)
test_predictions = overfit_tree.predict(X_test)

train_accuracy = accuracy_score(y_train, train_predictions)
validation_accuracy = accuracy_score(y_test, test_predictions)

accuracy_gap = train_accuracy - validation_accuracy

overfit_results = pd.DataFrame({
    "Metric": [
        "Training Accuracy",
        "Validation Accuracy",
        "Train-Validation Gap"
    ],
    "Score": [
        train_accuracy,
        validation_accuracy,
        accuracy_gap
    ]
})

overfit_results.style \
    .format({"Score": "{:.2%}"}) \
    .set_properties(**{
        'background-color': '#FDF2F8',
        'color': '#500724',
        'border-color': '#F9A8D4',
        'text-align': 'center',
        'padding': '10px'
    }) \
    .set_table_styles([
        {
            'selector': 'th',
            'props': [
                ('background-color', '#500724'),
                ('color', '#FCE7F3'),
                ('font-weight', 'bold'),
                ('border-color', '#F472B6'),
                ('text-align', 'center'),
                ('padding', '10px')
            ]
        },
        {
            'selector': 'td',
            'props': [
                ('border-color', '#F9A8D4'),
                ('padding', '10px')
            ]
        },
        {
            'selector': 'table',
            'props': [
                ('border-collapse', 'collapse'),
                ('border-radius', '10px'),
                ('overflow', 'hidden')
            ]
        }
    ])
```

Step 2 · Underfitting
</span>
</div>

<h2 style="color:#500724; font-size:24px; font-weight:600; margin:8px 0 12px 0;">
Deliberately Underfit a Decision Tree
</h2>

<p style="color:#831843; font-size:13px; line-height:1.8; margin:0;">
Now we will intentionally make the Decision Tree too simple by limiting its depth. The goal is to create a model that cannot capture enough of the patterns in the data, resulting in low performance on both training and validation sets.
</p>

<div style="margin-top:18px; background:#500724; border-radius:14px; padding:16px 18px; box-sizing:border-box;">

<div style="display:flex; align-items:center; gap:10px;">
<span style="font-size:18px;">🔎</span>
<span style="font-size:13px; color:#FCE7F3; font-weight:600;">
What are we looking for?
</span>
</div>

<p style="font-size:12px; color:#FBCFE8; line-height:1.8; margin:8px 0 0 0;">
Low training accuracy together with low validation accuracy, with a relatively small gap between them. This pattern indicates <strong>underfitting</strong>.
</p>

</div>

</div>
</div>

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import pandas as pd

underfit_tree = DecisionTreeClassifier(
    max_depth=1,
    random_state=42
)

underfit_tree.fit(X_train, y_train)

train_predictions_underfit = underfit_tree.predict(X_train)
test_predictions_underfit = underfit_tree.predict(X_test)

train_accuracy_underfit = accuracy_score(
    y_train,
    train_predictions_underfit
)

validation_accuracy_underfit = accuracy_score(
    y_test,
    test_predictions_underfit
)

accuracy_gap_underfit = (
    train_accuracy_underfit - validation_accuracy_underfit
)

underfit_results = pd.DataFrame({
    "Metric": [
        "Training Accuracy",
        "Validation Accuracy",
        "Train-Validation Gap"
    ],
    "Score": [
        train_accuracy_underfit,
        validation_accuracy_underfit,
        accuracy_gap_underfit
    ]
})

underfit_results.style \
    .format({"Score": "{:.2%}"}) \
    .set_properties(**{
        'background-color': '#FDF2F8',
        'color': '#500724',
        'border-color': '#F9A8D4',
        'text-align': 'center',
        'padding': '10px'
    }) \
    .set_table_styles([
        {
            'selector': 'th',
            'props': [
                ('background-color', '#500724'),
                ('color', '#FCE7F3'),
                ('font-weight', 'bold'),
                ('border-color', '#F472B6'),
                ('text-align', 'center'),
                ('padding', '10px')
            ]
        },
        {
            'selector': 'td',
            'props': [
                ('border-color', '#F9A8D4'),
                ('padding', '10px')
            ]
        },
        {
            'selector': 'table',
            'props': [
                ('border-collapse', 'collapse'),
                ('border-radius', '10px'),
                ('overflow', 'hidden')
            ]
        }
    ])
```

</span>

<h2 style="color:#500724; font-size:24px; font-weight:600; margin:8px 0 12px 0;">
Underfitting Analysis

<p style="color:#831843; font-size:13px; line-height:1.8; margin:0;">
The simple Decision Tree achieved <strong>92.31%</strong> training accuracy and <strong>92.11%</strong> validation accuracy, with only a <strong>0.20%</strong> gap between them.

<div style="display:flex; align-items:center; gap:10px; margin-bottom:10px;">
<span style="font-size:20px;">📊</span>
<span style="font-size:15px; color:#500724; font-weight:600;">
What do the scores tell us?
</span>
</div>

<p style="font-size:12px; color:#831843; line-height:1.8; margin:0;">
The very small train-validation gap shows that the model is generalizing consistently and is not suffering from overfitting. However, because both scores remain around 92%, this experiment does not demonstrate strong underfitting. The dataset can still be handled reasonably well by this simple tree.
</p>

</div>

<div style="margin-top:16px; background:#500724; border-radius:14px; padding:16px 18px; box-sizing:border-box;">

<div style="display:flex; align-items:center; gap:10px;">
<span style="font-size:18px;">🎯</span>

Diagnosis
</span>
</div>

<p style="font-size:12px; color:#FBCFE8; line-height:1.8; margin:8px 0 0 0;">
The model has a <strong>very small train-validation gap</strong>, but the evidence for strong underfitting is weak because its performance is still relatively high. This shows why model fit should be diagnosed using both the score levels and the gap between training and validation performance.

Step 3 · Fixing Overfitting
</span>
</div>

<h2 style="color:#500724; font-size:24px; font-weight:600; margin:8px 0 12px 0;">
Reducing Model Complexity
</h2>

<p style="color:#831843; font-size:13px; line-height:1.8; margin:0;">
The previous Decision Tree was deliberately made highly complex, resulting in a large gap between training and validation performance. To reduce overfitting, we will now simplify the model and compare the new results with the original scores.
</p>

<div style="height:1px; background:linear-gradient(90deg,#F472B6,transparent); margin:22px 0;"></div>

<div style="display:flex; gap:14px; flex-wrap:wrap; box-sizing:border-box; width:100%;">

<div style="width:calc(50% - 7px); box-sizing:border-box; background:#FFFFFF; border:1px solid #F9A8D4; border-radius:14px; padding:18px;">

<div style="font-size:21px; margin-bottom:8px;">📈</div>

<div style="font-size:15px; color:#500724; font-weight:600; margin-bottom:8px;">
Original Model
</div>

<p style="font-size:12px; color:#831843; line-height:1.7; margin:0;">
Training Accuracy: <strong>100%</strong><br>
Validation Accuracy: <strong>91.23%</strong><br>
Train-Validation Gap: <strong>8.77%</strong>
</p>

</div>

<div style="width:calc(50% - 7px); box-sizing:border-box; background:#FFFFFF; border:1px solid #F9A8D4; border-radius:14px; padding:18px;">

<div style="font-size:21px; margin-bottom:8px;">🎯</div>

<div style="font-size:15px; color:#500724; font-weight:600; margin-bottom:8px;">
Goal
</div>

<p style="font-size:12px; color:#831843; line-height:1.7; margin:0;">
Reduce the model's complexity so that it relies less on memorizing the training data and generalizes better to unseen data.
</p>

</div>

</div>

<div style="margin-top:16px; background:#500724; border-radius:14px; padding:16px 18px; box-sizing:border-box;">

<div style="display:flex; align-items:center; gap:10px;">
<span style="font-size:18px;">📉</span>
<span style="font-size:13px; color:#FCE7F3; font-weight:600;">
Expected Result
</span>
</div>

<p style="font-size:12px; color:#FBCFE8; line-height:1.8; margin:8px 0 0 0;">
A successful fix should reduce the train-validation gap while maintaining strong validation performance. This indicates that the model has become better at generalizing instead of memorizing the training data.
</p>

</div>

</div>
</div>

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import pandas as pd

fixed_tree = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

fixed_tree.fit(X_train, y_train)

train_predictions_fixed = fixed_tree.predict(X_train)
test_predictions_fixed = fixed_tree.predict(X_test)

train_accuracy_fixed = accuracy_score(
    y_train,
    train_predictions_fixed
)

validation_accuracy_fixed = accuracy_score(
    y_test,
    test_predictions_fixed
)

accuracy_gap_fixed = (
    train_accuracy_fixed - validation_accuracy_fixed
)

fixed_results = pd.DataFrame({
    "Metric": [
        "Training Accuracy",
        "Validation Accuracy",
        "Train-Validation Gap"
    ],
    "Score": [
        train_accuracy_fixed,
        validation_accuracy_fixed,
        accuracy_gap_fixed
    ]
})

fixed_results.style \
    .format({"Score": "{:.2%}"}) \
    .set_properties(**{
        'background-color': '#FDF2F8',
        'color': '#500724',
        'border-color': '#F9A8D4',
        'text-align': 'center',
        'padding': '10px'
    }) \
    .set_table_styles([
        {
            'selector': 'th',
            'props': [
                ('background-color', '#500724'),
                ('color', '#FCE7F3'),
                ('font-weight', 'bold'),
                ('border-color', '#F472B6'),
                ('text-align', 'center'),
                ('padding', '10px')
            ]
        },
        {
            'selector': 'td',
            'props': [
                ('border-color', '#F9A8D4'),
                ('padding', '10px')
            ]
        },
        {
            'selector': 'table',
            'props': [
                ('border-collapse', 'collapse'),
                ('border-radius', '10px'),
                ('overflow', 'hidden')
            ]
        }
    ])
```

<p style="font-size:12px; color:#831843; line-height:1.8; margin:0;">
