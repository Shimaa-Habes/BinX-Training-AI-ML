<div style="

📉 Finding the Best Number of Clusters
</div>

<div style="
    margin-top: 10px;
    font-size: 14px;
    line-height: 1.7;
    color: #CFFAFE;
">
K-Means requires us to choose the number of clusters <b>k</b> before training.
We will test different values of k and use the <b>Elbow Method</b> to identify
where increasing the number of clusters starts providing smaller improvements.
</div>

<div style="
    margin-top: 14px;
    background: #0E7490;
    border-radius: 10px;
    padding: 12px 16px;
    font-size: 14px;
">
💡 <b>Inertia</b> measures how close the data points are to their assigned cluster centers.
Lower inertia means tighter clusters.
</div>

</div>

```python
# Calculate K-Means inertia for different values of k

inertias = []

k_values = range(1, 11)

for k in k_values:
    km = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    km.fit(X_scaled)
    inertias.append(km.inertia_)

plt.figure(figsize=(9, 5))

plt.plot(
    k_values,
    inertias,
    marker="o",
    linewidth=2
)

plt.title("📉 Elbow Method — Choosing k", fontsize=16, fontweight="bold")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.xticks(k_values)
plt.grid(alpha=0.25)

plt.tight_layout()
plt.show()

display(HTML("""
<div style="
    background: linear-gradient(135deg, #0F172A, #164E63);
    border-radius: 14px;
    padding: 16px 20px;
    margin-top: 10px;
    color: #CFFAFE;
">
<div style="color:#67E8F9; font-size:11px; font-weight:700; letter-spacing:1.5px;">
ELBOW METHOD
</div>

<div style="font-size:14px; margin-top:7px; line-height:1.6;">
🔎 Look for the point where the curve begins to flatten.
This point is the <b>elbow</b> and provides a candidate value for <b>k</b>.
</div>
</div>
"""))
```

<div style="color:#334155;font-size:14px;margin-top:8px;line-height:1.6;">
K-Means works with numerical features and calculates distances between data points.
Therefore, we will use meaningful numerical customer attributes and exclude the
customer ID because it is only an identifier.

```python
# Select numerical features for clustering
features = ["Age", "Annual Income (k$)", "Spending Score (1-100)"]

X = df[features].copy()

display(X.head().style
        .set_caption("🎯 Features Selected for Clustering")
        .set_properties(**{
            "text-align": "center",
            "padding": "8px"
        })
        .set_table_styles([
            {
                "selector": "caption",
                "props": [
                    ("background-color", "#0F172A"),
                    ("color", "#67E8F9"),
                    ("font-size", "17px"),
                    ("font-weight", "800"),
                    ("text-align", "left"),
                    ("padding", "11px")
                ]
            },
            {
                "selector": "th",
                "props": [
                    ("background-color", "#164E63"),
                    ("color", "white")
                ]
            }
        ]))

display(HTML(f"""
<div style="
    background:linear-gradient(135deg,#0F172A,#164E63);
    border-radius:12px;
    padding:14px 18px;
    margin-top:12px;
    color:white;
">
<b style="color:#67E8F9;">CLUSTERING INPUT</b><br>
<span style="font-size:20px;font-weight:800;">
{X.shape[0]} customers × {X.shape[1]} numerical features
</span>
</div>
"""))
```

STEP 4
</div>

<div style="color:#0F172A;font-size:22px;font-weight:800;margin-top:5px;">
📏 Scale the Features
</div>

<div style="color:#334155;font-size:14px;margin-top:8px;line-height:1.6;">
K-Means is distance-based. If one feature has a larger numerical range than another,
it can dominate the distance calculation.
<br><br>
We use <b>StandardScaler</b> so that the features are placed on a comparable scale.
</div>

</div>

```python
# Standardize the clustering features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

scaled_df = pd.DataFrame(X_scaled, columns=features)

display(scaled_df.head().style
        .set_caption("📏 Scaled Features Preview")
        .set_properties(**{
            "text-align": "center",
            "padding": "8px"
        })
        .set_table_styles([
            {
                "selector": "caption",
                "props": [
                    ("background-color", "#0F172A"),
                    ("color", "#67E8F9"),
                    ("font-size", "17px"),
                    ("font-weight", "800"),
                    ("text-align", "left"),
                    ("padding", "11px")
                ]
            },
            {
                "selector": "th",
                "props": [
                    ("background-color", "#164E63"),
                    ("color", "white")
                ]
            }
        ]))

display(HTML("""
<div style="
    background:#ECFEFF;
    border:1px solid #A5F3FC;
    border-radius:12px;
    padding:14px 18px;
    margin-top:12px;
    color:#0F172A;
">
<b style="color:#0E7490;">✅ SCALING COMPLETE</b><br>
The numerical features are now standardized and ready for K-Means clustering.
</div>
"""))
```

</div>

<div style="color:#0F172A;font-size:22px;font-weight:800;margin-top:5px;">
🔵 K-Means Clustering
</div>

<div style="color:#334155;font-size:14px;margin-top:8px;line-height:1.6;">
K-Means requires us to choose the number of clusters <b>k</b>.
We will test several possible values and use the <b>Elbow Method</b> to identify
where adding more clusters stops providing a major improvement.
</div>

</div>

```python
# Calculate inertia for different k values
inertias = []

for k in range(1, 11):
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_scaled)
    inertias.append(km.inertia_)

# Plot elbow curve
plt.figure(figsize=(9, 5))
plt.plot(range(1, 11), inertias, marker="o")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("📉 Elbow Method — Choosing k")
plt.xticks(range(1, 11))
plt.grid(alpha=0.25)
plt.show()

# Styled result message
display(HTML("""
<div style="
    background:linear-gradient(135deg,#0F172A,#164E63);
    border-radius:12px;
    padding:15px 18px;
    color:white;
    margin-top:10px;
">
<div style="color:#67E8F9;font-size:11px;font-weight:800;letter-spacing:1.5px;">
ELBOW METHOD
</div>
<div style="font-size:15px;margin-top:6px;">
📉 The elbow point represents a value of <b>k</b> where increasing the number
of clusters provides diminishing improvement.
</div>
</div>
"""))
```

<div style="
    background:#F0FDFA;
    border-left:6px solid #0891B2;
    border-radius:12px;
    padding:18px;
    margin:15px 0;
">

<div style="color:#0E7490;font-size:12px;font-weight:800;letter-spacing:1.5px;">
STEP 6
</div>

<div style="color:#0F172A;font-size:22px;font-weight:800;margin-top:5px;">
📐 Silhouette Score
</div>

<div style="color:#334155;font-size:14px;margin-top:8px;line-height:1.6;">
The silhouette score gives us a quantitative way to evaluate how well-separated
the clusters are.
<br><br>
The score ranges from <b>-1 to +1</b>. Higher values generally indicate better-defined clusters.
</div>

</div>

```python
# Calculate silhouette scores for candidate k values
silhouette_scores = {}

for k in range(2, 11):
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores[k] = score

silhouette_df = pd.DataFrame({
    "k": list(silhouette_scores.keys()),
    "Silhouette Score": list(silhouette_scores.values())
})

display(
    silhouette_df.style
    .set_caption("📐 Silhouette Scores by Number of Clusters")
    .format({"Silhouette Score": "{:.3f}"})
    .set_properties(**{
        "text-align": "center",
        "padding": "8px"
    })
    .set_table_styles([
        {
            "selector": "caption",
            "props": [
                ("background-color", "#0F172A"),
                ("color", "#67E8F9"),
                ("font-size", "17px"),
                ("font-weight", "800"),
                ("text-align", "left"),
                ("padding", "11px")
            ]
        },
        {
            "selector": "th",
            "props": [
                ("background-color", "#164E63"),
                ("color", "white")
            ]
        }
    ])
)

best_k = max(silhouette_scores, key=silhouette_scores.get)
best_score = silhouette_scores[best_k]

display(HTML(f"""
<div style="
    background:linear-gradient(135deg,#0F172A,#164E63);
    border-radius:14px;
    padding:17px 20px;
    margin-top:12px;
    color:white;
">
<div style="color:#67E8F9;font-size:11px;font-weight:800;letter-spacing:1.5px;">
BEST SILHOUETTE RESULT
</div>

<div style="font-size:24px;font-weight:800;margin-top:5px;">
k = {best_k}
</div>

<div style="font-size:16px;color:#CFFAFE;margin-top:4px;">
Silhouette Score = {best_score:.3f}
</div>
</div>
"""))
```

    background:#F0FDFA;
    border-left:6px solid #0891B2;
    border-radius:12px;

STEP 7
</div>

<div style="color:#0F172A;font-size:22px;font-weight:800;margin-top:5px;">
🎯 Fit the Final K-Means Model
</div>

<div style="color:#334155;font-size:14px;margin-top:8px;">
Using the selected number of clusters, we train the final K-Means model and assign
each customer to a cluster.
</div>

</div>

```python
# Fit final K-Means model
kmeans = KMeans(n_clusters=best_k, random_state=42, n_init=10)

df["Cluster"] = kmeans.fit_predict(X_scaled)

# Display cluster counts
cluster_counts = (
    df["Cluster"]
    .value_counts()
    .sort_index()
    .rename("Customers")
    .to_frame()
)

display(
    cluster_counts.style
    .set_caption("👥 Customers per Cluster")
    .set_properties(**{
        "text-align": "center",
        "padding": "8px"
    })
    .set_table_styles([
        {
            "selector": "caption",
            "props": [
                ("background-color", "#0F172A"),
                ("color", "#67E8F9"),
                ("font-size", "17px"),
                ("font-weight", "800"),
                ("text-align", "left"),
                ("padding", "11px")
            ]
        },
        {
            "selector": "th",
            "props": [
                ("background-color", "#164E63"),
                ("color", "white")
            ]
        }
    ])
)

display(HTML(f"""
<div style="
    background:#ECFEFF;
    border:1px solid #A5F3FC;
    border-radius:12px;
    padding:14px 18px;
    margin-top:12px;
">
<b style="color:#0E7490;">✅ CLUSTERING COMPLETE</b><br>
<b>{best_k}</b> customer groups were created using K-Means.
</div>
"""))
```
