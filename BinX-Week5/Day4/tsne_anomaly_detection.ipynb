💡 <strong>In simple terms:</strong>
PCA gives us a broader view of the overall structure, while t-SNE zooms in

visualizations can reveal different patterns in the same dataset.

```python
from sklearn.manifold import TSNE

# Sample 10,000 transactions for faster visualization
df_tsne = df.sample(n=10000, random_state=42)

# Separate features from the target
X_tsne_input = df_tsne.drop(columns=["Class"])

# Apply t-SNE
tsne = TSNE(
    n_components=2,
    perplexity=30,
    random_state=42
)

X_tsne = tsne.fit_transform(X_tsne_input)

print("t-SNE shape:", X_tsne.shape)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    X_tsne[:, 0],
    X_tsne[:, 1],
    c=df_tsne["Class"],
    cmap="coolwarm",
    alpha=0.6,
    s=8
)

plt.title("t-SNE Visualization of Credit Card Transactions", fontsize=16, fontweight="bold")
plt.xlabel("t-SNE Dimension 1")
plt.ylabel("t-SNE Dimension 2")

plt.colorbar(scatter, label="Class")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

# Create a DataFrame with the t-SNE coordinates and class labels
tsne_results = pd.DataFrame({
    "TSNE_1": X_tsne[:, 0],
    "TSNE_2": X_tsne[:, 1],
    "Class": df_tsne["Class"].values
})

# Show basic statistics
print("Overall t-SNE statistics:")
print(tsne_results[["TSNE_1", "TSNE_2"]].describe())

print("\nClass distribution:")
print(tsne_results["Class"].value_counts())

print("\nClass-wise t-SNE statistics:")
print(
    tsne_results.groupby("Class")[["TSNE_1", "TSNE_2"]].agg(
        ["mean", "std", "min", "max"]
    )
)
```

🔍 What Makes a Point an Anomaly?
</div>

<div style="
    font-size:14px;
    line-height:1.85;

An <strong style="color:#B65F45;">anomaly</strong> is an observation that
does not follow the general behavior of the majority of the data. It may be
very different in its feature values or isolated from the normal observations.

<br><br>

Anomaly detection is often considered an
<strong style="color:#B65F45;">unsupervised learning</strong> task because
anomalies are usually rare and difficult to label in advance. Instead of
learning from predefined classes, the algorithm learns the general structure
of the data and identifies observations that appear unusual.

">
🔪 How Does Isolation Forest Work?
</div>

<div style="
    font-size:14px;
    line-height:1.85;
    color:#665E62;
">
The main idea behind Isolation Forest is simple:
<strong style="color:#B65F45;">anomalies are easier to isolate than normal
observations.</strong>

<br><br>

The algorithm randomly selects features and random split values to partition
the data. An unusual observation that is far away from the majority of the
data usually becomes isolated after only a few splits.

<br><br>

Normal observations tend to be part of dense groups, so they require more
splits to isolate. Therefore, observations that are isolated quickly receive
higher anomaly scores.

</div>

<div style="
    margin-top:18px;
    padding:14px 17px;
    background:#F8F1EE;
    border-radius:13px;
    font-size:13px;
    line-height:1.7;
    color:#6F5559;
">

💡 <strong>Key idea:</strong>
<strong>Easy to isolate → likely anomaly</strong> &nbsp; | &nbsp;
<strong>Hard to isolate → likely normal</strong>

</div>

</div>

<div style="
    width:100%;
    box-sizing:border-box;
    background:#FFFDFB;
    border:1px solid #E7DDD7;

<div style="

">
🎚️ Understanding <code style="
    background:#F3EBDD;
    color:#B65F45;
    padding:3px 7px;
    border-radius:6px;
">contamination</code>
</div>

<div style="
    font-size:14px;
    line-height:1.85;
    color:#665E62;
">
The <strong style="color:#B65F45;">contamination</strong> parameter represents
our estimated proportion of anomalies in the dataset. It helps Isolation
Forest determine the threshold used to classify observations as normal or
anomalous.

<br><br>

In this project, we use:

</div>

<div style="
    margin:16px 0;
    padding:16px;
    background:#F8E9E6;
    border-radius:13px;
    text-align:center;
    font-size:22px;
    font-weight:900;
    color:#B65F45;
">
contamination = 0.05
</div>

<div style="
    font-size:14px;
    line-height:1.8;
    color:#665E62;
">
This means we estimate that approximately <strong>5%</strong> of the
observations may be anomalous. The value is an assumption used by the
algorithm and does not mean that exactly 5% of the transactions are truly
fraudulent.
</div>

</div>

```python
from sklearn.ensemble import IsolationForest
```

```python
iso = IsolationForest(
    contamination=0.05,
    random_state=42
)

print("Isolation Forest model created successfully.")
```

```python
preds = iso.fit_predict(X_tsne_input)

print("Anomaly detection completed.")
print("Prediction shape:", preds.shape)
```

```python
normal_count = (preds == 1).sum()
anomaly_count = (preds == -1).sum()

print("Normal observations:", normal_count)
print("Anomalous observations:", anomaly_count)
```
