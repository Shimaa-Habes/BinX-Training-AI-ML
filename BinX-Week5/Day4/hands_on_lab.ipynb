🌀 Apply t-SNE

<div style="
    margin-top:12px;
    font-size:14px;
    line-height:1.8;
    color:#665E62;
">
We will reduce the high-dimensional credit card transaction data to two
dimensions using t-SNE. A sample of the dataset will be used to make the
visualization faster while preserving enough observations to reveal local
patterns.
</div>

</div>

```python
import pandas as pd

# Load the credit card fraud dataset
df = pd.read_csv("creditcard.csv")

print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())
```

```python
# Sample 10,000 transactions for the t-SNE lab
df_lab = df.sample(n=10000, random_state=42)

# Separate features from the target
X_lab = df_lab.drop(columns=["Class"])

print("Sample shape:", X_lab.shape)
```

```python
from sklearn.manifold import TSNE

tsne = TSNE(
    n_components=2,
    perplexity=30,
    random_state=42
)

X_tsne_lab = tsne.fit_transform(X_lab)

print("t-SNE shape:", X_tsne_lab.shape)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    X_tsne_lab[:, 0],
    X_tsne_lab[:, 1],
    c=df_lab["Class"],
    cmap="coolwarm",
    alpha=0.5,
    s=8
)

plt.title(
    "t-SNE Visualization of Credit Card Transactions",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("t-SNE Dimension 1")
plt.ylabel("t-SNE Dimension 2")

plt.colorbar(scatter, label="Class")

plt.tight_layout()
plt.show()
```

<div style="

Both PCA and t-SNE reduce high-dimensional data to two dimensions, but they
preserve different types of structure. We will apply PCA to the same sample
and compare its representation with the t-SNE visualization.
</div>

</div>

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

X_pca_lab = pca.fit_transform(X_lab)

print("PCA shape:", X_pca_lab.shape)
```

```python
plt.figure(figsize=(10, 7))

scatter = plt.scatter(
    X_pca_lab[:, 0],
    X_pca_lab[:, 1],
    c=df_lab["Class"],
    cmap="coolwarm",
    alpha=0.5,
    s=8
)

plt.title(
    "PCA Visualization of Credit Card Transactions",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")

plt.colorbar(scatter, label="Class")

plt.tight_layout()
plt.show()
```

PCA provides a more global view of the transaction data by preserving the

🌲 Detect Anomalies with Isolation Forest

<div style="

">
We will now apply Isolation Forest to the same 10,000 transactions. The
algorithm will identify observations that appear significantly different
from the majority of the dataset.
</div>

</div>

```python
from sklearn.ensemble import IsolationForest

iso = IsolationForest(
    contamination=0.05,
    random_state=42
)

predictions = iso.fit_predict(X_lab)

print("Anomaly detection completed.")
```

```python
normal_count = (predictions == 1).sum()
anomaly_count = (predictions == -1).sum()

print("Normal observations:", normal_count)
print("Anomalous observations:", anomaly_count)
```

```python
import matplotlib.pyplot as plt

# Identify normal and anomalous observations
normal = predictions == 1
anomalous = predictions == -1

plt.figure(figsize=(10, 7))

plt.scatter(
    X_tsne_lab[normal, 0],
    X_tsne_lab[normal, 1],
    alpha=0.35,
    s=8,
    label="Normal"
)

plt.scatter(
    X_tsne_lab[anomalous, 0],
    X_tsne_lab[anomalous, 1],
    alpha=0.8,
    s=18,
    label="Anomaly"
)

plt.title(
    "Normal vs. Anomalous Transactions — t-SNE",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("t-SNE Dimension 1")
plt.ylabel("t-SNE Dimension 2")

plt.legend()
plt.tight_layout()
plt.show()
```
