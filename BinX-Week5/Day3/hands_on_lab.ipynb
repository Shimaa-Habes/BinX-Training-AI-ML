👕 Step 1 — Load & Prepare the Dataset
</div>

<div style="
    font-size: 16px;
    margin-top: 12px;
    line-height: 1.7;
    color: #7C2D12;
">
Fashion-MNIST contains 70,000 grayscale images of clothing items.
Each image has a size of 28 × 28 pixels, giving us 784 pixel features.
</div>

<div style="
    margin-top: 18px;
    background: #FFEDD5;
    border-left: 5px solid #F97316;
    padding: 14px 18px;
    border-radius: 10px;
    font-size: 14px;
    line-height: 1.6;
">
🎯 <b>Goal:</b> Load the Fashion-MNIST training data, separate the
pixel features from the labels, and prepare the data for PCA.
</div>

</div>


```python
import pandas as pd
import numpy as np

# Load Fashion-MNIST training dataset
df = pd.read_csv("fashion-mnist_train.csv")

# Display basic information
print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())
```

```python
# Separate labels from pixel features

y = df["label"]
X = df.drop("label", axis=1)

print("Features shape:", X.shape)
print("Labels shape:", y.shape)
```

<div style="
    background: linear-gradient(135deg, #FFF7ED, #FFEDD5);
    border-radius: 20px;
    padding: 28px;
    color: #431407;

">
PCA PREPROCESSING
</div>

<div style="
    font-size: 28px;
    font-weight: 800;
    margin-top: 10px;
">
📏 Step 2 — Scale the Features
</div>

<div style="
    font-size: 16px;
    margin-top: 12px;
    line-height: 1.7;
    color: #7C2D12;
">
PCA is based on variance. Since pixel values range from 0 to 255,
we standardize the features before applying PCA so that the scale
of the features does not unfairly affect the principal components.
</div>

<div style="
    margin-top: 18px;
    background: #FFEDD5;
    border-left: 5px solid #F97316;
    padding: 14px 18px;
    border-radius: 10px;
    font-size: 14px;
">
💡 <b>StandardScaler</b> transforms the features to have a mean close
to 0 and a standard deviation close to 1.
</div>

</div>


```python
from sklearn.preprocessing import StandardScaler

# Standardize the pixel features
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Original shape:", X.shape)
print("Scaled shape:", X_scaled.shape)
```

<div style="
    background: linear-gradient(135deg, #FFF7ED, #FFEDD5);
    border-radius: 20px;
    padding: 28px;
    color: #431407;

">
DIMENSIONALITY REDUCTION
</div>

<div style="
    font-size: 28px;
    font-weight: 800;
    margin-top: 10px;
">
🧠 Step 3 — Apply Principal Component Analysis
</div>

<div style="
    font-size: 16px;
    margin-top: 12px;
    line-height: 1.7;
    color: #7C2D12;
">
The original Fashion-MNIST data contains 784 dimensions.
PCA finds new directions called principal components that capture
the greatest amount of variance in the dataset.
</div>

<div style="
    margin-top: 18px;
    background: #FFEDD5;
    border-left: 5px solid #F97316;
    padding: 14px 18px;
    border-radius: 10px;
    font-size: 14px;
">
🎯 <b>For this step:</b> We reduce the dataset to 2 principal
components so that we can visualize it in two dimensions.
</div>

</div>


```python
from sklearn.decomposition import PCA

# Reduce the data to 2 principal components
pca_2d = PCA(n_components=2)

X_pca_2d = pca_2d.fit_transform(X_scaled)

print("Original shape:", X.shape)
print("PCA shape:", X_pca_2d.shape)
```

<div style="
    background: linear-gradient(135deg, #FFF7ED, #FFEDD5);
    border-radius: 20px;
    padding: 28px;
    color: #431407;

">
PCA • EXPLAINED VARIANCE
</div>

<div style="
    font-size: 28px;
    font-weight: 800;
    margin-top: 10px;
">
📊 Step 4 — How Much Information Did We Keep?
</div>

<div style="
    font-size: 16px;
    margin-top: 12px;
    line-height: 1.7;
    color: #7C2D12;
">
The explained variance ratio tells us how much of the original
dataset's variance is captured by each principal component.
By adding these values together, we can measure how much information
was retained after dimensionality reduction.
</div>

<div style="
    margin-top: 18px;
    background: #FFEDD5;
    border-left: 5px solid #F97316;
    padding: 14px 18px;
    border-radius: 10px;
    font-size: 14px;
    line-height: 1.6;
">
💡 <b>Important:</b> The first principal component captures the
largest amount of variance, the second captures the next largest,
and so on.
</div>

</div>


```python
# Explained variance for the two PCA components

explained_variance = pca_2d.explained_variance_ratio_

print("Explained variance ratio:")
print(explained_variance)

print("\nTotal variance retained:")
print(explained_variance.sum())
```

<div style="
    background: linear-gradient(135deg, #FFF7ED, #FFEDD5);

PCA • COMPONENT SELECTION
</div>

<div style="
    font-size: 28px;
    font-weight: 800;
    margin-top: 10px;
">
🎯 Step 5 — Choosing the Number of Components
</div>

<div style="
    font-size: 16px;
    margin-top: 12px;
    line-height: 1.7;
    color: #7C2D12;
">
Using only two components is useful for visualization, but it may
discard a large amount of information. Therefore, we fit PCA using
all available components and calculate the cumulative explained
variance.
</div>

<div style="
    margin-top: 18px;
    background: #FFEDD5;
    border-left: 5px solid #F97316;
    padding: 14px 18px;
    border-radius: 10px;
    font-size: 14px;
    line-height: 1.6;
">
🎯 <b>Target:</b> Find the smallest number of components that retains
approximately <b>95%</b> of the original variance.
</div>

</div>


```python
# Fit PCA with all available components

pca_full = PCA()

X_pca_full = pca_full.fit_transform(X_scaled)

# Calculate cumulative explained variance
cumulative_variance = np.cumsum(
    pca_full.explained_variance_ratio_
)

# Find the number of components needed to retain 95% variance
n_components_95 = np.argmax(
    cumulative_variance >= 0.95
) + 1

print("Original number of features:", X.shape[1])
print("Components needed for 95% variance:", n_components_95)
print(
    "Variance retained:",
    cumulative_variance[n_components_95 - 1]
)
```

<div style="

📈 Step 6 — Cumulative Explained Variance
</div>

<div style="

">

🔎 The point where the curve reaches <b>95%</b> tells us the number
of components required to preserve most of the dataset's variance.
</div>

</div>


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance
)

plt.axhline(
    y=0.95,
    linestyle="--",
    label="95% Variance"
)

plt.axvline(
    x=n_components_95,
    linestyle="--",
    label=f"{n_components_95} Components"
)

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Fashion-MNIST — PCA Cumulative Explained Variance")

plt.legend()
plt.grid(True)

plt.show()
```

🧭 Step 7 — 2D PCA Visualization

<p style="font-size: 16px; line-height: 1.7; color: #7C2D12;">
To visualize the high-dimensional Fashion-MNIST dataset, we reduce the
original <b>784 features to only 2 principal components</b>.

</p>

<div style="
    background: #FFEDD5;

🎯 <b>Goal:</b> Create a 2D visualization and use the clothing labels

```python
# Reduce the scaled data to 2 principal components
pca_2d = PCA(n_components=2)

X_pca_2d = pca_2d.fit_transform(X_scaled)

print("Original number of features:", X_scaled.shape[1])
print("Reduced number of components:", X_pca_2d.shape[1])
print("Shape of reduced data:", X_pca_2d.shape)
```

```python
plt.figure(figsize=(12, 8))

scatter = plt.scatter(
    X_pca_2d[:, 0],
    X_pca_2d[:, 1],
    c=y,
    alpha=0.5,
    s=8,
    cmap="tab10"
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("Fashion-MNIST — PCA 2D Visualization")

plt.colorbar(
    scatter,
    label="Clothing Class"
)

plt.grid(alpha=0.2)
plt.show()
```
