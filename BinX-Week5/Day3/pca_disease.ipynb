📐 3.1 — The Curse of Dimensionality
</div>

<p style="font-size: 16px; line-height: 1.8;">
As the number of features increases, machine learning models have to work
with a much larger and more complex feature space.
This can make the data more difficult to analyze and visualize.
</p>

<div style="
    background: #FFFFFF;
    border-radius: 14px;
    padding: 16px 20px;
    border-left: 5px solid #E11D48;
">

<b style="color:#BE123C;">⚠️ Why is high dimensionality a problem?</b>

<ul style="font-size:15px; line-height:1.9;">
    <li>Data becomes more sparse.</li>
    <li>Distances between observations become less meaningful.</li>
    <li>Models can become more prone to overfitting.</li>
    <li>Visualizing data becomes difficult when there are many features.</li>
    <li>Models may require more computational resources.</li>
</ul>

</div>

<div style="
    background: #FFE4E6;
    padding: 14px 18px;
    border-radius: 12px;
    margin-top: 16px;
    font-size: 15px;
    line-height: 1.7;
">

💡 <b>Dimensionality Reduction</b> helps by transforming a large number
of features into a smaller set of informative components while attempting
to preserve the important structure and variation in the data.

</div>

</div>

🧠 3.2 — What PCA Does
</div>

<p style="font-size: 16px; line-height: 1.8;">
<b>Principal Component Analysis (PCA)</b> transforms the original features
into new features called <b>Principal Components</b>.
These components are designed to capture the directions of greatest
variance in the data.
</p>

<div style="
    background: #FFFFFF;
    border-radius: 14px;
    padding: 16px 20px;
    border-left: 5px solid #E11D48;
">

<b style="color:#BE123C;">🔎 How it works</b>

<ul style="font-size:15px; line-height:1.9;">
    <li><b>PC1</b> captures the largest amount of variance.</li>
    <li><b>PC2</b> captures the next largest amount of variance.</li>
    <li>Each component is a combination of the original features.</li>
    <li>Keeping fewer components reduces the dimensionality of the dataset.</li>
</ul>

</div>

<div style="
    background: #FFE4E6;
    padding: 14px 18px;
    border-radius: 12px;
    margin-top: 16px;
    font-size: 15px;
    line-height: 1.7;
">

🎯 <b>Important:</b> PCA is variance-based, so the features should be
scaled before applying PCA. Otherwise, features with larger numerical
ranges could dominate the principal components.

</div>

</div>


```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer

# Load the Heart Disease dataset
df = pd.read_csv("heart_disease_uci.csv")

# Separate features and target
X = df.drop(columns=["id", "num"])
y = df["num"]

# Encode categorical features
X = pd.get_dummies(X, drop_first=True)

# Convert all features to numeric values
X = X.astype(float)

# Check missing values
print("Missing values before imputation:")
print(X.isnull().sum().sum())

# Impute missing values using the median
imputer = SimpleImputer(strategy="median")
X_imputed = imputer.fit_transform(X)

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)

print("\nOriginal feature count:", X.shape[1])
print("Scaled data shape:", X_scaled.shape)
print("Missing values after imputation:", np.isnan(X_scaled).sum())
```

```python
from sklearn.decomposition import PCA
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
import pandas as pd
import numpy as np

# Make sure X is numeric
X_numeric = X.copy()

# Replace infinite values with NaN
X_numeric = X_numeric.replace([np.inf, -np.inf], np.nan)

# Remove columns that are completely empty
X_numeric = X_numeric.dropna(axis=1, how="all")

# Impute remaining missing values
imputer = SimpleImputer(strategy="median")
X_imputed = imputer.fit_transform(X_numeric)

# Scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_imputed)

# Verify that there are no NaN values
print("Missing values after preprocessing:", np.isnan(X_scaled).sum())

# Apply PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Original number of features:", X_scaled.shape[1])
print("Reduced number of components:", X_pca.shape[1])
```

📊 3.3 — Explained Variance
</div>

<p style="font-size: 16px; line-height: 1.8;">
The <b>explained variance ratio</b> tells us how much of the total
variance in the original dataset is captured by each principal component.
</p>

<div style="
    background: #FFFFFF;
    border-radius: 14px;
    padding: 16px 20px;
    border-left: 5px solid #E11D48;
">

<b style="color:#BE123C;">📌 Key Idea</b>

<ul style="font-size:15px; line-height:1.9;">
    <li>Each principal component explains a portion of the total variance.</li>
    <li>Adding the components gives the cumulative variance retained.</li>
    <li>A common target is to retain approximately <b>95%</b> of the variance.</li>
</ul>

</div>

🔍 3.4 — When to Use PCA
</div>

<p style="font-size: 16px; line-height: 1.8; color: #7C2D12;">
PCA is useful when a dataset contains a large number of features or when many features are correlated.
It transforms the original features into a smaller set of principal components while preserving most of the important variation in the data.
</p>

<div style="
    background: #FFFFFF;
    padding: 16px 20px;
    border-radius: 12px;
    margin-top: 14px;
    border-left: 5px solid #F97316;
">

<b style="color: #C2410C;">✅ PCA can be used to:</b>

<ul style="font-size: 15px; line-height: 1.8; color: #7C2D12;">
<li>Reduce the number of features and make the dataset smaller.</li>
<li>Speed up machine learning models by using fewer dimensions.</li>
<li>Reduce redundant or highly correlated features.</li>
<li>Make high-dimensional data easier to visualize in 2D or 3D.</li>
</ul>

</div>

<div style="
    background: #FFF7ED;
    padding: 16px 20px;
    border-radius: 12px;
    margin-top: 14px;
    border-left: 5px solid #EA580C;
">

<b style="color: #C2410C;">⚠️ Trade-off:</b>

<p style="font-size: 15px; line-height: 1.7; color: #7C2D12;">
The new principal components are combinations of the original features.
Therefore, PCA can make the data less interpretable because each component no longer represents one original feature directly.
</p>

</div>

📊 Step 2 — Cumulative Explained Variance
</div>

<p style="
    font-size: 16px;
    line-height: 1.8;
    color: #7C2D12;
">
Now that the features have been standardized, we apply PCA to all available
components and calculate how much variance is retained as the number of
components increases.
</p>

<div style="
    background: #FFFFFF;
    padding: 16px 20px;
    border-radius: 12px;
    border-left: 5px solid #F97316;

<b style="color: #C2410C;">🎯 What are we looking for?</b>

<p style="font-size: 15px; line-height: 1.7; color: #7C2D12;">
We will use the cumulative explained variance to determine how many principal
components are needed to preserve approximately <b>95% of the information</b>
in the original dataset.
</p>

</div>

</div>


```python
from sklearn.decomposition import PCA
import numpy as np
import matplotlib.pyplot as plt

# Fit PCA using all available components
pca_full = PCA()
X_pca_full = pca_full.fit_transform(X_scaled)

# Calculate cumulative explained variance
cumulative_variance = np.cumsum(
    pca_full.explained_variance_ratio_
)

# Plot cumulative explained variance
plt.figure(figsize=(10, 6))

plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker="o",
    markersize=3
)

plt.axhline(
    y=0.95,
    linestyle="--",
    label="95% Variance"
)

plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA — Cumulative Explained Variance")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

🎯 Step 3 — Choosing the Components
</div>

<p style="
    font-size: 16px;
    line-height: 1.8;
    color: #7C2D12;
">
The cumulative explained variance curve helps us identify the smallest number

</p>

<div style="
    background: #FFFFFF;
    padding: 16px 20px;

📌 <b style="color: #C2410C;">Decision Rule:</b>
Choose the first component count where cumulative explained variance reaches
or exceeds <b>95%</b>.
</div>

</div>


```python
# Find the number of components needed to retain 95% variance
n_components_95 = np.argmax(cumulative_variance >= 0.95) + 1

variance_95 = cumulative_variance[n_components_95 - 1]

print("Original number of features:", X_scaled.shape[1])
print("Components needed for 95% variance:", n_components_95)
print("Variance retained:", variance_95)
```
