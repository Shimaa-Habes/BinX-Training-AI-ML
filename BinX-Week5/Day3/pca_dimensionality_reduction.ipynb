">
📐 3.1 — The Curse of Dimensionality
</h2>

<p style="
    color: #596579;
    font-size: 15px;
    line-height: 1.9;
">
When a dataset contains a very large number of features, it becomes
<strong>high-dimensional</strong>. Fashion-MNIST is a good example because
each image is represented by <strong>784 pixel features</strong>.
</p>

<p style="
    color: #596579;
    font-size: 15px;
    line-height: 1.9;
">
As the number of dimensions increases, several problems can appear:
</p>

<div style="
    display: grid;
    grid-template-columns: repeat(2, 1fr);
    gap: 12px;
    margin-top: 18px;
">

<div style="
    background:#FFFFFF;
    padding:16px;
    border-radius:12px;
    border:1px solid #D5EAE6;
">
<b style="color:#287C78;">🌫️ Sparse Data</b>
<p style="color:#6B7280;font-size:13px;">
Data points become more spread out in a high-dimensional space.
</p>
</div>

<div style="
    background:#FFFFFF;
    padding:16px;
    border-radius:12px;
    border:1px solid #D5EAE6;
">
<b style="color:#6B6680;">📏 Distance Problems</b>
<p style="color:#6B7280;font-size:13px;">
Distances between data points become less informative as dimensions increase.
</p>
</div>

<div style="
    background:#FFFFFF;
    padding:16px;
    border-radius:12px;
    border:1px solid #D5EAE6;
">
<b style="color:#E0A72E;">⚠️ Overfitting</b>
<p style="color:#6B7280;font-size:13px;">
Models may learn unnecessary patterns from many redundant features.
</p>
</div>

<div style="
    background:#FFFFFF;
    padding:16px;
    border-radius:12px;
    border:1px solid #D5EAE6;
">
<b style="color:#287C78;">👁️ Visualization</b>
<p style="color:#6B7280;font-size:13px;">
We cannot directly visualize data with hundreds of dimensions.
</p>
</div>

</div>

<div style="
    background:#FFF4CC;
    border-radius:12px;
    padding:16px 20px;
    margin-top:20px;
    color:#66541A;
    line-height:1.7;
">
💡 <strong>Key Idea:</strong>
Dimensionality reduction helps us compress a large number of features
into fewer dimensions while preserving the most important information.
</div>

</div>

<div style="
    background: linear-gradient(120deg, #F4F1FA, #E8F6F3);

">
🧭 3.2 — What PCA Does
</h2>

<p style="
    color:#596579;
    font-size:15px;
    line-height:1.9;
">
<strong>Principal Component Analysis (PCA)</strong> is a dimensionality
reduction technique that creates new features called
<strong>principal components</strong>.
</p>

<p style="
    color:#596579;
    font-size:15px;
    line-height:1.9;
">
Instead of keeping the original 784 pixel features separately, PCA looks
for directions in the data where the greatest amount of
<strong>variation</strong> occurs.
</p>

<div style="
    background:#FFFFFF;
    border-radius:14px;
    padding:20px;
    margin:20px 0;
    border:1px solid #DCD5EA;
">

<div style="text-align:center;font-size:17px;font-weight:800;color:#287C78;">
Original Features
</div>

<div style="text-align:center;font-size:25px;color:#E0A72E;margin:8px;">
784 pixels
&nbsp; → &nbsp;
PCA
&nbsp; → &nbsp;
Fewer Components
</div>

<div style="text-align:center;font-size:14px;color:#6B7280;">
while preserving important variation in the dataset
</div>

</div>

<p style="
    color:#596579;
    font-size:15px;
    line-height:1.9;
">
The first principal component captures the <strong>largest amount of
variance</strong>. The second captures the next largest amount, and so on.
Each component is a combination of the original features.
</p>

<div style="
    background:#FFF4CC;
    border-radius:12px;
    padding:15px 18px;
    color:#66541A;
">
💡 <strong>Important:</strong> PCA does not simply delete random features.
It creates new combinations of the original features that capture the
most important patterns in the data.
</div>

</div>

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
train_df = pd.read_csv("fashion-mnist_train.csv")
test_df = pd.read_csv("fashion-mnist_test.csv")

print("Training shape:", train_df.shape)
print("Testing shape:", test_df.shape)
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
# Separate features from labels
X = train_df.drop("label", axis=1)
y = train_df["label"]

print("Features shape:", X.shape)
print("Labels shape:", y.shape)
```

```python
# Scale the data before applying PCA
X_scaled = StandardScaler().fit_transform(X)

# Apply PCA and reduce the data to 2 dimensions
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Original shape:", X.shape)
print("Reduced shape:", X_pca.shape)
```

">
⚖️ 3.3 — Why Do We Scale Before PCA?
</h2>

<p style="
    color:#596579;
    font-size:15px;
    line-height:1.9;
">
PCA is based on <strong>variance</strong>. If different features have very
different scales, features with larger numerical ranges can have a much
greater influence on the principal components.
</p>

<p style="
    color:#596579;
    font-size:15px;
    line-height:1.9;
">
For this reason, we normally apply <strong>StandardScaler</strong> before
PCA so that the features are placed on a comparable scale.
</p>

<div style="
    background:#FFFFFF;
    border-radius:13px;
    padding:18px;
    margin-top:18px;
    text-align:center;
    border:1px solid #CDE6E2;
">

<div style="font-size:16px;font-weight:800;color:#287C78;">
StandardScaler
</div>

<div style="font-size:22px;color:#E0A72E;margin:8px 0;">
Original Data → Scaled Data → PCA
</div>

<div style="font-size:13px;color:#6B7280;">
Scaling helps PCA treat the features more fairly.
</div>

</div>

</div>

">
📊 3.4 — Explained Variance Ratio
</h2>

<p style="
    color:#596579;
    font-size:15px;
    line-height:1.9;
">
After applying PCA, we need to know how much information each principal
component preserves. This is measured using the
<strong>explained variance ratio</strong>.
</p>

<p style="
    color:#596579;
    font-size:15px;
    line-height:1.9;
">
For example, if the first component explains <strong>30%</strong> of the
variance and the second explains <strong>20%</strong>, keeping both
components preserves approximately <strong>50%</strong> of the total variance.
</p>

<div style="
    background:#FFFFFF;
    border-radius:13px;
    padding:18px;
    border-left:5px solid #E0A72E;
    margin-top:18px;
">

<p style="color:#596579;line-height:1.7;margin:0;">
<strong>Explained Variance Ratio</strong> → information retained by each component
<br><br>
<strong>Cumulative Explained Variance</strong> → total information retained
by all components up to that point
</p>

</div>

<div style="
    background:#E8F6F3;
    border-radius:12px;
    padding:15px 18px;
    margin-top:18px;
    color:#245E5B;
">
🎯 A common goal is to choose enough components to preserve approximately
<strong>95% of the total variance</strong>.
</div>

</div>

    font-size:27px;
">
🎯 3.5 — When Should We Use PCA?
</h2>

<p style="
    color:#596579;
    font-size:15px;
    line-height:1.9;
">
PCA is useful when a dataset contains many features and we want a more
compact representation of the data.
</p>

<ul style="
    color:#596579;
    font-size:15px;
    line-height:2;
">
<li><strong>Reduce computational cost:</strong> fewer features can make some models faster.</li>
<li><strong>Remove redundancy:</strong> correlated information can be represented using fewer components.</li>
<li><strong>Reduce overfitting:</strong> removing unnecessary dimensions can sometimes improve generalization.</li>
<li><strong>Visualization:</strong> high-dimensional data can be reduced to 2D or 3D for plotting.</li>
</ul>

<div style="
    background:#FFF4CC;
    border-radius:12px;
    padding:16px 20px;
    color:#66541A;
    margin-top:18px;
">
⚠️ <strong>Trade-off:</strong>
Principal components are combinations of the original features, so they are
usually less interpretable than the original columns.
</div>
