">
🔵 DBSCAN Clustering
</h2>

<p style="
    color: #E9D5FF;
    font-size: 18px;
    line-height: 1.9;
">
DBSCAN stands for <strong style="color:#C4B5FD;">
Density-Based Spatial Clustering of Applications with Noise
</strong>.
Instead of requiring us to specify the number of clusters, DBSCAN groups points
based on their density.
</p>

<ul style="
    color: #E9D5FF;
    font-size: 18px;
    line-height: 2;
">

<li>
📏 <strong style="color:#C4B5FD;">eps</strong> — defines the maximum distance
between neighboring points.
</li>

<li>
🔢 <strong style="color:#C4B5FD;">min_samples</strong> — defines the minimum
number of nearby points required to form a dense region.
</li>

<li>
⚠️ Points that do not belong to a dense region are marked as
<strong style="color:#F0ABFC;">noise (-1)</strong>.
</li>

</ul>

<p style="
    color: #E9D5FF;
    font-size: 18px;
    line-height: 1.9;
">
DBSCAN is particularly useful for our dataset because the Moon clusters have a
<strong style="color:#C4B5FD;">curved, non-linear shape</strong>.
</p>

</div>

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(
    eps=0.15,
    min_samples=5
)

dbscan_labels = dbscan.fit_predict(X)

print("DBSCAN labels:")
print(pd.Series(dbscan_labels).value_counts().sort_index())
```

```python
n_clusters = len(set(dbscan_labels)) - (1 if -1 in dbscan_labels else 0)
n_noise = list(dbscan_labels).count(-1)

print("Number of clusters:", n_clusters)
print("Number of noise points:", n_noise)
```

```python
plt.figure(figsize=(10, 6))

scatter = plt.scatter(
    X["X1"],
    X["X2"],
    c=dbscan_labels,
    s=35,
    alpha=0.8
)

plt.title("DBSCAN Clustering", fontsize=16, fontweight="bold")
plt.xlabel("X1")
plt.ylabel("X2")
plt.grid(alpha=0.2)

plt.show()
```

">
🌳 Hierarchical Clustering
</h2>

<p style="
    color: #E9D5FF;
    font-size: 18px;
    line-height: 1.9;
">
Hierarchical clustering builds a hierarchy of clusters by repeatedly merging
the closest groups of data points.
</p>

<p style="
    color: #E9D5FF;
    font-size: 18px;
    line-height: 1.9;
">
The result is represented using a <strong style="color:#C4B5FD;">
dendrogram</strong>, which allows us to visualize how the observations are
grouped together at different distance levels.
</p>

<ul style="
    color: #E9D5FF;
    font-size: 18px;
    line-height: 2;
">

<li>🌱 Each point starts as its own cluster.</li>

<li>🔗 The closest clusters are progressively merged.</li>

<li>📏 The vertical height represents the distance between merged clusters.</li>

<li>✂️ Cutting the dendrogram at a chosen height determines the final clusters.</li>

</ul>

</div>

```python
from scipy.cluster.hierarchy import dendrogram, linkage

# Build the hierarchical clustering structure
Z = linkage(X, method="ward")

plt.figure(figsize=(12, 6))

dendrogram(
    Z,
    truncate_mode="lastp",
    p=30
)

plt.title(
    "Hierarchical Clustering Dendrogram",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Cluster / Data Points")
plt.ylabel("Distance")

plt.grid(alpha=0.2)
plt.show()
```

```python
from scipy.cluster.hierarchy import fcluster

# Cut the hierarchy into 2 clusters
hierarchical_labels = fcluster(
    Z,
    t=2,
    criterion="maxclust"
)

print("Number of clusters:", len(set(hierarchical_labels)))
print("\nCluster counts:")
print(pd.Series(hierarchical_labels).value_counts().sort_index())
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    X["X1"],
    X["X2"],
    c=hierarchical_labels,
    s=35,
    alpha=0.8
)

plt.title(
    "Hierarchical Clustering",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("X1")
plt.ylabel("X2")

plt.grid(alpha=0.2)
plt.show()
```

">
🟣 K-Means Clustering
</h2>

<p style="
    color: #E9D5FF;
    font-size: 18px;
    line-height: 1.9;
">
K-Means divides the data into a predefined number of clusters by assigning each
point to the nearest cluster center.
</p>

<p style="
    color: #E9D5FF;
    font-size: 18px;
    line-height: 1.9;
">
For this dataset, we use <strong style="color:#C4B5FD;">k = 2</strong>
because the dataset visually contains two moon-shaped groups.
</p>

<p style="
    color: #E9D5FF;
    font-size: 18px;
    line-height: 1.9;
">
Although K-Means is simple and useful, it may struggle with this dataset because
its clusters are <strong style="color:#C4B5FD;">curved rather than round</strong>.
</p>

</div>

```python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans.fit_predict(X)

print("Number of clusters:", len(set(kmeans_labels)))
print("\nCluster counts:")
print(pd.Series(kmeans_labels).value_counts().sort_index())
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    X["X1"],
    X["X2"],
    c=kmeans_labels,
    s=35,
    alpha=0.8
)

plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    marker="X",
    s=200,
    edgecolor="black",
    linewidth=1.5,
    label="Centroids"
)

plt.title(
    "K-Means Clustering",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("X1")
plt.ylabel("X2")
plt.legend()
plt.grid(alpha=0.2)

plt.show()
```
