```python
import tensorflow as tf
import pandas as pd
from IPython.display import display, HTML

def build_model(
    hidden_units=32,
    num_layers=1,
    learning_rate=0.001,
    dropout_rate=0.2
):
    model = tf.keras.Sequential()

    model.add(
        tf.keras.layers.Input(shape=(X_train.shape[1],))
    )

    for _ in range(num_layers):
        model.add(
            tf.keras.layers.Dense(
                hidden_units,
                activation="relu"
            )
        )
        model.add(
            tf.keras.layers.Dropout(dropout_rate)
        )

    model.add(
        tf.keras.layers.Dense(
            1,
            activation="sigmoid"
        )
    )

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=learning_rate
        ),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model


experiments = [
    ("Baseline", 0.001),
    ("Lower Learning Rate", 0.0005),
    ("Higher Learning Rate", 0.005)
]

results = []

for name, learning_rate in experiments:

    model = build_model(
        hidden_units=32,
        num_layers=1,
        learning_rate=learning_rate,
        dropout_rate=0.2
    )

    early_stop = tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=2,
        restore_best_weights=True
    )

    history = model.fit(
        X_train,
        y_train,
        validation_data=(X_val, y_val),
        epochs=8,
        batch_size=64,
        callbacks=[early_stop],
        verbose=0
    )

    best_val_accuracy = max(
        history.history["val_accuracy"]
    )

    results.append({
        "Experiment": name,
        "Learning Rate": learning_rate,
        "Best Validation Accuracy": best_val_accuracy
    })


results_df = pd.DataFrame(results)

table_html = results_df.to_html(
    index=False,
    border=0,
    classes="learning-rate-table",
    formatters={
        "Learning Rate": lambda x: f"{x:.4f}",
        "Best Validation Accuracy": lambda x: f"{x:.4f}"
    }
)

display(HTML(f"""
<style>

.learning-rate-table {{
    width:100%;
    border-collapse:separate;
    border-spacing:0;
    font-family:'Segoe UI',Arial,sans-serif;
    font-size:15px;
    border-radius:16px;
    overflow:hidden;
    border:1px solid #C7E9E4;
    box-shadow:0 8px 24px rgba(15,118,110,0.08);
}}

.learning-rate-table th {{
    background:#E0F7EF;
    color:#047857;
    font-weight:900;
    padding:14px 18px;
    text-align:center;
    border-bottom:2px solid #A7E8D0;
}}

.learning-rate-table td {{
    background:#FFFFFF;
    color:#334155;
    font-weight:700;
    padding:13px 18px;
    text-align:center;
    border-bottom:1px solid #E2E8F0;
}}

.learning-rate-table tr:nth-child(even) td {{
    background:#F8FFFC;
}}

.learning-rate-table tr:hover td {{
    background:#FFF7FB;
}}

.learning-rate-table td:first-child {{
    color:#164E63;
    font-weight:900;
}}

.learning-rate-table td:last-child {{
    color:#BE185D;
    font-weight:900;
}}

.learning-rate-table tr:last-child td {{
    border-bottom:none;
}}

</style>

<div style="
    background:
        radial-gradient(
            ellipse at 90% 15%,
            rgba(236,72,153,0.08),
            transparent 28%
        ),
        radial-gradient(
            ellipse at 10% 85%,
            rgba(16,185,129,0.08),
            transparent 28%
        ),
        linear-gradient(
            135deg,
            #F4FFFA 0%,
            #FFFFFF 52%,
            #FFF9FC 100%
        );
    padding:20px 24px;
    border-radius:18px;
    border:1px solid #C7E9E4;
    box-shadow:0 6px 20px rgba(15,118,110,0.07);
    font-family:'Segoe UI',Arial,sans-serif;
    margin:18px 0 12px 0;
">

<div style="
    color:#0EA5A4;
    font-size:13px;
    font-weight:900;
    letter-spacing:2px;
">
HYPERPARAMETER TUNING
</div>

<div style="
    color:#164E63;
    font-size:24px;
    font-weight:900;
    margin-top:5px;
">
Learning Rate Experiments
</div>

<div style="
    color:#64748B;
    font-size:14px;
    font-weight:600;
    margin-top:6px;
">
Testing different learning rates while keeping the
network architecture and dropout fixed.
</div>

</div>

{table_html}
"""))
```

```python
import tensorflow as tf
import pandas as pd
from IPython.display import display, HTML

def build_model(hidden_units=32, num_layers=1, learning_rate=0.001, dropout_rate=0.2):
    model = tf.keras.Sequential()

    model.add(tf.keras.layers.Input(shape=(X_train.shape[1],)))

    for _ in range(num_layers):
        model.add(tf.keras.layers.Dense(hidden_units, activation="relu"))
        model.add(tf.keras.layers.Dropout(dropout_rate))

    model.add(tf.keras.layers.Dense(1, activation="sigmoid"))

    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=learning_rate),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    return model


experiments_width_depth = [
    {"name": "Baseline", "hidden_units": 32, "num_layers": 1},
    {"name": "Wider Network", "hidden_units": 64, "num_layers": 1},
    {"name": "Deeper Network", "hidden_units": 32, "num_layers": 2}
]

results_width_depth = []

for exp in experiments_width_depth:

    model_tuned = build_model(
        hidden_units=exp["hidden_units"],
        num_layers=exp["num_layers"],
        learning_rate=0.001,
        dropout_rate=0.2
    )

    early_stop = tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=2,
        restore_best_weights=True
    )

    history = model_tuned.fit(
        X_train,
        y_train,
        validation_data=(X_val, y_val),
        epochs=8,
        batch_size=64,
        callbacks=[early_stop],
        verbose=0
    )

    best_val_accuracy = max(history.history["val_accuracy"])

    results_width_depth.append({
        "Experiment": exp["name"],
        "Hidden Units": exp["hidden_units"],
        "Layers": exp["num_layers"],
        "Best Validation Accuracy": best_val_accuracy
    })


results_width_depth_df = pd.DataFrame(results_width_depth)


display(HTML("""
<div style="
    width:100%;
    box-sizing:border-box;
    background:
        radial-gradient(ellipse at 90% 10%,rgba(236,72,153,0.08),transparent 25%),
        radial-gradient(ellipse at 10% 90%,rgba(16,185,129,0.10),transparent 27%),
        linear-gradient(135deg,#F4FFFA 0%,#FFFFFF 52%,#FFF8FB 100%);
    padding:20px 24px;
    border-radius:17px;
    font-family:'Segoe UI',Arial,sans-serif;
    border:1px solid #C7E9E4;
    box-shadow:0 7px 20px rgba(15,118,110,0.07);
    margin:18px 0 12px 0;
">

<div style="
    color:#0F766E;
    font-size:13px;
    font-weight:900;
    letter-spacing:2px;
    margin-bottom:6px;
">
WIDTH / DEPTH EXPERIMENTS
</div>

<div style="
    color:#164E63;
    font-size:22px;
    font-weight:900;
">
Comparing Network Capacity
</div>

</div>
"""))


table_html = results_width_depth_df.to_html(
    index=False,
    classes="tuning-table",
    float_format=lambda x: f"{x:.4f}"
)


display(HTML(f"""
<style>

.tuning-table {{
    width:100%;
    border-collapse:separate;
    border-spacing:0;
    font-family:'Segoe UI',Arial,sans-serif;
    font-size:15px;
    border-radius:14px;
    overflow:hidden;
    border:1px solid #A7E8D0;
    box-shadow:0 6px 18px rgba(15,118,110,0.08);
}}

.tuning-table th {{
    background:#E0F7EF;
    color:#047857;
    font-weight:900;
    padding:13px;
    text-align:center;
    border-bottom:1px solid #A7E8D0;
}}

.tuning-table td {{
    background:#FFFFFF;
    color:#334155;
    padding:12px;
    text-align:center;
    border-bottom:1px solid #E2E8F0;
}}

.tuning-table tr:last-child td {{
    border-bottom:none;
}}

.tuning-table tr:hover td {{
    background:#F0FDFA;
}}

</style>

{table_html}
"""))
```
