```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow import keras
from tensorflow.keras import layers
from IPython.display import display, HTML

df = pd.read_csv("../Day1/heart_disease_uci.csv", sep=";")

X = df.drop(columns=["cardio"])
y = df["cardio"]

X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train_raw)
X_test = scaler.transform(X_test_raw)

input_dim = X_train.shape[1]

baseline_model = keras.Sequential([
    layers.Input(shape=(input_dim,)),
    layers.Dense(64, activation="relu"),
    layers.Dense(32, activation="relu"),
    layers.Dense(1, activation="sigmoid")
])

rows = []

for layer in baseline_model.layers:
    output_shape = tuple(layer.output.shape)
    params = layer.count_params()

    rows.append(f"""
    <tr>
        <td style="
            padding:13px 16px;
            text-align:left;
            border-bottom:1px solid #E5D8CB;
            color:#4A4338;
            font-weight:800;
        ">
            {layer.name}
        </td>

        <td style="
            padding:13px 16px;
            text-align:left;
            border-bottom:1px solid #E5D8CB;
            color:#684535;
            font-weight:700;
        ">
            {layer.__class__.__name__}
        </td>

        <td style="
            padding:13px 16px;
            text-align:left;
            border-bottom:1px solid #E5D8CB;
            color:#687455;
            font-weight:700;
        ">
            {output_shape}
        </td>

        <td style="
            padding:13px 16px;
            text-align:left;
            border-bottom:1px solid #E5D8CB;
            color:#9A6947;
            font-weight:900;
        ">
            {params:,}
        </td>
    </tr>
    """)

total_params = baseline_model.count_params()

display(HTML(f"""
<div style="
    width:100%;
    box-sizing:border-box;
    background:#FFFDF9;
    padding:24px 26px;
    border-radius:20px;
    font-family:'Segoe UI',Arial,sans-serif;
    border:1px solid #DDD4C4;
    box-shadow:0 8px 22px rgba(91,76,53,0.07);
    overflow:hidden;
">

<div style="
    color:#9A6947;
    font-size:12px;
    font-weight:900;
    letter-spacing:1.8px;
    margin-bottom:6px;
    text-align:left;
">
MODEL SUMMARY
</div>

<div style="
    color:#2F2A24;
    font-size:25px;
    font-weight:900;
    margin-bottom:18px;
    text-align:left;
">
CardioML Baseline Network
</div>

<table style="
    width:100%;
    border-collapse:collapse;
    border:1px solid #D8C6B7;
    font-size:13px;
    text-align:left;
">

<thead>
<tr style="
    background:#8B5E3C;
    color:#FFFDF9;
">

<th style="padding:14px 16px;text-align:left;font-weight:900;">
Layer
</th>

<th style="padding:14px 16px;text-align:left;font-weight:900;">
Type
</th>

<th style="padding:14px 16px;text-align:left;font-weight:900;">
Output Shape
</th>

<th style="padding:14px 16px;text-align:left;font-weight:900;">
Parameters
</th>

</tr>
</thead>

<tbody>
{''.join(rows)}
</tbody>

</table>


<div style="
    margin-top:15px;
    padding:14px 17px;
    background:#F5EFE6;
    border:1px solid #E0D0BE;
    border-radius:13px;
    display:flex;
    justify-content:space-between;
    align-items:center;
">

<div style="
    color:#9A6947;
    font-size:11px;
    font-weight:900;
    letter-spacing:1.2px;
">
TOTAL PARAMETERS
</div>

<div style="
    color:#684535;
    font-size:19px;
    font-weight:900;
">
{total_params:,}
</div>

</div>


<div style="
    margin-top:14px;
    color:#6B665D;
    font-size:12px;
    font-weight:700;
    text-align:left;
">
Dataset: {len(df):,} records • Input features: {input_dim} • Binary target: cardio
</div>

</div>
"""))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss",
    linewidth=2
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss",
    linewidth=2
)

plt.xlabel("Epoch", fontsize=12)
plt.ylabel("Loss", fontsize=12)
plt.title("Training vs Validation Loss", fontsize=15, fontweight="bold")
plt.legend()
plt.grid(alpha=0.25)

plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy",
    linewidth=2
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy",
    linewidth=2
)

plt.xlabel("Epoch", fontsize=12)
plt.ylabel("Accuracy", fontsize=12)
plt.title("Training vs Validation Accuracy", fontsize=15, fontweight="bold")
plt.legend()
plt.grid(alpha=0.25)

plt.show()
```

```python
from tensorflow import keras
from tensorflow.keras import layers

improved_model = keras.Sequential([
    layers.Input(shape=(input_dim,)),

    layers.Dense(128),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.Dropout(0.20),

    layers.Dense(64),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.Dropout(0.20),

    layers.Dense(32),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.Dropout(0.10),

    layers.Dense(1, activation="sigmoid")
])

improved_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

callbacks = [
    keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=8,
        restore_best_weights=True
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=4,
        min_lr=1e-6
    )
]

improved_history = improved_model.fit(
    X_train,
    y_train,
    validation_split=0.20,
    epochs=100,
    batch_size=64,
    callbacks=callbacks,
    verbose=1
)
```

<div style="color:#6B665D;font-size:15px;font-weight:600;line-height:1.9;margin-top:9px;">
Instead of relying only on the original measurements, we create two additional
features that represent medically meaningful relationships between existing variables.
This experiment tests whether these derived features provide useful information
for predicting cardiovascular disease.
</div>

<div style="margin-top:18px;padding:17px 20px;background:#F5EFE6;border:1px solid #E0D0BE;border-left:5px solid #B97852;border-radius:15px;">

<div style="color:#9A6947;font-size:12px;font-weight:900;letter-spacing:1.3px;">
BMI
</div>

<div style="color:#4A4338;font-size:14px;line-height:1.85;font-weight:700;margin-top:6px;">
BMI combines height and weight into one feature that represents body-mass level.
We calculate it using weight divided by height in meters squared.
</div>

</div>

<div style="margin-top:12px;padding:17px 20px;background:#EFF1E8;border:1px solid #D0D4C2;border-left:5px solid #687455;border-radius:15px;">

<div style="color:#687455;font-size:12px;font-weight:900;letter-spacing:1.3px;">
PULSE PRESSURE
</div>

<div style="color:#4A4F3E;font-size:14px;line-height:1.85;font-weight:700;margin-top:6px;">
Pulse Pressure is calculated as systolic blood pressure minus diastolic blood pressure.
It provides the model with a derived representation of blood-pressure behavior.
</div>

</div>

</div>


```python
df_temp = df.copy()

# BMI (Body Mass Index)
# BMI = weight / (height in meters)^2
df_temp["bmi"] = df_temp["weight"] / (df_temp["height"] / 100) ** 2

# PP (Pulse Pressure)
# PP = systolic blood pressure - diastolic blood pressure
df_temp["pulse_pressure"] = df_temp["ap_hi"] - df_temp["ap_lo"]

print("Original Shape:", df.shape)
print("Temporary Shape:", df_temp.shape)

display(df_temp[[
    "height",
    "weight",
    "ap_hi",
    "ap_lo",
    "bmi",
    "pulse_pressure",
    "cardio"
]].head())
```

<div style="color:#4A4338;font-size:14px;line-height:1.85;font-weight:700;margin-top:6px;">

<div style="margin-top:12px;padding:17px 20px;background:#EFF1E8;border:1px solid #D0D4C2;border-left:5px solid #687455;border-radius:15px;">

<div style="color:#4A4F3E;font-size:14px;line-height:1.85;font-weight:700;margin-top:6px;">

the number of detected outliers, and the median used for replacement.

```python
import numpy as np

df_clean = df_clean.copy()

for col in ["ap_hi", "ap_lo"]:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    median = df_clean[col].median()

    outliers = (df_clean[col] < lower_bound) | (df_clean[col] > upper_bound)

    print(f"{col}:")
    print("  Lower Bound:", lower_bound)
    print("  Upper Bound:", upper_bound)
    print("  Outliers:", outliers.sum())
    print("  Median:", median)

    df_clean.loc[outliers, col] = median

print("\nAfter outlier treatment:", df_clean.shape)
```

Identify Physiologically Invalid Values

<div style="

    font-weight:600;
    margin-top:11px;

We create a separate copy of the dataset and define rules for detecting
invalid blood-pressure measurements. The checks include unrealistic
systolic and diastolic ranges and incorrect systolic-to-diastolic ordering.
</div>

</div>

```python
df_bp = df.copy()

# Invalid blood pressure values
invalid_bp = (
    (df_bp["ap_hi"] < 70) |
    (df_bp["ap_hi"] > 250) |
    (df_bp["ap_lo"] < 40) |
    (df_bp["ap_lo"] > 150) |
    (df_bp["ap_hi"] < df_bp["ap_lo"])
)

print("Invalid BP rows:", invalid_bp.sum())
```

```python
invalid_ap_hi = (
    (df_bp["ap_hi"] < 70) |
    (df_bp["ap_hi"] > 250)
)

invalid_ap_lo = (
    (df_bp["ap_lo"] < 40) |
    (df_bp["ap_lo"] > 150)
)

invalid_order = df_bp["ap_hi"] < df_bp["ap_lo"]

print("Invalid ap_hi:", invalid_ap_hi.sum())
print("Invalid ap_lo:", invalid_ap_lo.sum())
print("ap_hi < ap_lo:", invalid_order.sum())
```

```python
df_bp_clean = df.copy()

# Medians from valid blood pressure values
valid_ap_hi = df_bp_clean.loc[
    (df_bp_clean["ap_hi"] >= 70) &
    (df_bp_clean["ap_hi"] <= 250),
    "ap_hi"
].median()

valid_ap_lo = df_bp_clean.loc[
    (df_bp_clean["ap_lo"] >= 40) &
    (df_bp_clean["ap_lo"] <= 150),
    "ap_lo"
].median()

# Replace invalid ap_hi values
invalid_ap_hi = (
    (df_bp_clean["ap_hi"] < 70) |
    (df_bp_clean["ap_hi"] > 250)
)

df_bp_clean.loc[invalid_ap_hi, "ap_hi"] = valid_ap_hi

# Replace invalid ap_lo values
invalid_ap_lo = (
    (df_bp_clean["ap_lo"] < 40) |
    (df_bp_clean["ap_lo"] > 150)
)

df_bp_clean.loc[invalid_ap_lo, "ap_lo"] = valid_ap_lo

# Fix cases where systolic became lower than diastolic
invalid_order = df_bp_clean["ap_hi"] < df_bp_clean["ap_lo"]

df_bp_clean.loc[invalid_order, "ap_hi"] = valid_ap_hi
df_bp_clean.loc[invalid_order, "ap_lo"] = valid_ap_lo

print("Original Shape:", df.shape)
print("After BP Cleaning:", df_bp_clean.shape)

print("\nRemaining invalid:")
print("ap_hi < 70:", (df_bp_clean["ap_hi"] < 70).sum())
print("ap_hi > 250:", (df_bp_clean["ap_hi"] > 250).sum())
print("ap_lo < 40:", (df_bp_clean["ap_lo"] < 40).sum())
print("ap_lo > 150:", (df_bp_clean["ap_lo"] > 150).sum())
print("ap_hi < ap_lo:", (df_bp_clean["ap_hi"] < df_bp_clean["ap_lo"]).sum())
```

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler
import numpy as np

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

cv_scores = []

for fold, (train_idx, val_idx) in enumerate(
    skf.split(X_bp, y_bp),
    start=1
):

    X_cv_train = X_bp.iloc[train_idx]
    X_cv_val = X_bp.iloc[val_idx]

    y_cv_train = y_bp.iloc[train_idx]
    y_cv_val = y_bp.iloc[val_idx]

    # Scale inside each fold
    scaler_cv = StandardScaler()

    X_cv_train_scaled = scaler_cv.fit_transform(X_cv_train)
    X_cv_val_scaled = scaler_cv.transform(X_cv_val)

    # Build the same architecture
    cv_model = keras.Sequential([
        layers.Input(shape=(X_cv_train_scaled.shape[1],)),

        layers.Dense(128),
        layers.BatchNormalization(),
        layers.Activation("relu"),
        layers.Dropout(0.20),

        layers.Dense(64),
        layers.BatchNormalization(),
        layers.Activation("relu"),
        layers.Dropout(0.20),

        layers.Dense(32),
        layers.BatchNormalization(),
        layers.Activation("relu"),
        layers.Dropout(0.10),

        layers.Dense(1, activation="sigmoid")
    ])

    cv_model.compile(
        optimizer=keras.optimizers.Adam(learning_rate=0.001),
        loss="binary_crossentropy",
        metrics=["accuracy"]
    )

    cv_model.fit(
        X_cv_train_scaled,
        y_cv_train,
        validation_split=0.20,
        epochs=100,
        batch_size=64,
        callbacks=callbacks,
        verbose=0
    )

    _, fold_accuracy = cv_model.evaluate(
        X_cv_val_scaled,
        y_cv_val,
        verbose=0
    )

    cv_scores.append(fold_accuracy)

    print(
        f"Fold {fold} Accuracy: "
        f"{fold_accuracy * 100:.2f}%"
    )

print("\nCross-Validation Summary")
print(f"Mean Accuracy: {np.mean(cv_scores) * 100:.2f}%")
print(f"Standard Deviation: {np.std(cv_scores) * 100:.2f}%")
```
