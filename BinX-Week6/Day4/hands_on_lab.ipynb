```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow import keras
from tensorflow.keras import layers
from IPython.display import display, HTML

df = pd.read_csv("../Day1/heart_disease_uci.csv", sep=";")

X = df.drop(columns=["cardio"])
y = df["cardio"]

X_train_raw, X_test_raw, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train_raw)
X_test = scaler.transform(X_test_raw)

input_dim = X_train.shape[1]

baseline_model = keras.Sequential([
    layers.Input(shape=(input_dim,)),
    layers.Dense(64, activation="relu"),
    layers.Dense(32, activation="relu"),
    layers.Dense(1, activation="sigmoid")
])

rows = []

for layer in baseline_model.layers:
    output_shape = tuple(layer.output.shape)
    params = layer.count_params()

    rows.append(f"""
    <tr>
        <td style="
            padding:13px 16px;
            text-align:left;
            border-bottom:1px solid #E5D8CB;
            color:#4A4338;
            font-weight:800;
        ">
            {layer.name}
        </td>

        <td style="
            padding:13px 16px;
            text-align:left;
            border-bottom:1px solid #E5D8CB;
            color:#684535;
            font-weight:700;
        ">
            {layer.__class__.__name__}
        </td>

        <td style="
            padding:13px 16px;
            text-align:left;
            border-bottom:1px solid #E5D8CB;
            color:#687455;
            font-weight:700;
        ">
            {output_shape}
        </td>

        <td style="
            padding:13px 16px;
            text-align:left;
            border-bottom:1px solid #E5D8CB;
            color:#9A6947;
            font-weight:900;
        ">
            {params:,}
        </td>
    </tr>
    """)

total_params = baseline_model.count_params()

display(HTML(f"""
<div style="
    width:100%;
    box-sizing:border-box;
    background:#FFFDF9;
    padding:24px 26px;
    border-radius:20px;
    font-family:'Segoe UI',Arial,sans-serif;
    border:1px solid #DDD4C4;
    box-shadow:0 8px 22px rgba(91,76,53,0.07);
    overflow:hidden;
">

<div style="
    color:#9A6947;
    font-size:12px;
    font-weight:900;
    letter-spacing:1.8px;
    margin-bottom:6px;
    text-align:left;
">
MODEL SUMMARY
</div>

<div style="
    color:#2F2A24;
    font-size:25px;
    font-weight:900;
    margin-bottom:18px;
    text-align:left;
">
CardioML Baseline Network
</div>

<table style="
    width:100%;
    border-collapse:collapse;
    border:1px solid #D8C6B7;
    font-size:13px;
    text-align:left;
">

<thead>
<tr style="
    background:#8B5E3C;
    color:#FFFDF9;
">

<th style="padding:14px 16px;text-align:left;font-weight:900;">
Layer
</th>

<th style="padding:14px 16px;text-align:left;font-weight:900;">
Type
</th>

<th style="padding:14px 16px;text-align:left;font-weight:900;">
Output Shape
</th>

<th style="padding:14px 16px;text-align:left;font-weight:900;">
Parameters
</th>

</tr>
</thead>

<tbody>
{''.join(rows)}
</tbody>

</table>


<div style="
    margin-top:15px;
    padding:14px 17px;
    background:#F5EFE6;
    border:1px solid #E0D0BE;
    border-radius:13px;
    display:flex;
    justify-content:space-between;
    align-items:center;
">

<div style="
    color:#9A6947;
    font-size:11px;
    font-weight:900;
    letter-spacing:1.2px;
">
TOTAL PARAMETERS
</div>

<div style="
    color:#684535;
    font-size:19px;
    font-weight:900;
">
{total_params:,}
</div>

</div>


<div style="
    margin-top:14px;
    color:#6B665D;
    font-size:12px;
    font-weight:700;
    text-align:left;
">
Dataset: {len(df):,} records • Input features: {input_dim} • Binary target: cardio
</div>

</div>
"""))
```

```python
history = baseline_model.fit(
    X_train,
    y_train,
    validation_split=0.20,
    epochs=30,
    batch_size=32,
    verbose=1
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss",
    linewidth=2
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss",
    linewidth=2
)

plt.xlabel("Epoch", fontsize=12)
plt.ylabel("Loss", fontsize=12)
plt.title("Training vs Validation Loss", fontsize=15, fontweight="bold")
plt.legend()
plt.grid(alpha=0.25)

plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy",
    linewidth=2
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy",
    linewidth=2
)

plt.xlabel("Epoch", fontsize=12)
plt.ylabel("Accuracy", fontsize=12)
plt.title("Training vs Validation Accuracy", fontsize=15, fontweight="bold")
plt.legend()
plt.grid(alpha=0.25)

plt.show()
```

```python
from tensorflow import keras
from tensorflow.keras import layers

improved_model = keras.Sequential([
    layers.Input(shape=(input_dim,)),

    layers.Dense(128),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.Dropout(0.20),

    layers.Dense(64),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.Dropout(0.20),

    layers.Dense(32),
    layers.BatchNormalization(),
    layers.Activation("relu"),
    layers.Dropout(0.10),

    layers.Dense(1, activation="sigmoid")
])

improved_model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=0.001),
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

callbacks = [
    keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=8,
        restore_best_weights=True
    ),
    keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=4,
        min_lr=1e-6
    )
]

improved_history = improved_model.fit(
    X_train,
    y_train,
    validation_split=0.20,
    epochs=100,
    batch_size=32,
    callbacks=callbacks,
    verbose=1
)
```
