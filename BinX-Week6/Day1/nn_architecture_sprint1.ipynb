WHAT A NEURON DOES
</div>

</div>

<div style="
    background:#FFFFFF;
    border:1px solid #D7E0EA;
    border-radius:16px;
    padding:20px 22px;
">

<p style="
    color:#172554;
    font-size:14px;
    line-height:1.75;
    font-weight:700;
    margin:0 0 16px 0;
">
A neuron takes its inputs, multiplies each by a weight, sums them up,
adds a bias, then passes the result through an activation function.
</p>

<div style="
    background:#F8FAFC;
    border:1px solid #D7E0EA;
    border-radius:13px;
    padding:15px 20px;
    display:flex;
    align-items:center;
    justify-content:center;
    gap:16px;
    flex-wrap:wrap;
">

<div style="
    color:#2563EB;
    font-size:10px;
    font-weight:900;
    letter-spacing:1.3px;
">
NEURON FORMULA
</div>

<div style="
    color:#172554;
    font-size:15px;
    font-weight:900;
">
z = dot(inputs, weights) + bias
</div>

<div style="
    color:#64748B;
    font-size:14px;
    font-weight:900;
">
→
</div>

<div style="
    color:#075985;
    font-size:15px;
    font-weight:900;
">
output = activation(z)
</div>

</div>

</div>

</div>

```python
import numpy as np

# ============================================
# Conceptual Neuron — connects back to the Week 2 dot product
# ============================================

# Example: 3 input features (like age, resting BP, cholesterol - simplified/scaled)
inputs = np.array([0.63, 0.45, 0.72])

# Each input has a weight (learned during training — random for now)
weights = np.array([0.8, -0.3, 0.5])

# Bias shifts the result
bias = 0.1

# Step 1: weighted sum (the Week 2 dot product) + bias
z = np.dot(inputs, weights) + bias
print(f"z (weighted sum + bias) = {z:.4f}")

# Step 2: activation function (sigmoid, squashes to 0-1)
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

output = sigmoid(z)
print(f"output = activation(z) = {output:.4f}")
```
