Activation functions transform the values produced by a neural network
layer before passing them to the next layer. Different activation
functions have different output ranges and behaviors.
</p>

<p style="
    color:#334155;
    font-size:14px;
    line-height:1.8;
    font-weight:600;
    margin:0;
">
In this step, we will plot <strong style="color:#065F46;">ReLU</strong>,
<strong style="color:#0D9488;">Sigmoid</strong>, and
<strong style="color:#166534;">Tanh</strong> over the same range of
input values. This allows us to visually compare how each function
transforms the input.
</p>

</div>

<div style="
    margin-top:16px;
    background:#F8FAFC;
    border:1px solid #D7E0EA;
    border-radius:13px;
    padding:14px 18px;
    color:#475569;
    font-size:12px;
    font-weight:700;
    text-align:center;
">
<strong>Goal:</strong> See the behavior and output range of each activation function.
</div>

</div>

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 500)

relu = np.maximum(0, x)
sigmoid = 1 / (1 + np.exp(-x))
tanh = np.tanh(x)

plt.figure(figsize=(12, 7))

plt.plot(x, relu, label="ReLU", linewidth=2)
plt.plot(x, sigmoid, label="Sigmoid", linewidth=2)
plt.plot(x, tanh, label="Tanh", linewidth=2)

plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)

plt.title(
    "Activation Functions: ReLU, Sigmoid, and Tanh",
    fontsize=16,
    fontweight="bold"
)

plt.xlabel("Input (x)", fontsize=12)
plt.ylabel("Activation Output", fontsize=12)

plt.grid(alpha=0.25)
plt.legend()
plt.show()
```

<div style="

<p style="
    color:#334155;

    margin:0 0 14px 0;
">
A <strong style="color:#065F46;">forward pass</strong> is the process of
passing an input through the neural network to produce a prediction.
Each layer performs a weighted sum using its weights and biases, then
applies an activation function.
</p>

<p style="
    color:#334155;
    font-size:14px;
    line-height:1.8;
    font-weight:600;
    margin:0 0 14px 0;
">
For this exercise, we will use a tiny network with
<strong style="color:#0D9488;">2 input features → 2 hidden neurons → 1 output neuron</strong>.
ReLU will be used in the hidden layer, while Sigmoid will be used in
the output layer.
</p>

<div style="
    background:#F0FDFA;
    border:1px solid #99F6E4;
    border-radius:14px;
    padding:15px 18px;
    color:#115E59;
    font-size:12px;
    font-weight:800;
    text-align:center;
">
Input → Linear Transformation → ReLU → Linear Transformation → Sigmoid → Prediction
</div>

</div>

</div>

```python

x = np.array([[2.0, 1.0]])

# Layer 1: 2 inputs → 2 hidden neurons
W1 = np.array([
    [0.5, -0.5],
    [1.0,  0.5]
])

b1 = np.array([[0.5, 0.0]])

# Layer 2: 2 hidden neurons → 1 output neuron
W2 = np.array([
    [1.0],
    [-1.0]
])

b2 = np.array([[0.5]])


# ReLU activation
def relu(z):
    return np.maximum(0, z)


# Sigmoid activation
def sigmoid(z):
    return 1 / (1 + np.exp(-z))


# ----- Layer 1 -----
z1 = x @ W1 + b1
a1 = relu(z1)

# ----- Layer 2 -----
z2 = a1 @ W2 + b2
y_pred = sigmoid(z2)


# Display the forward-pass results
print("Input (x):")
print(x)

print("\nHidden Layer - Weighted Sum (z1):")
print(z1)

print("\nHidden Layer - After ReLU (a1):")
print(a1)

print("\nOutput Layer - Weighted Sum (z2):")
print(z2)

print("\nFinal Prediction - After Sigmoid:")
print(y_pred)
```
