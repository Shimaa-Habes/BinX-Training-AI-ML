<div style="color:#0D9488; font-size:11px; font-weight:900; letter-spacing:2px;">SECTION 2.1</div>
<div style="color:#022C22; font-size:20px; font-weight:900; line-height:1.3;">Why Activation Functions Matter</div>
</div>
</div>

<div style="background:#FFFFFF; border:1px solid #BFE3D3; border-radius:16px; padding:20px 22px;">

<p style="color:#334155; font-size:14px; line-height:1.8; font-weight:600; margin:0 0 14px 0;">
Imagine a network with 10 hidden layers, but every layer just does
<code style="background:#F0FDFA;padding:2px 6px;border-radius:5px;">weighted sum + bias</code>
with no activation. Mathematically, stacking linear operations always
collapses back into <b>one single linear operation</b> — no matter how many layers you add,
the whole network behaves exactly like plain linear regression.
</p>

<p style="color:#334155; font-size:14px; line-height:1.8; font-weight:600; margin:0 0 14px 0;">
That's a problem, because most real relationships in data — like how blood pressure,
cholesterol, and age combine to predict heart disease — aren't straight lines.
They're curved, conditional, and interact with each other in complex ways.
A purely linear model can never capture that, regardless of depth.
</p>

<p style="color:#334155; font-size:14px; line-height:1.8; font-weight:600; margin:0;">
The <span style="color:#10B981;">activation function</span> is what breaks this limitation.
By applying a non-linear transformation after every layer, each layer can bend
and reshape the data differently, letting the network build up increasingly
complex, curved decision boundaries as it goes deeper. This is the single ingredient
that separates a "deep" network from just a longer linear equation —
and it's why depth actually helps.
</p>

</div>

<div style="display:flex; gap:9px; flex-wrap:nowrap; margin-top:16px;">
<div style="height:30px; box-sizing:border-box; display:flex; align-items:center; justify-content:center; background:#D1FAE5; color:#065F46; border:1px solid #A7F3D0; padding:0 12px; border-radius:9px; font-size:10px; font-weight:900; white-space:nowrap;">RELU</div>
<div style="height:30px; box-sizing:border-box; display:flex; align-items:center; justify-content:center; background:#CCFBF1; color:#115E59; border:1px solid #99F6E4; padding:0 12px; border-radius:9px; font-size:10px; font-weight:900; white-space:nowrap;">SIGMOID</div>
<div style="height:30px; box-sizing:border-box; display:flex; align-items:center; justify-content:center; background:#DCFCE7; color:#166534; border:1px solid #BBF7D0; padding:0 12px; border-radius:9px; font-size:10px; font-weight:900; white-space:nowrap;">SOFTMAX</div>

<div style="color:#166534; font-size:12px; font-weight:900; letter-spacing:2px;">WHY STACKING LINEAR LAYERS FAILS</div>
</div>

<div style="background:#FFFFFF; border:1px solid #BFE3D3; border-radius:16px; padding:20px 22px;">

<p style="color:#334155; font-size:14px; line-height:1.8; font-weight:600; margin:0 0 16px 0;">
Say layer 1 computes <code style="background:#F0FDFA;padding:2px 6px;border-radius:5px;">h = W₁x + b₁</code>,
and layer 2 takes that output and computes
<code style="background:#F0FDFA;padding:2px 6px;border-radius:5px;">y = W₂h + b₂</code> — with no activation in between.
</p>

<div style="background:#F8FAFC; border:1px solid #D7E0EA; border-radius:13px; padding:16px 20px; margin-bottom:16px;">
<div style="color:#0D9488; font-size:10px; font-weight:900; letter-spacing:1.3px; margin-bottom:8px;">SUBSTITUTE h INTO y</div>
<div style="color:#022C22; font-size:14px; font-weight:900; line-height:1.8;">
y = W₂(W₁x + b₁) + b₂<br>
y = (W₂W₁)x + (W₂b₁ + b₂)
</div>
</div>

<p style="color:#334155; font-size:14px; line-height:1.8; font-weight:600; margin:0;">
That's it — <code style="background:#F0FDFA;padding:2px 6px;border-radius:5px;">W₂W₁</code> is just
another matrix, and <code style="background:#F0FDFA;padding:2px 6px;border-radius:5px;">W₂b₁ + b₂</code>
is just another bias. Two "layers" collapsed into
<span style="color:#10B981;">one single linear equation</span> — same shape as one layer.
Add a hundred more layers with no activation, and the same thing happens: it all
flattens into one linear function no matter the depth.
</p>

<div style="background:#FFFFFF; border:1px solid #BFE3D3; border-radius:16px; padding:20px 22px;">

<p style="color:#334155; font-size:14px; line-height:1.8; font-weight:600; margin:0;">
Non-linearity is the one thing that turns a stack of layers into an actual
<span style="color:#10B981;">deep</span> network instead of a disguised linear model.

heart disease) instead of just fitting a straight line through it.

```python
import numpy as np


def relu(z):
    return np.maximum(0, z)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

X = np.array([0.63, 0.45, 0.72])

W1 = np.array([[0.8, -0.3, 0.5],
               [0.2,  0.6, -0.4]])
b1 = np.array([0.1, -0.1])

W2 = np.array([0.7, -0.5])
b2 = 0.05

z1 = np.dot(W1, X) + b1          
layer1 = relu(z1)                 
print(f"z1 = {z1}")
print(f"layer1 (after ReLU) = {layer1}")

z2 = np.dot(W2, layer1) + b2      
output = sigmoid(z2)              
print(f"z2 = {z2:.4f}")
print(f"output = sigmoid(z2) = {output:.4f}")
```
