🖼️ Spatial Data vs 🔄 Sequential Data
</div>

<div style="
    color:#4F5D63;
    font-size:15px;
    line-height:1.9;
    font-weight:500;
">

<b style="color:#24323D;">CNNs</b> are designed mainly for
<b style="color:#2A9D8F;">spatial data</b>, such as images.
They learn patterns based on the relationships between nearby pixels
and their position in the image. For example, a CNN can learn edges,
shapes, textures, and eventually more complex visual patterns.

<br><br>

Sequential data is different because the
<b style="color:#F26B5B;">order of the elements matters</b>.
Examples include text, time series, speech, and audio. For this type
of data, we need an architecture that can process elements in sequence
and use information from previous elements when interpreting the
current one.

<br><br>

This is where <b style="color:#F26B5B;">Recurrent Neural Networks (RNNs)</b>
become useful. Unlike a CNN, an RNN processes the sequence step by step
while maintaining a <b style="color:#2A9D8F;">hidden state</b> that carries
information from earlier steps.

</div>

</div>

🔤 Why Does Order Matter?
</div>

<div style="
    color:#4F5D63;
    font-size:15px;
    line-height:1.9;
    font-weight:500;
">

In sequential data, changing the order of the elements can change
the meaning completely.

<br><br>

<div style="

“Good, the movie was not.”

The sentences contain almost the same words, but the way the words
are arranged changes how we interpret the sentence.

<br><br>

The first sentence clearly expresses a
<b style="color:#F26B5B;">negative opinion</b> about the movie.
A model that understands language therefore cannot simply look at
which words exist; it must also understand
<b style="color:#2A9D8F;">the order in which they appear</b>
and how they relate to each other.

<br><br>

This is why sequence models need to be
<b style="color:#E88955;">order-aware</b>.

</div>

🧠 The Need for Memory
</div>

<div style="
    color:#4F5D63;
    font-size:15px;
    line-height:1.9;
    font-weight:500;
">

A model working with sequential data needs to consider not only
the current element, but also what came
<b style="color:#F26B5B;">before it</b>.

<br><br>

When reading a sentence, for example, the meaning of a word can
depend on the words that appeared earlier. The model therefore needs
a way to carry useful information from previous steps into the
current step.

<br><br>

RNNs achieve this through a
<b style="color:#2A9D8F;">hidden state</b>, which acts as a form of
memory. At each step, the RNN receives the current input and combines
it with information stored from previous steps.

<br><br>

<div style="

Previous Information

Current Input
<span style="color:#F6A06D;"> → </span>
Updated Memory
<span style="color:#2A9D8F;"> → </span>
Next Step
</div>

This ability to carry context forward is what makes
<b style="color:#2A9D8F;">RNNs suitable for sequential data</b>.

🧠 Hidden State — The RNN's Memory
</div>

<div style="
    color:#4F5D63;
    font-size:15px;
    line-height:1.9;
    font-weight:500;
">

An <b style="color:#F26B5B;">RNN processes a sequence one element at a time</b>.
At every step, it receives the current input and combines it with
information from the previous step.

<br><br>

The information carried from one step to the next is called the
<b style="color:#2A9D8F;">hidden state</b>. We can think of it as the
RNN's memory because it stores useful information about what the model
has already seen.

<br><br>

After processing each element, the hidden state is updated and passed
to the next step. This allows the RNN to use
<b style="color:#E88955;">previous context when processing new information</b>.

</div>

</div>

Updated Hidden State
<span style="color:#F26B5B;"> → </span>

</div>


<!-- Limitation -->
<div>

<div style="
    color:#E88955;
    font-size:19px;

    margin-bottom:11px;
">
⚠️ Limitation of Plain RNNs
</div>

<div style="
    color:#4F5D63;
    font-size:15px;
    line-height:1.9;
    font-weight:500;
">

Although the hidden state gives an RNN the ability to remember previous
information, a <b style="color:#F26B5B;">plain RNN struggles with long sequences</b>.

<br><br>

As the sequence becomes longer, information from earlier time steps has
to pass through many recurrent steps. During training, the gradients
can become extremely small as they are propagated backward through
these steps.

<br><br>

This is known as the
<b style="color:#E88955;">vanishing-gradient problem</b>. As a result,
the RNN may gradually lose important information from far earlier in
the sequence.

<br><br>

This limitation is one of the main reasons why more advanced recurrent
architectures such as <b style="color:#2A9D8F;">LSTMs and GRUs</b> were
developed.

</div>

</div>

</div>

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

rnn_model = Sequential([
    SimpleRNN(32, activation="tanh", input_shape=(10, 1)),
    Dense(1)
])

rnn_model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

rnn_model.summary()
```

```python
import numpy as np
import time
from IPython.display import display, HTML


sequence = np.arange(1, 201, dtype=np.float32)

X = []
y = []

sequence_length = 10

for i in range(len(sequence) - sequence_length):
    X.append(sequence[i:i + sequence_length])
    y.append(sequence[i + sequence_length])

X = np.array(X)
y = np.array(y)

X = X.reshape((X.shape[0], X.shape[1], 1))

split = int(len(X) * 0.8)

X_train = X[:split]
y_train = y[:split]

X_val = X[split:]
y_val = y[split:]



rnn_model = Sequential([
    SimpleRNN(32, activation="tanh", input_shape=(sequence_length, 1)),
    Dense(1)
])

rnn_model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)



start_time = time.time()

rnn_history = rnn_model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=20,
    batch_size=16,
    verbose=1
)

training_time = time.time() - start_time



loss, mae = rnn_model.evaluate(
    X_val,
    y_val,
    verbose=0
)

best_val_loss = min(rnn_history.history["val_loss"])
best_val_mae = min(rnn_history.history["val_mae"])



display(HTML(f"""
<div style="
    width:100%;
    max-width:100%;
    box-sizing:border-box;
    margin:25px 0;
    padding:26px 30px;
    background:#FFFFFF;
    border:1px solid #E8E8E8;
    border-radius:18px;
    box-shadow:0 6px 18px rgba(36,50,61,0.06);
    font-family:'Segoe UI',Arial,sans-serif;
">

<div style="
    color:#F26B5B;
    font-size:12px;
    font-weight:900;
    letter-spacing:2px;
    margin-bottom:8px;
">
RNN TRAINING RESULT
</div>

<div style="
    color:#24323D;
    font-size:24px;
    font-weight:900;
    margin-bottom:20px;
">
🔄 Plain RNN — Sequential Prediction
</div>

<div style="
    display:flex;
    flex-wrap:wrap;
    gap:12px;
">

<div style="
    flex:1;
    min-width:170px;
    box-sizing:border-box;
    padding:17px;
    background:#FFF5F1;
    border-radius:12px;
    border-left:4px solid #F26B5B;
">
<div style="color:#697780;font-size:12px;font-weight:800;">
VALIDATION LOSS
</div>
<div style="color:#24323D;font-size:23px;font-weight:900;margin-top:5px;">
{loss:.4f}
</div>
</div>

<div style="
    flex:1;
    min-width:170px;
    box-sizing:border-box;
    padding:17px;
    background:#F2FAF8;
    border-radius:12px;
    border-left:4px solid #2A9D8F;
">
<div style="color:#697780;font-size:12px;font-weight:800;">
VALIDATION MAE
</div>
<div style="color:#24323D;font-size:23px;font-weight:900;margin-top:5px;">
{mae:.4f}
</div>
</div>

<div style="
    flex:1;
    min-width:170px;
    box-sizing:border-box;
    padding:17px;
    background:#FFF9F3;
    border-radius:12px;
    border-left:4px solid #F6A06D;
">
<div style="color:#697780;font-size:12px;font-weight:800;">
TRAINING TIME
</div>
<div style="color:#24323D;font-size:23px;font-weight:900;margin-top:5px;">
{training_time:.2f}s
</div>
</div>

</div>

<div style="
    margin-top:18px;
    padding:15px 18px;
    background:#F7F8F8;
    border-radius:11px;
    color:#4F5D63;
    font-size:14px;
    line-height:1.7;
    font-weight:600;
">
The RNN successfully learned a simple sequential pattern by using
previous time-step information through its hidden state.
</div>

</div>
"""))
```

📉 How the Gradient Vanishes
</div>

<div style="
    color:#4F5D63;
    font-size:14px;
    line-height:1.75;
    font-weight:500;
">
During training, the RNN uses
<b style="color:#F26B5B;">Backpropagation Through Time (BPTT)</b>
to send the error signal backward through the time steps.

When the sequence is long, the gradient is repeatedly multiplied by
values related to the recurrent weights and activation function.

If these values are smaller than 1, the gradient becomes
<b style="color:#E88955;">smaller at every time step</b>.
After many steps, it can shrink exponentially and become extremely close to zero.

This is known as the
<b style="color:#F26B5B;">Vanishing Gradient Problem</b>.
</div>

</div>

<div style="
    width:100%;
    box-sizing:border-box;
    margin:20px 0;

⚠️ What Happens to the RNN?
</div>

<div style="
    color:#4F5D63;
    font-size:14px;
    line-height:1.75;
    font-weight:500;
">
When the gradient becomes very small, earlier time steps receive
<b style="color:#F26B5B;">almost no useful learning signal</b>.

As a result, a plain RNN has difficulty learning relationships between
information that is far apart in a sequence.

The model may gradually
<b style="color:#E88955;">forget important information from earlier time steps</b>
and focus more on recent information.

For example, if important information appears near the beginning of a
long sequence, the RNN may struggle to use it when making a prediction
near the end.
</div>

💡 Why Were LSTMs Introduced?
</div>

<div style="
    color:#4F5D63;
    font-size:14px;
    line-height:1.75;
    font-weight:500;
">
The vanishing gradient problem is one of the main limitations that
motivated the development of
<b style="color:#2A9D8F;">Long Short-Term Memory (LSTM)</b> networks.

LSTMs were specifically designed to preserve important information
over longer sequences using a
<b style="color:#F26B5B;">memory cell and gates</b>
that control what information should be kept, updated, or forgotten.

This allows LSTMs to handle
<b style="color:#E88955;">long-term dependencies</b>
much better than a plain RNN.
</div>

🧠 LSTM Gates — Controlling Memory
</div>

<div style="
    color:#4F5D63;
    font-size:14px;
    line-height:1.75;
    font-weight:500;
">
An <b style="color:#2A9D8F;">LSTM</b> improves the basic RNN by introducing a
memory cell and gates that control how information flows through the network.

The <b style="color:#F26B5B;">forget gate</b> decides what old information
should be removed from the memory.

The <b style="color:#E88955;">input gate</b> decides what new information
should be added to the memory.

The <b style="color:#2A9D8F;">output gate</b> controls what information
from the memory should be passed to the next step and used as the output.

Together, these gates allow the LSTM to keep important information for
longer periods instead of gradually losing it.
</div>

</div>

⚡ GRU — A Simpler Recurrent Architecture
</div>

<div style="
    color:#4F5D63;
    font-size:14px;
    line-height:1.75;
    font-weight:500;
">
A <b style="color:#E88955;">GRU (Gated Recurrent Unit)</b> is another recurrent
architecture designed to handle the long-term dependency problem.

Unlike an LSTM, a GRU uses a
<b style="color:#F26B5B;">simpler gating mechanism</b>
with fewer gates and does not use a separate memory cell.

Because of its simpler structure, a GRU generally has
<b style="color:#2A9D8F;">fewer parameters</b> and can train faster than an LSTM.

Despite being simpler, GRUs can achieve
<b style="color:#E88955;">similar performance</b>
to LSTMs on many sequence-learning tasks.
</div>

```python
import pandas as pd
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

train_df = pd.read_csv("mitbih_train.csv")
test_df = pd.read_csv("mitbih_test.csv")

X_train = train_df.iloc[:, :-1].values.astype(np.float32)
y_train = train_df.iloc[:, -1].values.astype(np.int32)

X_test = test_df.iloc[:, :-1].values.astype(np.float32)
y_test = test_df.iloc[:, -1].values.astype(np.int32)

X_train = X_train.reshape((X_train.shape[0], X_train.shape[1], 1))
X_test = X_test.reshape((X_test.shape[0], X_test.shape[1], 1))

lstm_model = Sequential([
    LSTM(32, input_shape=(187, 1)),
    Dense(5, activation="softmax")
])

lstm_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

lstm_model.summary()
```

The MIT-BIH dataset contains ECG signals represented as sequential

patterns across time to classify each heartbeat into its corresponding class.

```python
import time

start_time = time.time()

lstm_history = lstm_model.fit(
    X_train,
    y_train,
    validation_split=0.2,
    epochs=10,
    batch_size=64,
    verbose=1
)

lstm_training_time = time.time() - start_time

lstm_loss, lstm_accuracy = lstm_model.evaluate(
    X_test,
    y_test,
    verbose=0
)

best_lstm_accuracy = max(lstm_history.history["val_accuracy"])
```

```python
from sklearn.model_selection import train_test_split
import time

X_train_rnn, X_val_rnn, y_train_rnn, y_val_rnn = train_test_split(
    X_train,
    y_train,
    test_size=0.2,
    random_state=42,
    stratify=y_train,
    shuffle=True
)

start_time = time.time()

lstm_history = lstm_model.fit(
    X_train_rnn,
    y_train_rnn,
    validation_data=(X_val_rnn, y_val_rnn),
    epochs=10,
    batch_size=64,
    verbose=1
)

lstm_training_time = time.time() - start_time

lstm_loss, lstm_accuracy = lstm_model.evaluate(
    X_test,
    y_test,
    verbose=0
)

best_lstm_accuracy = max(lstm_history.history["val_accuracy"])
```
