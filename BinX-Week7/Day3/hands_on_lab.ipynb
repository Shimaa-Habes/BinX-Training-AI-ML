```python
import numpy as np
import pandas as pd

train_df = pd.read_csv("mitbih_train.csv")
test_df = pd.read_csv("mitbih_test.csv")

X_train = train_df.iloc[:, :-1].values.astype(np.float32)
y_train = train_df.iloc[:, -1].values.astype(np.int32)

X_test = test_df.iloc[:, :-1].values.astype(np.float32)
y_test = test_df.iloc[:, -1].values.astype(np.int32)

X_train = X_train.reshape(X_train.shape[0], X_train.shape[1], 1)
X_test = X_test.reshape(X_test.shape[0], X_test.shape[1], 1)

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
import tensorflow as tf
import time
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

lstm_lab_model = Sequential([
    LSTM(32, input_shape=(X_train.shape[1], 1)),
    Dense(5, activation="softmax")
])

lstm_lab_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

start_time = time.time()

lstm_lab_history = lstm_lab_model.fit(
    X_train,
    y_train,
    validation_split=0.2,
    epochs=10,
    batch_size=64,
    verbose=1
)

lstm_lab_training_time = time.time() - start_time

lstm_lab_loss, lstm_lab_accuracy = lstm_lab_model.evaluate(
    X_test,
    y_test,
    verbose=0
)

lstm_lab_best_val_accuracy = max(
    lstm_lab_history.history["val_accuracy"]
)

print(f"Best Validation Accuracy: {lstm_lab_best_val_accuracy:.4f}")
print(f"Test Accuracy: {lstm_lab_accuracy:.4f}")
print(f"Training Time: {lstm_lab_training_time:.2f}s")
```

```python
from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.2,
    random_state=42,
    stratify=y_train
)

lstm_lab_model = Sequential([
    LSTM(32, input_shape=(X_train.shape[1], 1)),
    Dense(5, activation="softmax")
])

lstm_lab_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

start_time = time.time()

lstm_lab_history = lstm_lab_model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=10,
    batch_size=64,
    verbose=1
)

lstm_lab_training_time = time.time() - start_time

lstm_lab_loss, lstm_lab_accuracy = lstm_lab_model.evaluate(
    X_test,
    y_test,
    verbose=0
)

lstm_lab_best_val_accuracy = max(
    lstm_lab_history.history["val_accuracy"]
)

print(f"Best Validation Accuracy: {lstm_lab_best_val_accuracy:.4f}")
print(f"Test Accuracy: {lstm_lab_accuracy:.4f}")
print(f"Training Time: {lstm_lab_training_time:.2f}s")
```

```python
rnn_test_accuracy = 0.8276
lstm_test_accuracy = lstm_lab_accuracy

difference = lstm_test_accuracy - rnn_test_accuracy

print(f"Plain RNN Test Accuracy: {rnn_test_accuracy:.4f}")
print(f"LSTM Test Accuracy: {lstm_test_accuracy:.4f}")
print(f"Improvement: {difference:.4f}")
```

ECG signals are sequential, meaning that the order of measurements
contains important information about the shape and pattern of each
heartbeat.

The LSTM processes the signal step by step and maintains information
from previous time steps through its internal memory. This allows the
model to recognize patterns that depend on how the signal changes over
time rather than treating each measurement independently.

Because the LSTM is designed to preserve useful information across
longer sequences, it achieved a higher test accuracy than the plain RNN
on this dataset.
