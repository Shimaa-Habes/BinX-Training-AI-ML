Pooling
</div>

<div style="
    color:#596357;
    font-size:16px;
    line-height:1.9;
    font-weight:600;
    margin-bottom:22px;
">
After convolution, the feature map can still be relatively large.
Pooling reduces its spatial dimensions while keeping the most important
information detected by the convolution filters.
</div>


<div style="
    padding:22px 24px;

Max Pooling
</div>

<div style="
    color:#4D574B;
    font-size:16px;
    line-height:1.9;
    font-weight:600;
">
Max Pooling looks at a small region of the feature map, commonly a
<strong>2×2 window</strong>, and keeps only the largest value.
This preserves the strongest detected signal while reducing the size
of the feature map.
</div>

</div>


<div style="
    color:#8F2838;
    font-size:18px;
    font-weight:900;
    margin-bottom:12px;
">
Why Do We Use Pooling?
</div>

<div style="
    color:#596357;
    font-size:16px;
    line-height:2;
    font-weight:600;
">

<div style="margin-bottom:7px;">
<span style="color:#8F2838;font-weight:900;">01</span>
&nbsp; Reduces the spatial size of feature maps.
</div>

<div style="margin-bottom:7px;">
<span style="color:#8F2838;font-weight:900;">02</span>
&nbsp; Reduces computation and the number of values processed by later layers.
</div>

<div style="margin-bottom:7px;">
<span style="color:#8F2838;font-weight:900;">03</span>
&nbsp; Helps control overfitting by making the representation more compact.
</div>

<div>
<span style="color:#8F2838;font-weight:900;">04</span>
&nbsp; Makes the network more robust to small shifts in the input.
</div>

</div>


<div style="

Convolution + Pooling
</div>

<div style="
    color:#596357;
    font-size:16px;
    line-height:1.9;
    font-weight:600;
">
Convolution detects useful patterns such as edges, shapes, and textures,
while pooling progressively shrinks the resulting feature maps.
By alternating between convolution and pooling, the CNN gradually
transforms the original image into a smaller and more meaningful
representation.
</div>

</div>

</div>

<div style="

">
Before applying Max Pooling, we first need a small feature map to work with.
But this time, the values are not random. 👀
<br><br>
Each number is generated using a simple rule based on its
<strong>row and column position</strong>.
</div>

<div style="
    margin-top:20px;
    padding:20px 22px;

    background:#EEF4F0;
    border:1px solid #C9D9CF;

💡 <strong>Your Challenge:</strong>

Look at the matrix, try to discover the hidden pattern,

<div style="
    margin-top:18px;

">
🎯 Goal: Discover the rule → Generate the matrix → Apply 2×2 Max Pooling
</div>

```python
import numpy as np
from IPython.display import display, HTML

feature_map = np.fromfunction(
    lambda i, j: (i + 1) * (j + 2),
    (4, 4),
    dtype=int
)

pool_size = 2

pooled_height = feature_map.shape[0] // pool_size
pooled_width = feature_map.shape[1] // pool_size

pooled_feature_map = np.zeros(
    (pooled_height, pooled_width),
    dtype=int
)

for i in range(pooled_height):
    for j in range(pooled_width):
        window = feature_map[
            i * pool_size:(i + 1) * pool_size,
            j * pool_size:(j + 1) * pool_size
        ]

        pooled_feature_map[i, j] = np.max(window)

size_reduction = (
    (feature_map.size - pooled_feature_map.size)
    / feature_map.size
) * 100

original_text = np.array2string(
    feature_map,
    separator="   "
)

pooled_text = np.array2string(
    pooled_feature_map,
    separator="   "
)

display(HTML(f"""
<div style="
    width:100%;
    max-width:100%;
    box-sizing:border-box;
    margin:25px 0;
    padding:32px 38px;
    background:linear-gradient(135deg,#FFFDF8 0%,#F5EFE3 100%);
    border:1px solid #D7CCBC;
    border-radius:20px;
    box-shadow:0 7px 20px rgba(24,61,43,0.07);
    font-family:'Segoe UI',Arial,sans-serif;
    overflow:hidden;
">

<div style="
    color:#9A6A32;
    font-size:13px;
    font-weight:900;
    letter-spacing:2px;
    margin-bottom:20px;
">
MAX POOLING RESULT
</div>

<div style="
    display:flex;
    flex-wrap:wrap;
    gap:18px;
    width:100%;
    box-sizing:border-box;
">

<div style="
    flex:1 1 280px;
    min-width:0;
    box-sizing:border-box;
    padding:24px;
    background:#FFFFFF;
    border:1px solid #D8CCBC;
    border-radius:15px;
">

<div style="
    color:#183D2B;
    font-size:20px;
    font-weight:900;
    margin-bottom:15px;
">
Original Feature Map
</div>

<pre style="
    margin:0;
    color:#4D574B;
    font-family:'Consolas','Courier New',monospace;
    font-size:20px;
    line-height:1.7;
    font-weight:700;
    white-space:pre-wrap;
">
{original_text}
</pre>

<div style="
    margin-top:16px;
    color:#8F2838;
    font-size:15px;
    font-weight:800;
">
Shape: {feature_map.shape}
</div>

</div>

<div style="
    flex:1 1 280px;
    min-width:0;
    box-sizing:border-box;
    padding:24px;
    background:#EEF4F0;
    border:1px solid #C9D9CF;
    border-radius:15px;
">

<div style="
    color:#315C47;
    font-size:20px;
    font-weight:900;
    margin-bottom:15px;
">
Pooled Feature Map
</div>

<pre style="
    margin:0;
    color:#315C47;
    font-family:'Consolas','Courier New',monospace;
    font-size:20px;
    line-height:1.7;
    font-weight:700;
    white-space:pre-wrap;
">
{pooled_text}
</pre>

<div style="
    margin-top:16px;
    color:#8F2838;
    font-size:15px;
    font-weight:800;
">
Shape: {pooled_feature_map.shape}
</div>

</div>

</div>

<div style="
    margin-top:18px;
    padding:19px 23px;
    box-sizing:border-box;
    background:#F7ECE9;
    border:1px solid #E1C9C1;
    border-radius:15px;
">

<div style="
    color:#8F2838;
    font-size:16px;
    font-weight:900;
    margin-bottom:7px;
">
Pooling Window
</div>

<div style="
    color:#604A45;
    font-size:16px;
    line-height:1.8;
    font-weight:700;
">
{pool_size} × {pool_size}
&nbsp; → &nbsp;
The maximum value is selected from each region.
</div>

</div>

<div style="
    margin-top:18px;
    display:flex;
    flex-wrap:wrap;
    align-items:center;
    justify-content:space-between;
    gap:15px;
    padding:19px 23px;
    box-sizing:border-box;
    background:#EAF1EC;
    border:1px solid #C7D8CD;
    border-radius:15px;
">

<div style="
    color:#315C47;
    font-size:16px;
    font-weight:900;
">
Size Reduction
</div>

<div style="
    color:#183D2B;
    font-size:20px;
    font-weight:900;
">
{feature_map.size} values → {pooled_feature_map.size} values
&nbsp; ({size_reduction:.0f}% reduction)
</div>

</div>

</div>
"""))
```

<div style="

Why Do We Keep the Maximum Value? 🎯
</div>

<div style="

Max Pooling keeps the <strong>largest value</strong> from each 2×2 region

Instead of keeping every value, pooling keeps the most important signal
while reducing the size of the feature map.
</div>

```python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

num_classes = 3

model = Sequential([
    Conv2D(
        32,
        (3, 3),
        activation="relu",
        input_shape=(128, 128, 3)
    ),

    MaxPooling2D((2, 2)),

    Conv2D(
        64,
        (3, 3),
        activation="relu"
    ),

    MaxPooling2D((2, 2)),

    Flatten(),

    Dense(
        64,
        activation="relu"
    ),

    Dense(
        num_classes,
        activation="softmax"
    )
])
```

```python
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2

base = MobileNetV2(
    include_top=False,
    weights="imagenet",
    input_shape=(128, 128, 3)
)

base.trainable = False

print("Base model:", base.name)
print("Input shape:", base.input_shape)
print("Output shape:", base.output_shape)
print("Trainable:", base.trainable)
```

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense

num_classes = 3

model = Sequential([
    base,
    GlobalAveragePooling2D(),
    Dense(64, activation="relu"),
    Dense(num_classes, activation="softmax")
])

print("Model built successfully.")
print("Number of classes:", num_classes)
```

```python
import tensorflow as tf
import time

from tensorflow.keras import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense

IMG_SIZE = (128, 128)
BATCH_SIZE = 32
EPOCHS = 5

train_ds = tf.keras.utils.image_dataset_from_directory(
    "train",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=True,
    seed=42
)

test_ds = tf.keras.utils.image_dataset_from_directory(
    "test",
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    label_mode="binary",
    shuffle=False
)

class_names = train_ds.class_names

AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(AUTOTUNE)
test_ds = test_ds.prefetch(AUTOTUNE)

print("Classes:", class_names)
print("Number of classes:", len(class_names))
```

```python
scratch_model = Sequential([
    Conv2D(16, (3, 3), activation="relu", input_shape=(128, 128, 3)),
    MaxPooling2D((2, 2)),
    Conv2D(32, (3, 3), activation="relu"),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(32, activation="relu"),
    Dense(1, activation="sigmoid")
])

scratch_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

start_time = time.time()

scratch_history = scratch_model.fit(
    train_ds,
    validation_data=test_ds,
    epochs=EPOCHS
)

scratch_training_time = time.time() - start_time
```

```python
scratch_loss, scratch_accuracy = scratch_model.evaluate(
    test_ds,
    verbose=0
)

scratch_val_accuracy = max(
    scratch_history.history["val_accuracy"]
)

scratch_trainable_params = scratch_model.count_params()

print("Test Accuracy:", round(scratch_accuracy, 4))
print("Best Validation Accuracy:", round(scratch_val_accuracy, 4))
print("Training Time:", round(scratch_training_time, 2), "seconds")
print("Trainable Parameters:", scratch_trainable_params)
```

```python
from tensorflow.keras.layers import RandomFlip, RandomRotation, RandomZoom
```

```python
augment = Sequential([
    RandomFlip("horizontal"),
    RandomRotation(0.1),
    RandomZoom(0.1)
])

augmentation_model = Sequential([
    augment,
    Conv2D(16, (3, 3), activation="relu", input_shape=(128, 128, 3)),
    MaxPooling2D((2, 2)),
    Conv2D(32, (3, 3), activation="relu"),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(32, activation="relu"),
    Dense(1, activation="sigmoid")
])

augmentation_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

start_time = time.time()

augmentation_history = augmentation_model.fit(
    train_ds,
    validation_data=test_ds,
    epochs=EPOCHS
)

augmentation_training_time = time.time() - start_time
```

```python
augmentation_loss, augmentation_accuracy = augmentation_model.evaluate(
    test_ds,
    verbose=0
)

augmentation_val_accuracy = max(
    augmentation_history.history["val_accuracy"]
)

augmentation_trainable_params = augmentation_model.count_params()

print("Test Accuracy:", round(augmentation_accuracy, 4))
print("Best Validation Accuracy:", round(augmentation_val_accuracy, 4))
print("Training Time:", round(augmentation_training_time, 2), "seconds")
print("Trainable Parameters:", augmentation_trainable_params)
```

```python
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

base = MobileNetV2(
    weights="imagenet",
    include_top=False,
    input_shape=(128, 128, 3)
)

base.trainable = False
```

```python
from tensorflow.keras.layers import GlobalAveragePooling2D
```

```python
transfer_model = Sequential([
    base,
    GlobalAveragePooling2D(),
    Dense(32, activation="relu"),
    Dense(1, activation="sigmoid")
])

transfer_model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

start_time = time.time()

transfer_train_ds = train_ds.map(
    lambda x, y: (preprocess_input(x), y),
    num_parallel_calls=tf.data.AUTOTUNE
).prefetch(tf.data.AUTOTUNE)

transfer_test_ds = test_ds.map(
    lambda x, y: (preprocess_input(x), y),
    num_parallel_calls=tf.data.AUTOTUNE
).prefetch(tf.data.AUTOTUNE)

transfer_history = transfer_model.fit(
    transfer_train_ds,
    validation_data=transfer_test_ds,
    epochs=EPOCHS
)

transfer_training_time = time.time() - start_time
```

```python
import numpy as np

transfer_loss, transfer_accuracy = transfer_model.evaluate(
    transfer_test_ds,
    verbose=0
)

transfer_val_accuracy = max(
    transfer_history.history["val_accuracy"]
)

transfer_trainable_params = sum(
    np.prod(v.shape) for v in transfer_model.trainable_weights
)

print("Test Accuracy:", round(transfer_accuracy, 4))
print("Best Validation Accuracy:", round(transfer_val_accuracy, 4))
print("Training Time:", round(transfer_training_time, 2), "seconds")
print("Trainable Parameters:", transfer_trainable_params)
```
