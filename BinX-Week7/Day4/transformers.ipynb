<div style="font-size:22px;font-weight:700;color:#4338CA;">⚠️ 4.1 The Limitation of RNNs</div>
<div style="font-size:13px;color:#6B7280;margin-top:6px;">Understanding why sequential processing became a limitation</div>
</div>

<div style="width:100%;box-sizing:border-box;padding:20px 22px;border-radius:15px;background:#FFFFFF;border:1px solid #DDD6FE;box-shadow:0 4px 12px rgba(79,70,229,0.05);font-family:Arial,sans-serif;">
<div style="font-size:19px;font-weight:700;color:#6D28D9;margin-bottom:10px;">🔄 Step-by-Step Processing</div>
<div style="font-size:14px;line-height:1.75;color:#4B5563;">
RNNs process sequence data one step at a time. Each element must be processed before the model can move to the next one. This sequential nature makes training slower because the computations cannot be fully parallelized.
<br><br>
For example, when processing a sequence of words, the RNN reads the first word, then the second, then the third, and continues in the same order. Later steps have to wait for earlier steps to finish.
</div>
</div>

<div style="width:100%;box-sizing:border-box;padding:20px 22px;border-radius:15px;background:#FFFFFF;border:1px solid #C7D2FE;box-shadow:0 4px 12px rgba(79,70,229,0.05);font-family:Arial,sans-serif;">
<div style="font-size:19px;font-weight:700;color:#4F46E5;margin-bottom:10px;">🔗 Long-Range Dependencies</div>
<div style="font-size:14px;line-height:1.75;color:#4B5563;">
RNNs can struggle when important information appears far apart in a sequence. Although LSTMs were designed to handle longer dependencies better than basic RNNs, they still process information sequentially and can have difficulty preserving information across very long sequences.
</div>
</div>

<div style="width:100%;box-sizing:border-box;padding:20px 22px;border-radius:15px;background:linear-gradient(135deg,#F0FDFF,#F5F3FF);border:1px solid #A5F3FC;box-shadow:0 4px 12px rgba(6,182,212,0.06);font-family:Arial,sans-serif;">
<div style="font-size:19px;font-weight:700;color:#0E7490;margin-bottom:10px;">🚀 The Shift to Transformers</div>
<div style="font-size:14px;line-height:1.75;color:#4B5563;">
In 2017, the Transformer architecture was introduced and removed recurrence entirely. Instead of processing a sequence step by step, Transformers use Attention to allow different positions to interact directly with each other.
<br><br>
This architecture became the foundation of many modern AI systems. Models such as GPT and BERT are Transformer-based architectures and use Attention to understand relationships between different parts of the input.
</div>
</div>


<div style="width:100%;box-sizing:border-box;padding:17px 22px;border-radius:15px;background:linear-gradient(135deg,#FFF7ED,#FFF1F2);border:1px solid #FED7AA;box-shadow:0 4px 12px rgba(249,115,22,0.06);font-family:Arial,sans-serif;">
<div style="font-size:22px;font-weight:700;color:#C2410C;">🧠 4.2 The Attention Mechanism</div>
<div style="font-size:13px;color:#6B7280;margin-top:6px;">Understanding how Transformers focus on the most relevant information</div>
</div>

<div style="width:100%;box-sizing:border-box;padding:20px 22px;border-radius:15px;background:#FFFFFF;border:1px solid #FED7AA;box-shadow:0 4px 12px rgba(249,115,22,0.05);font-family:Arial,sans-serif;">
<div style="font-size:19px;font-weight:700;color:#EA580C;margin-bottom:10px;">👀 What is Attention?</div>
<div style="font-size:14px;line-height:1.75;color:#4B5563;">
Attention allows each element in a sequence to look at the other elements and determine which ones are important for understanding the current element.
<br><br>
Instead of relying only on the previous information, the model can consider the entire sequence and focus on the parts that are most relevant.
</div>
</div>

<div style="width:100%;box-sizing:border-box;padding:20px 22px;border-radius:15px;background:linear-gradient(135deg,#FFFBEB,#FFF7ED);border:1px solid #FCD34D;box-shadow:0 4px 12px rgba(245,158,11,0.05);font-family:Arial,sans-serif;">
<div style="font-size:19px;font-weight:700;color:#B45309;margin-bottom:10px;">🔎 Self-Attention</div>
<div style="font-size:14px;line-height:1.75;color:#4B5563;">
Self-Attention means that each element in the sequence looks at all the other elements, including itself, and assigns different importance weights to them.
<br><br>
The model learns which elements are more relevant to the current element and gives them higher attention. This helps the model understand relationships and context within the whole sequence.
</div>
</div>

<div style="width:100%;box-sizing:border-box;padding:20px 22px;border-radius:15px;background:#FFFFFF;border:1px solid #FDBA74;box-shadow:0 4px 12px rgba(249,115,22,0.05);font-family:Arial,sans-serif;">
<div style="font-size:19px;font-weight:700;color:#C2410C;margin-bottom:10px;">⚡ Parallel Processing</div>
<div style="font-size:14px;line-height:1.75;color:#4B5563;">
Unlike RNNs, Transformers do not need to process the sequence one element at a time. Attention allows all positions in the sequence to be processed at the same time.
<br><br>
This parallel processing makes Transformer training much faster and allows the model to work efficiently with large amounts of data.
</div>
</div>

<div style="width:100%;box-sizing:border-box;padding:20px 22px;border-radius:15px;background:linear-gradient(135deg,#F0FDFA,#ECFEFF);border:1px solid #99F6E4;box-shadow:0 4px 12px rgba(20,184,166,0.05);font-family:Arial,sans-serif;">
<div style="font-size:19px;font-weight:700;color:#0F766E;margin-bottom:10px;">🔗 Long-Range Context</div>
<div style="font-size:14px;line-height:1.75;color:#4B5563;">
Attention gives each element direct access to other elements in the sequence, even when they are far apart.
<br><br>
Because the information does not have to pass through many sequential steps, Transformers can capture long-range relationships more effectively and reduce the vanishing-gradient problem associated with long sequential paths.
</div>
</div>

<div style="width:100%;box-sizing:border-box;padding:20px 22px;border-radius:15px;background:#FFFFFF;border:1px solid #C4B5FD;box-shadow:0 4px 12px rgba(124,58,237,0.05);font-family:Arial,sans-serif;">
<div style="font-size:19px;font-weight:700;color:#6D28D9;margin-bottom:10px;">💡 Simple Example</div>
<div style="font-size:14px;line-height:1.75;color:#4B5563;">
Consider the sentence:
<br><br>
<div style="padding:12px 16px;border-radius:10px;background:#F5F3FF;border-left:4px solid #8B5CF6;font-size:15px;color:#4338CA;font-weight:600;">"The cat was tired because it had been playing all day."</div>
<br>
When the model processes the word <b>"it"</b>, Attention allows it to look at the other words in the sentence and determine which one is most relevant. In this case, it can learn that <b>"it"</b> refers to <b>"the cat"</b>.
<br><br>
This is the main idea of Attention: the model can directly look at other parts of the sequence to understand the meaning of the current element.
</div>
</div>

<div style="width:100%;box-sizing:border-box;padding:17px 22px;border-radius:15px;background:linear-gradient(135deg,#F0FDF4,#EFF6FF);border:1px solid #BBF7D0;box-shadow:0 4px 12px rgba(34,197,94,0.06);font-family:Arial,sans-serif;">
<div style="font-size:22px;font-weight:700;color:#166534;">🏗️ 4.3 The Transformer Architecture</div>
<div style="font-size:13px;color:#6B7280;margin-top:6px;">Understanding the main building blocks of a Transformer</div>
</div>

<div style="width:100%;box-sizing:border-box;padding:20px 22px;border-radius:15px;background:#FFFFFF;border:1px solid #BFDBFE;box-shadow:0 4px 12px rgba(59,130,246,0.05);font-family:Arial,sans-serif;">
<div style="font-size:19px;font-weight:700;color:#2563EB;margin-bottom:10px;">🧩 Stacked Layers</div>
<div style="font-size:14px;line-height:1.75;color:#4B5563;">
A Transformer is built from multiple layers stacked on top of each other. Each layer contains an Attention mechanism followed by a Feed-Forward network.
<br><br>
The <b>Attention layer</b> allows the model to understand relationships between different elements in the sequence, while the <b>Feed-Forward layer</b> further processes and transforms the information learned from Attention.
<br><br>
By stacking multiple layers, the model can gradually learn more complex patterns and relationships from the input.
</div>
</div>

<div style="width:100%;box-sizing:border-box;padding:20px 22px;border-radius:15px;background:linear-gradient(135deg,#FFF7ED,#FFFBEB);border:1px solid #FED7AA;box-shadow:0 4px 12px rgba(249,115,22,0.05);font-family:Arial,sans-serif;">
<div style="font-size:19px;font-weight:700;color:#C2410C;margin-bottom:10px;">📍 Positional Encoding</div>
<div style="font-size:14px;line-height:1.75;color:#4B5563;">
Transformers process all positions in a sequence in parallel, so they do not naturally know the order of the elements.
<br><br>
<b>Positional Encoding</b> adds information about the position of each element. This allows the Transformer to distinguish between different orders and understand that the first element comes before the second, the second before the third, and so on.
<br><br>
For example, the sentences <b>"The dog chased the cat"</b> and <b>"The cat chased the dog"</b> contain the same words but have different meanings because the order is different. Positional information helps the model recognize this difference.
</div>
</div>

<div style="width:100%;box-sizing:border-box;padding:20px 22px;border-radius:15px;background:linear-gradient(135deg,#F5F3FF,#F0FDFA);border:1px solid #C4B5FD;box-shadow:0 4px 12px rgba(124,58,237,0.05);font-family:Arial,sans-serif;">
<div style="font-size:19px;font-weight:700;color:#6D28D9;margin-bottom:10px;">🎯 What We Need to Understand</div>
<div style="font-size:14px;line-height:1.75;color:#4B5563;">
For this stage, understanding the Transformer architecture conceptually is enough. We do not need to build every component from scratch.
<br><br>
Our main goal is to understand how <b>Attention</b>, <b>Feed-Forward layers</b>, and <b>Positional Encoding</b> work together. In practice, we will use <b>pre-trained Transformer models</b> instead of implementing the full architecture manually.
</div>
</div>

<div style="font-size:22px;font-weight:700;color:#4338CA;">🤗 4.4 Pre-trained Transformers & Hugging Face</div>
<div style="font-size:13px;color:#6B7280;margin-top:6px;">Using powerful pre-trained Transformer models with Hugging Face</div>
</div>

<div style="width:100%;box-sizing:border-box;padding:20px 22px;border-radius:15px;background:#FFFFFF;border:1px solid #C7D2FE;box-shadow:0 4px 12px rgba(79,70,229,0.05);font-family:Arial,sans-serif;">
<div style="font-size:19px;font-weight:700;color:#4F46E5;margin-bottom:10px;">🔄 From Transfer Learning to Transformers</div>
<div style="font-size:14px;line-height:1.75;color:#4B5563;">
The idea is similar to <b>Transfer Learning</b> from Day 2. Instead of training a deep learning model completely from scratch, we start with a model that has already learned useful patterns from a large amount of data.
<br><br>
In Day 2, we applied this idea to <b>CNNs for computer vision</b>. Today, we apply the same principle to <b>Transformers for NLP</b>. We reuse a pre-trained Transformer and use it for a specific text task with minimal additional work.
</div>
</div>

<div style="width:100%;box-sizing:border-box;padding:20px 22px;border-radius:15px;background:linear-gradient(135deg,#F0FDFA,#ECFEFF);border:1px solid #99F6E4;box-shadow:0 4px 12px rgba(20,184,166,0.05);font-family:Arial,sans-serif;">
<div style="font-size:19px;font-weight:700;color:#0F766E;margin-bottom:10px;">🧠 Popular Pre-trained Transformers</div>
<div style="font-size:14px;line-height:1.75;color:#4B5563;">
<b>BERT</b> is a Transformer model designed to understand the context of words by considering information from both directions of a sentence.
<br><br>
<b>DistilBERT</b> is a smaller and faster version of BERT that keeps much of its language understanding while using fewer resources.
<br><br>
<b>GPT-2</b> is a generative Transformer model designed to predict and generate text based on the context it receives.
<br><br>
These models were pre-trained on very large collections of text, allowing them to learn general language patterns before being adapted to specific tasks.
</div>
