```python
import numpy as np
import matplotlib.pyplot as plt

image = np.zeros((100, 100))

image[20:80, 20:30] = 1.0
image[20:30, 20:80] = 1.0
image[70:80, 20:80] = 1.0

edge_filter = np.array([
    [-1, 0, 1],
    [-1, 0, 1],
    [-1, 0, 1]
])

kernel_size = 3

output_height = image.shape[0] - kernel_size + 1
output_width = image.shape[1] - kernel_size + 1

feature_map = np.zeros((output_height, output_width))

for i in range(output_height):
    for j in range(output_width):

        image_patch = image[
            i:i + kernel_size,
            j:j + kernel_size
        ]

        feature_map[i, j] = np.sum(
            image_patch * edge_filter
        )

# Visualization
fig, axes = plt.subplots(
    1,
    2,
    figsize=(13, 5),
    constrained_layout=True
)

axes[0].imshow(
    image,
    cmap="gray",
    interpolation="nearest"
)

axes[0].set_title(
    "Original Image",
    fontsize=17,
    fontweight="bold",
    pad=14
)

axes[0].set_xticks([])
axes[0].set_yticks([])

im = axes[1].imshow(
    feature_map,
    cmap="gray",
    interpolation="nearest"
)

axes[1].set_title(
    "Feature Map",
    fontsize=17,
    fontweight="bold",
    pad=14
)

axes[1].set_xticks([])
axes[1].set_yticks([])

cbar = fig.colorbar(
    im,
    ax=axes[1],
    fraction=0.046,
    pad=0.04
)

cbar.set_label(
    "Filter Response",
    fontsize=11,
    fontweight="600"
)

fig.suptitle(
    "3×3 Edge Detection Using Convolution",
    fontsize=20,
    fontweight="bold"
)

plt.show()
```

Parameter Sharing
</div>

<div style="
    color:#4D574B;
    font-size:16px;
    line-height:1.9;
    font-weight:600;
">
CNNs reuse the same filter weights across different locations of an image.
This is called <span style="color:#9F2638;font-weight:900;">parameter sharing</span>.
It allows a CNN to detect the same pattern wherever it appears while using
far fewer trainable parameters than a fully connected layer.
</div>
