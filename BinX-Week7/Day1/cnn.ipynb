">
1.2 Why Dense Networks Fail on Images
</div>

<div style="
    color:#596357;
    font-size:17px;
    line-height:1.7;
    font-weight:600;
    margin-top:10px;
">
Understanding why fully connected networks are inefficient for image data.
</div>

</div>

<div style="
    width:100%;
    box-sizing:border-box;
    padding:34px 40px;
    margin:22px 0;
    background:#FFFDF8;
    border:1px solid #D5C7B5;
    border-radius:24px;
    box-shadow:0 8px 22px rgba(24,61,43,0.07);
    font-family:'Segoe UI',Arial,sans-serif;
">

<div style="
    color:#183D2B;
    font-size:28px;
    font-weight:900;
    margin-bottom:18px;
">
The Problem with Dense Networks
</div>

<div style="
    color:#4D574B;
    font-size:17px;
    line-height:1.9;
    font-weight:600;
">
A <span style="color:#9F2638;font-weight:900;">200 × 200 RGB image</span>
contains:
</div>

<div style="
    margin:20px 0;
    padding:22px;
    box-sizing:border-box;
    width:100%;
    background:#F1F4EC;
    border:1px solid #CBD5C4;
    border-radius:16px;
    text-align:center;
">

<div style="
    color:#183D2B;
    font-size:29px;
    font-weight:900;
">
200 × 200 × 3 = 120,000
</div>

<div style="
    color:#68735F;
    font-size:14px;
    font-weight:700;
    margin-top:7px;
">
input values
</div>

</div>

<div style="
    color:#4D574B;
    font-size:17px;
    line-height:1.9;
    font-weight:600;
">
If these 120,000 values are connected directly to a dense layer with
<span style="color:#9F2638;font-weight:900;">128 neurons</span>,
the first layer alone requires more than
<span style="color:#183D2B;font-weight:900;">15 million trainable weights</span>.
</div>

<div style="
    margin-top:27px;
    padding-top:24px;
    border-top:2px solid #E5DED2;
">

<div style="
    color:#9F2638;
    font-size:13px;
    font-weight:900;
    letter-spacing:1.8px;
    margin-bottom:16px;
">
TWO KEY PROBLEMS
</div>

<div style="
    width:100%;
    box-sizing:border-box;
    background:#F8EDE8;
    border:1px solid #E2C8BC;
    border-radius:16px;
    padding:21px 24px;
    margin-bottom:14px;
">

<div style="
    color:#70202D;
    font-size:19px;
    font-weight:900;
    margin-bottom:7px;
">
01 · Spatial Structure
</div>

<div style="
    color:#604A45;
    font-size:15px;
    line-height:1.8;
    font-weight:600;
">
Nearby pixels are related. A pixel's meaning depends partly on the pixels
around it, but a dense network treats every input connection independently
and does not explicitly preserve this spatial relationship.
</div>

</div>

<div style="
    width:100%;
    box-sizing:border-box;
    background:#EEF2E9;
    border:1px solid #CBD5C4;
    border-radius:16px;
    padding:21px 24px;
">

<div style="
    color:#183D2B;
    font-size:19px;
    font-weight:900;
    margin-bottom:7px;
">
02 · Translation Invariance
</div>

<div style="
    color:#4D574B;
    font-size:15px;
    line-height:1.8;
    font-weight:600;
">
The same pattern can appear in different locations of an image. Dense networks
do not naturally reuse the same learned pattern across different positions.
CNNs solve this by applying the same filters throughout the image.
</div>

</div>

</div>

```python
import numpy as np

# Create a sample 100x100 grayscale image
image = np.random.rand(100, 100)

# Define a vertical edge detection kernel
kernel = np.array([
    [-1, 0, 1],
    [-1, 0, 1],
    [-1, 0, 1]
])

# Apply convolution (valid padding, stride = 1)
feature_map = np.zeros((98, 98))

for i in range(98):
    for j in range(98):
        patch = image[i:i+3, j:j+3]
        feature_map[i, j] = np.sum(patch * kernel)

print("Feature Map Shape:", feature_map.shape)
```
