```python
from fastapi import FastAPI
import joblib

app = FastAPI()

model = joblib.load("../Day1/model.joblib")
scaler = joblib.load("../Day1/scaler.joblib")

print("FastAPI application created.")
print("Model and scaler loaded successfully.")
```

```python
from pydantic import BaseModel

class InputData(BaseModel):
    mean_radius: float
    mean_texture: float
    mean_perimeter: float
    mean_area: float
    mean_smoothness: float
    mean_compactness: float
    mean_concavity: float
    mean_concave_points: float
    mean_symmetry: float
    mean_fractal_dimension: float
    radius_error: float
    texture_error: float
    perimeter_error: float
    area_error: float
    smoothness_error: float
    compactness_error: float
    concavity_error: float
    concave_points_error: float
    symmetry_error: float
    fractal_dimension_error: float
    worst_radius: float
    worst_texture: float
    worst_perimeter: float
    worst_area: float
    worst_smoothness: float
    worst_compactness: float
    worst_concavity: float
    worst_concave_points: float
    worst_symmetry: float
    worst_fractal_dimension: float

print("Input schema created successfully.")
```

```python
@app.post("/predict")
def predict(data: InputData):
    input_data = [[
        data.mean_radius,
        data.mean_texture,
        data.mean_perimeter,
        data.mean_area,
        data.mean_smoothness,
        data.mean_compactness,
        data.mean_concavity,
        data.mean_concave_points,
        data.mean_symmetry,
        data.mean_fractal_dimension,
        data.radius_error,
        data.texture_error,
        data.perimeter_error,
        data.area_error,
        data.smoothness_error,
        data.compactness_error,
        data.concavity_error,
        data.concave_points_error,
        data.symmetry_error,
        data.fractal_dimension_error,
        data.worst_radius,
        data.worst_texture,
        data.worst_perimeter,
        data.worst_area,
        data.worst_smoothness,
        data.worst_compactness,
        data.worst_concavity,
        data.worst_concave_points,
        data.worst_symmetry,
        data.worst_fractal_dimension
    ]]

    X = scaler.transform(input_data)
    pred = model.predict(X)

    return {"prediction": int(pred[0])}
```

    Define the Pydantic Input Schema
</h2>
<p style="

    color:#5F625A;

">
    I defined a <strong>Pydantic BaseModel</strong> called
    <strong>InputData</strong> to validate the data received by the
    <strong>/predict</strong> endpoint.
</p>
<p style="
    margin:14px 0 0 0;

">
    The schema contains all <strong>30 features</strong> from the
    Breast Cancer dataset, and each feature is defined as a
    <strong>float</strong>. This ensures that the API receives the
    correct feature names and data types before making a prediction.
</p>

```python
from pydantic import BaseModel

class InputData(BaseModel):
    mean_radius: float
    mean_texture: float
    mean_perimeter: float
    mean_area: float
    mean_smoothness: float
    mean_compactness: float
    mean_concavity: float
    mean_concave_points: float
    mean_symmetry: float
    mean_fractal_dimension: float
    radius_error: float
    texture_error: float
    perimeter_error: float
    area_error: float
    smoothness_error: float
    compactness_error: float
    concavity_error: float
    concave_points_error: float
    symmetry_error: float
    fractal_dimension_error: float
    worst_radius: float
    worst_texture: float
    worst_perimeter: float
    worst_area: float
    worst_smoothness: float
    worst_compactness: float
    worst_concavity: float
    worst_concave_points: float
    worst_symmetry: float
    worst_fractal_dimension: float

print("Input schema created successfully.")
```

    Implement the /predict Endpoint
</h2>
<p style="
    margin:0;
    color:#5F625A;
    font-size:15px;
    line-height:1.9;
">
    I implemented the <strong>POST /predict</strong> endpoint to receive
    validated input from the Pydantic schema and arrange the 30 features
    in the same order used during model training.
</p>
<p style="
    margin:14px 0 0 0;
    color:#5F625A;
    font-size:15px;
    line-height:1.9;
">
    The input is first transformed using the saved
    <strong>scaler.joblib</strong>. The scaled data is then passed to the
    saved <strong>model.joblib</strong> to generate a prediction.
    Finally, the prediction is returned as a JSON response.
</p>
<div style="
    margin-top:18px;
    padding:15px 20px;
    background:#E9EDE6;
    border:1px solid #D4DCCF;
    border-radius:12px;
    color:#4F6048;
    font-size:14px;
    font-weight:700;
">
    Input → Pydantic Validation → Saved Scaler → Saved Model → JSON Response
</div>

</div>


```python
@app.post("/predict")
def predict(data: InputData):
    input_data = [[
        data.mean_radius,
        data.mean_texture,
        data.mean_perimeter,
        data.mean_area,
        data.mean_smoothness,
        data.mean_compactness,
        data.mean_concavity,
        data.mean_concave_points,
        data.mean_symmetry,
        data.mean_fractal_dimension,
        data.radius_error,
        data.texture_error,
        data.perimeter_error,
        data.area_error,
        data.smoothness_error,
        data.compactness_error,
        data.concavity_error,
        data.concave_points_error,
        data.symmetry_error,
        data.fractal_dimension_error,
        data.worst_radius,
        data.worst_texture,
        data.worst_perimeter,
        data.worst_area,
        data.worst_smoothness,
        data.worst_compactness,
        data.worst_concavity,
        data.worst_concave_points,
        data.worst_symmetry,
        data.worst_fractal_dimension
    ]]

    X = scaler.transform(input_data)
    pred = model.predict(X)

    return {"prediction": int(pred[0])}
```
