<h2 style="color:#1F7A66; font-size:24px; margin:0; font-weight:bold;">
📦 4.2 The requirements.txt Is Critical
</h2>
<p style="color:#3D6B5C; font-size:16px; line-height:1.7; margin-top:10px; font-weight:bold;">
The hosting platform builds a fresh environment from my requirements.txt
— if it's missing a library or has a wrong version, the deployment
fails. This is where Day 1's pinned requirements pay off.
</p>
</div>

```python
requirements_content = """streamlit==1.36.0
scikit-learn==1.5.0
joblib==1.4.2
numpy==1.26.4
pandas==2.2.2
"""

with open("requirements.txt", "w") as f:
    f.write(requirements_content)

print("requirements.txt created:")
print(requirements_content)
```

<div style="background-color:#F0FAF7; border-left:5px solid #5FB49C; padding:18px 22px; border-radius:14px; margin-top:10px; width:100%; box-sizing:border-box;">
<p style="color:#3D6B5C; font-size:16px; line-height:1.7; margin:0; font-weight:bold;">
📌 A missing library, or a version different from what was used locally,
can silently break the build — or worse, install successfully but
produce different results than the local app. Pinning exact versions is
what makes a deployment reproducible.
</p>
</div>

```python
import requests

LIVE_URL = "https://your-username-your-space.hf.space"

test_inputs = [
    {"feature_1": 5.1, "feature_2": 3.5},
    {"feature_1": 6.7, "feature_2": 3.0},
]

for i, sample in enumerate(test_inputs, start=1):
    try:
        response = requests.post(f"{LIVE_URL}/predict", json=sample, timeout=10)
        print(f"Test {i} - Input: {sample}")
        print(f"Live prediction: {response.json()}")
    except Exception as e:
        print(f"Test {i} failed:", e)
```
