<h1 style="color: #5F6655; margin-bottom: 18px;">🚀 3.1 Why Streamlit</h1>

<p style="font-size: 16px; line-height: 1.8; color: #4F504A;">
FastAPI and Streamlit serve different purposes when working with a machine learning model.
FastAPI provides an API that allows <b>programs</b> to communicate with the model,
while Streamlit provides an interactive interface designed for <b>people</b>.
</p>

<h3 style="color: #6F7764;">🔌 FastAPI — For Programs</h3>

<p style="font-size: 15px; line-height: 1.8; color: #4F504A;">
FastAPI allows another application, such as a website or mobile application,
to send data to the machine learning model through an HTTP request.
The model processes the data and returns the prediction as a JSON response.
</p>

<h3 style="color: #6F7764;">🖥️ Streamlit — For People</h3>

<p style="font-size: 15px; line-height: 1.8; color: #4F504A;">
Streamlit creates a simple and interactive web interface directly from Python.
Users can enter values using UI widgets, click a button, and receive the model's
prediction without writing code or sending API requests manually.
</p>

<h3 style="color: #6F7764;">⚡ Why Streamlit?</h3>

<ul style="font-size: 15px; line-height: 1.9; color: #4F504A;">
<li>Builds interactive web interfaces directly from Python.</li>
<li>Does not require HTML, CSS, or JavaScript.</li>
<li>Provides simple widgets for collecting user input.</li>
<li>Makes machine learning models easier for non-technical users to interact with.</li>
<li>Provides a practical interface for live project demonstrations.</li>
</ul>

<h1 style="color: #5F6655; margin-bottom: 18px;">🧩 3.3 Common Streamlit Widgets</h1>

<p style="color: #5B5C56; font-size: 16px; line-height: 1.7;">
Streamlit provides different widgets for collecting user input and displaying application results.
The appropriate widget depends on the type of data used by the machine learning project.
</p>

<h3 style="color: #68735D;">🔢 Numeric Inputs</h3>

<p style="color: #5B5C56; font-size: 15px; line-height: 1.7;">
<strong>st.number_input</strong> allows users to enter numeric values directly.
It is suitable for the Breast Cancer project because all 30 model features are numeric.
</p>

<p style="color: #5B5C56; font-size: 15px; line-height: 1.7;">
<strong>st.slider</strong> is another option for numeric values.
It allows users to select a value by moving a slider within a defined range.
</p>

<h3 style="color: #68735D;">🏷️ Categorical Inputs</h3>

<p style="color: #5B5C56; font-size: 15px; line-height: 1.7;">
<strong>st.selectbox</strong> displays a list of options where the user selects one value.
<strong>st.radio</strong> displays options as radio buttons and is also useful for categorical data.
</p>

<h3 style="color: #68735D;">📝 Text Inputs</h3>

<p style="color: #5B5C56; font-size: 15px; line-height: 1.7;">
<strong>st.text_input</strong> is used for short text values, while
<strong>st.text_area</strong> is useful for longer text.
These widgets are useful for text-based machine learning or application projects.
</p>

<h3 style="color: #68735D;">📁 File Uploads</h3>

<p style="color: #5B5C56; font-size: 15px; line-height: 1.7;">
<strong>st.file_uploader</strong> allows users to upload files such as images or CSV datasets.
It can be useful in projects where the model receives files instead of individual values.
</p>

<h3 style="color: #68735D;">📊 Displaying Results</h3>

<p style="color: #5B5C56; font-size: 15px; line-height: 1.7;">
<strong>st.write</strong> displays general text or data.
<strong>st.success</strong> displays a successful result, while
<strong>st.error</strong> displays an error or negative result.
</p>
